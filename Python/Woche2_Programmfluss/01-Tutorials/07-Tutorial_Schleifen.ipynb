# `While`-Schleifen

Eine `while`-Schleife ist in der Lage, Code auszuführen *solange* eine Bedingung erfüllt (`True`) ist. 


## Syntax:
```python
while Bedingung:
    Code der ausgeführt werden soll, solange Bedingung erfüllt ist
```

> <span style="color:red"> **_WICHTIG:_** </span> \
Wie beim `If-Else` Konstrukt, müssen wir hier daran denken, den Code einzurücken.

### Beispiele:

Die folgende Zelle zeigt eine typische Anwendung einer `while`-Scheife. 

Die Variable `x` wird in jedem Schleifendurchgang um 1 erhöht. 

Zusätzlich soll der aktuelle Wert der Variablen ausgegeben werden. 

Sobald `x` den Wert 8 erreicht, ist die Bedingung nicht mehr erfüllt, was zum Verlassen der Schleife nach dem aktuellen Durchlauf führt.

````python
x = 0

while x < 8:
    x = x + 1  # x wird um 1 erhöht
    print("Der aktuelle Wert von x ist", x)
````

Ein anderes typisches Beispiel ist bei einer Nutzereingabe durch die Funktion `input()`. 

Im folgenden Beispiel wird so lange um eine Eingabe gebeten, bis diese korrekt ist.

````python
user_input = ""

while not (user_input == "A" or user_input == "a"):
    user_input = input("Was ist der erste Buchstabe des Alphabets?\n")

print("Korrekt!")  # Diese Zeile wird erst erreicht, wenn die Schleife verlassen wurde.
````

Wir sehen hier auch eine etwas komplexere Bedingung in der Schleife. 

Der Teil in der Klammer ist `True`, wenn die Nutzereingabe `A` oder `a` ist. 

Dies wird durch `not` verneint, d.h. die Schleife wird solange durchlaufen, wie die Eingabe **nicht** `A` oder `a` lautet.

## Flags und Schleifen

Eine weitere Art der Anwendung von `while`-Schleifen ist in Kombination mit **Flags**. 

**Flags** sind boolsche Variablen, die nur den Zweck haben, zu speichern, ob die Schleife beendet werden soll oder nicht. 

Dadurch können wir innerhalb der Schleife mehrere Bedingungen definieren, die die Schleife beenden.

### Beispiel:

Das folgende Beispiel zeigt eine praktische Anwendung, die Schleife kann nun durch unterschiedliche Bedingungen beendet werden.

````python
aktiv = True

while aktiv:
    input_a = input(
        "Gibt mir die erste Zahl ein, die ich addieren soll. Zum Beenden gibt 'quit' ein.\n"
    )
    input_b = input(
        "Gibt mir die zweite Zahl ein, die ich addieren soll. Zum Beenden gibt 'quit' ein.\n"
    )

    if input_a == "quit" or input_b == "quit":
        aktiv = False
    else:
        print(int(input_a) + int(input_b))
````

# `For`-Schleifen

`For`-Schleifen werden grundsätzlich verwendet um den gleichen Code mehrmals auszuführen. 

In Python werden `for`-Schleifen über eine Sequenz von Objekten ausgeführt und für jedes Element wird der Code *innerhalb* der `for`-Schleife ausgeführt. 

Im Gegensatz zu `while`-Schleifen wird daher die Schleife nicht beendet, wenn eine Bedingung nicht mehr erfüllt ist, sondern wenn keine Elemente mehr vorhanden sind.

## Syntax:
```python
for element in sequence:
    Code der ausgeführt werden soll, solange es Elemente in sequence gibt
```
Hierbei ist 
- `sequence` Eine Sequenz von Objekten (meistens Listen).
- `element`: Ein frei wählbarer Name. In jedem Schleifendurchgang enthält `element` das aktuelle Element aus `sequence`.

Der auszuführende Code muss wie gewohnt **eingerückt** werden.

### Beispiel:

Das folgende Beispiel zeigt eine `for`-Schleife angewendet auf eine Liste. 

Bei jedem Schleifendurchgang wird das nächste Element aus der Liste genommen und in der Variable `element` gespeichert. 

Statt `element` können wir einen beliebigen Namen für diese Variable verwenden. 

Diese Variable wird auf den Bildschirm durch `print` ausgegeben. 

Wir sehen, dass die Schleife automatisch beendet wird, sobald das letzte Element erreicht wurde.

````python
my_list = ["A", 1, 2, "B"]

for element in my_list:
    print("Aktuelles Element:", element)
````

## `For`-Schleifen in Verbindung mit `range`

In der Praxis steht man häufig vor der Aufgabe, Code, z.B. 7-mal auszuführen. 

Dafür erzeugt man eine Liste mit den ersten 7 Zahlen und iteriert darüber. 

Dadurch lässt sich der gleiche Code mehrmals ausführen.

> <span style="color:green"> **_ERINNERUNG:_**\
 </span> `range(7)` erzeugt eine Liste mit den ersten 7 Zahlen `[0,1,2,3,4,5,6]`.

### Beispiele:

````python
for e in range(7):
    print("Diese Zeile wird 7-mal ausgegeben!")
````

Wir sehen, dass wir die Variable `e` nicht verwenden müssen. 

Ein weiteres Beispiel:

````python
names = ["Hans", "Sabrina", "Fritz"]

for name in names:
    print("Hallo", name)
````

Wir können beliebige Operationen mit den einzelnen Elementen in einer `for`-Schleife ausführen.

````python
names = ["Hans", "Sabrina", "Fritz"]
for name in names:
    name = name.upper()
    print("Hallo", name)
````

Die einzelnen Elemente `name` sind Strings, daher können wir die Operation `upper()` anwenden.

## Index mitzählen (`enumerate`)

Manchmal wollen wir nicht nur den Wert der Elemente in einer Liste haben, sondern auch ihren Index. Dies könnte man auf folgende Art lösen:

````python
my_list = ["A", "B", "C"]

for index in range(len(my_list)):
    print("Element an Index", index, ": ", my_list[index])
````

> <span style="color:red"> **_WICHTIG:_**\
 </span> Das funktioniert, ist aber nicht der "pythonische" Weg. \
 Wollen wir auf die einzelnen Elemente zugreifen, müssen wir das über ihren Index machen, davon ist abzuraten!

Besser ist es, `enumerate` zu verwenden, was eine Liste als Argument erwartet und für jeden Schleifendurchgang das Element selbst **und** seinen Index zurückliefert. 

Daher können wir in der Schleife sowohl direkt auf den Index als auch den Wert selber zugreifen. 

Für jeden Schleifendurchgang wird im nächsten Beispiel der Wert des Elements in `element` und sein Index in `index` gespeichert. 

````python
my_list = ["A", "B", "C"]

for index, element in enumerate(my_list):
    print("Element an Index", index, ": ", element)
````

Die Ausgabe ist die Selbe wie oben, der Code liest sich jedoch deutlich schöner!

Auch mit `enumerate` ist uns die Bezeichnung der Variablen überlassen. Wir könnten auch schreiben:

````python
l = ["A", "B", "C"]

for i, e in enumerate(l):
    print("Element an Index", i, ": ", e)
````

> <span style="color:blue"> **_INFO:_**\
 </span> Es ist immer besser lange und beschreibende Namen für Variablen zu verwenden! 

## `For`-Schleifen in Verbindung mit Wörterbüchern

Häufig wollen wir über ein Wörterbuch iterieren und die einzelnen Werte zu den jeweiligen Schlüsseln ausgeben lassen. 

Hier hilft die Methode `my_dict.keys()` für ein beliebiges Wörterbuch `my_dict`. 

Diese Methode liefert als Ergebnis ein Objekt mit sämtlichen Schlüsseln des Wörterbuchs, z.B.

````python
countries = {"Deutschland": "Berlin", "Frankreich": "Paris", "Polen": "Warschau"}
print(countries.keys())
````

Über dieses Objekt können wir nun wie gewohnt in einer Schleife iterieren:

````python
for key in countries.keys():
    print("Die Hauptstadt von", key, "ist", countries[key])
````

Genau das gleiche kann man auch durch `dict.items()` erreichen. 

Ähnlich wie bei `enumerate` können wir jetzt beides `key` und `value` iterieren und brauchen nicht den Wert durch den Index zu finden.

````python
for key, value in countries.items():
    # value statt countries[key]
    print("Die Hauptstadt von", key, "ist", value)
````

# Aus einer Schleife "ausbrechen"

Manchmal wollen wir eine `for` bzw. `while`-Schleife vorzeitig abbrechen. 

Dies wird über das Schlüsselwort `break` erreicht. 

Wird innerhalb einer Schleife ein `break` erreicht, wird die Schleife **sofort** verlassen!

Im folgenden Beispiel wird die Schleife (vorzeitig) beendet, wenn `x` den Wert `5` annimmt.



````python
x = 0

while x < 10:
    x += 1
    if x == 5:
        break
    print(x)

print("Hier befinden wir uns außerhalb der while-Schleife!")
````

> <span style="color:blue"> **_ERINNERUNG:_**\
</span>Es gibt zwei Wege in Python einen Wert zu inkrementieren. \
Entweder benutzt man `x = x + 1` oder `x += 1`. \
Das gilt auch für die restlichen mathematischen Operatoren.  

Auf diese Art kann man theoretisch auch eine Endlosschleife verlassen. 

Das nächste Beispiel zeigt eine Schleife, deren Abbruchbedingung niemals `False` wird. 

Dennoch können wir die Schleife verlassen durch `break`!

````python
while True:
    my_string = input("Schleife beenden? j/n ")
    if my_string == "j":
        break
    print("Schleife läuft noch!")

print("Schleife verlassen!")
````

Auch für `for`-Schleifen funktioniert `break`.

````python
my_list = [1, 2, 3, "stop", 4]

for element in my_list:
    if element == "stop":
        break

    print(element)
````

Das letzte Element der Liste (`4`) wird nicht erreicht, da vorher ein `break` erreicht wird.

# Eine Schleife "fortsetzen"

Mit dem Schlüsselwort `continue` können wir direkt zum Schleifenkopf zurückkehren, ohne weiteren Code innerhalb der Schleife auszuführen.

Dies kann sich als nützlich erweisen.

Im folgenden Programm werden alle Zahlen zwischen 0-20 ausgegeben, welche durch 3 teilbar sind. 

Wenn eine Zahl nicht durch 3 teilbar ist, wird durch `continue` direkt wieder an Beginn der Schleife gesprungen, das `print` wird in diesem Fall nicht erreicht.

````python
for i in range(20):
    if not i % 3 == 0:
        continue

    print(i, "ist durch 3 teilbar.")
````

`continue` ist auch in `while`-Schleifen verwendbar. 

Im folgenden Beispiel werden nur die ungeraden Zahlen ausgegeben, wird nämlich eine gerade Zahl erreicht, sorgt `continue` dafür, dass das `print` nicht erreicht wird.

````python
x = 0

while x < 10:
    x += 1

    if x % 2 == 0:  # Zahl ist gerade
        continue

    print("Aktueller Wert von x:", x)
````
