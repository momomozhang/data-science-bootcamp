# Docstrings

Bei größeren Pythonprojekten ist eine Dokumentation des Codes sehr wichtig. Besonders bei komplexen Funktionen und längerem Code kann man leicht den Durchblick verlieren. Eine Dokumentation hilft auch anderen Programmierern dabei, deinen Code zu verstehen.

**Docstrings** bieten eine einfache Möglichkeit einen Teil des Programmes, sei es eine Funktion, eine Klasse oder ein Modul, zu dokumentieren. 

Ein Docstring wird durch drei Einzel- `'''` oder Doppelanführungszeichen `"""` definiert.

Wenn man eine Funktion dokumentiert, gibt man als erstes eine kleine Beschreibung der Funktion an. Als nächstes spezifiziert man die Argumente welche die Funktion erwartet und das Ergebnis, welches von der Funktion zurückgeliefert wird. Dadurch wird anderen Programmierern die Funktionsweise der Funktion schneller klar, als wenn sie *nur* den Code zur Verfügung hätten. Unten siehst du ein kurzes Beispiel, wie für eine (einfache) Funktion der Docstring aussehen kann:

```python
def add_numbers(x, y):
    """
    Liefert die Summe der zwei Zahlen.

        Parameters:
            x (int): Erste Zahl
            y (int): Zweite Zahl

        Returns:
            sum (int): Die Summe von x und y.
    """

    return x + y
```

Diesen Docstring kann man im Code der Funktion sehen, wenn es unklar ist was für Input oder Ouput ein Funktion hat oder um schnell zu verstehen was die Funktion macht. 

Zusätzlich findet sich der Docstring auch unter dem Attribut `__doc__` der Funktion als String. 

```python
print(add_numbers.__doc__)
```

Das gleiche Ergebnis erhalten wir auch über die Built-In Funktion `help`:

```python
help(add_numbers)
```

Docstrings kannst du in allen großen Projekten (manchmal mit unterschiedlichen Dokumentationsstil) beobachten. Unten ist z.B. eine Funktion aus dem [Seaborn Code](https://github.com/mwaskom/seaborn/blob/master/seaborn/utils.py) gezeigt.
