# Die `dir`-Funktion

Häufig braucht man einen Überblick über die verfügbaren Methoden und Attributen eines Objektes, sei es weil man den Namen einer Funktion vergessen hat oder weil man nach alternativen Lösungen sucht. 

In Python ist [`dir()`](https://docs.python.org/3/library/functions.html#dir) eine integrierte Funktion, die verwendet wird, um eine Liste der in einem bestimmten Namespace definierten Eigenschaften und Methoden zu erhalten. Wenn `dir()` ohne Argumente aufgerufen wird, gibt es die Namen im aktuellen lokalen Namespace zurück. Wenn es auf ein Objekt angewendet wird, gibt `dir()` die Namen zurück, die in dem Namespace des Objekts definiert sind. 

Die zurückgegebene Liste enthält sowohl die Standardattribute des Objekts, also die built-in Funktionen, welche für alle Objekte definiert sind, wie z.B. `__str__` oder `__init__`, als auch die Attribute, die vom Benutzer definiert wurden; alle aber ohne deren Werte. 

`dir()` ist besonders nützlich, wenn man ein neues Modul oder eine neue Klasse erstellt, um sicherzustellen, dass alle erforderlichen Attribute und Methoden vorhanden sind.

## Beispiele:


```python
liste = [1, 2, 3]
dir(liste)
```

```python
class BeispielKlasse:
    def __init__(self):
        self.name = "Beispiel"

    def beispiel_methode(self):
        print("Dies ist eine Beispiel-Methode")


print(dir(BeispielKlasse))
```
