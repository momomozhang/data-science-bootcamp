## Aufgabe 1: Wörterbuch umdrehen

Schreibe eine Funktion `reverse_dict`, die ein Wörterbuch als Argument (`dictionary`) erhält. Die Funktion soll die Schlüssel und Werte umtauschen und das neue Wörterbuch zurückgeben.

### Anforderungen:
- Die Funktion `reverse_dict(dictionary: dict) -> dict` nimmt ein Wörterbuch als Argument und gibt ein neues Wörterbuch zurück, in dem die Schlüssel und Werte vertauscht sind.

**Hinweis:** Die Methode `dictionary.items()` gibt eine Ansicht des Wörterbuchs als `dict_items`-Objekt zurück. Dieses enthält alle Schlüssel-Wert-Paare des Wörterbuchs und lässt sich bequem in Schleifen oder anderen Iterationsstrukturen verwenden.

```python
# Lösungsvorschlag 1 - mit Dictionary Comprehension
def reverse_dict(dictionary: dict) -> dict:
    return {
        value: key for key, value in dictionary.items()
    }  # achtet auf die vertauschte Reihenfolge und die Einfachheit der Lösung!
```

```python
# Lösungsvorschlag 2 - mit map und lambda
def reverse_dict(dictionary: dict) -> dict:
    return dict(map(lambda kv: (kv[1], kv[0]), dictionary.items()))  # kv: key-value
```

```python
# Lösungsvorschlag 3 - mit for-Schleife
def reverse_dict(dictionary: dict) -> dict:
    reversed_dict = {}
    for key, value in dictionary.items():
        reversed_dict[value] = key
    return reversed_dict
```

```python
# Lösungsvorschlag 4 - mit zip und Dictionary Constructor
def reverse_dict(dictionary: dict) -> dict:
    keys, values = zip(*dictionary.items())
    return dict(zip(values, keys))
```

```python
# Testcases
print(reverse_dict({"a": 1, "b": 2, "c": 3}))  # {1: 'a', 2: 'b', 3: 'c'}
print(reverse_dict({"x": 10, "y": 20}))  # {10: 'x', 20: 'y'}
print(reverse_dict({1: "eins", 2: "zwei"}))  # {'eins': 1, 'zwei': 2}
print(
    reverse_dict({"schluessel": "wert", "flux": "kompensator", "drittes": "element"})
)  # {'wert': 'schluessel', 'kompensator': 'flux', 'element': 'drittes'}
```

## Aufgabe 2: Summe fehlender Zahlen

Schreibe eine Funktion `sum_missing_numbers`, die eine Liste mit Zahlen `numbers: list[int]` entgegennimmt und die Summe der Zahlen zurückgibt, die zwischen dem Maximalwert und dem Minimalwert fehlen.

### Anforderungen:
- Die Funktion `sum_missing_numbers(numbers: list[int]) -> int` soll eine Liste von ganzen Zahlen als Argument nehmen und die Summe der fehlenden Zahlen zwischen dem maximalen und minimalen Wert zurückgeben.
- Nutze ruhig die Python-Funktionen `min()` und `max()`


```python
# Lösungsvorschlag 1 - for-Schleife
def sum_missing_numbers(numbers: list[int]) -> int:
    min_num, max_num = min(numbers), max(numbers)
    total_sum = 0
    for num in range(min_num, max_num + 1):
        if num not in numbers:
            total_sum += num
    return total_sum
```

```python
# Lösungsvorschlag 2 - mit list comprehension und sum
def sum_missing_numbers(numbers: list[int]) -> int:
    min_num, max_num = min(numbers), max(numbers)
    return sum(num for num in range(min_num, max_num + 1) if num not in numbers)
```

```python
# Lösungsvorschlag 3 - mit Set-Differenz
def sum_missing_numbers(numbers: list[int]) -> int:
    min_num, max_num = min(numbers), max(numbers)
    all_numbers = set(range(min_num, max_num + 1))
    missing_numbers = all_numbers - set(numbers)
    return sum(missing_numbers)
```

```python
# Testfälle müssen alle True sein.
print(sum_missing_numbers([1, 3, 5]) == 6)  # 2 + 4
print(sum_missing_numbers([1, 2, 3]) == 0)  # keine fehlenden Zahlen
print(sum_missing_numbers([1, 4, 3, 7]) == 13)  # 2 + 5 + 6
print(sum_missing_numbers([10, 14, 11, 13]) == 12)  # 12
print(sum_missing_numbers([0, 2, 8, 6, 5]) == 15)  # 1 + 3 + 4 + 7
```

## Aufgabe 3: Harshad-Zahl

Eine [Harshad-Zahl](https://de.wikipedia.org/wiki/Harshad-Zahl) ist eine Zahl, die ohne Rest durch die Summe ihrer Ziffern teilbar ist.

- Schreibe eine Funktion `is_harshad_number`, die eine Zahl (`number: int`) übergeben bekommt und überprüft, ob diese Zahl eine Harshad-Zahl ist.

### Anforderungen:
- Die Funktion `is_harshad_number(number: int) -> bool` soll eine einzelne Zahl entgegennehmen.
- Sie soll `True` zurückgeben, wenn die Zahl eine Harshad-Zahl ist, ansonsten `False`.


```python
# Lösungsvorschlag 1 - sum und generator expression
def is_harshad_number(number: int) -> bool:
    digit_sum = sum(int(digit) for digit in str(number))
    return number % digit_sum == 0
```

```python
# Lösungsvorschlag 2 - mit while-Schleife
def is_harshad_number(number: int) -> bool:
    original_number = number
    digit_sum = 0
    while number > 0:
        digit_sum += number % 10
        number //= 10
    return original_number % digit_sum == 0
```

```python
# Lösungsvorschlag 3 - mit for-Schleife
def is_harshad_number(number: int) -> bool:
    original_number = number
    digit_sum = 0
    for digit in str(number):
        digit_sum += int(digit)
    return original_number % digit_sum == 0
```

```python
# Lösungsvorschlag 4 - Funktionaler Ansatz: sum + map
def is_harshad_number(number: int) -> bool:
    digit_sum = sum(map(int, str(number)))
    return number % digit_sum == 0
```

```python
# Lösungsvorschlag 5 - Funktionaler Ansatz: Rekursion
def digit_sum_recursive(number: int) -> int:
    if number == 0:
        return 0
    return number % 10 + digit_sum_recursive(number // 10)


def is_harshad_number(number: int) -> bool:
    return number % digit_sum_recursive(number) == 0
```

```python
# Testcases
print(is_harshad_number(10))  # True
print(is_harshad_number(12))  # True
print(is_harshad_number(15))  # False
print(is_harshad_number(49))  # False
print(is_harshad_number(777))  # True
```

## Aufgabe 4: Letzte Ziffer prüfen

Schreibe eine Funktion `has_same_last_digit`, die drei Zahlen `x`, `y` und `z` als Argumente entgegennimmt. Die Funktion soll `True` zurückgeben, wenn die letzte Ziffer von `x * y` der letzten Ziffer von `z` entspricht.

### Anforderungen:
- Die Funktion `has_same_last_digit(x: int, y: int, z: int) -> bool` nimmt drei ganze Zahlen entgegen und gibt `True` zurück, wenn die letzte Ziffer des Produkts `x * y` mit der letzten Ziffer von `z` übereinstimmt.
- Ansonsten soll die Funktion `False` zurückgeben.


```python
# Lösungsvorschlag 1 - direkter Vergleich


def has_same_last_digit(x: int, y: int, z: int) -> bool:
    return (x * y) % 10 == z % 10
```

```python
# Lösungsvorschlag 2 - String-Konvertierung


def has_same_last_digit(x: int, y: int, z: int) -> bool:
    return str(x * y)[-1] == str(z)[-1]
```

```python
# Lösungsvorschlag 3 - einzelne Rechnungen


def has_same_last_digit(x: int, y: int, z: int) -> bool:
    product_last_digit = x * y % 10
    z_last_digit = z % 10
    return product_last_digit == z_last_digit
```

```python
# Lösungsvorschlag 4 - verschachtelte Funktion


def has_same_last_digit(x: int, y: int, z: int) -> bool:
    def last_digit(n: int) -> int:
        return abs(n) % 10

    return last_digit(x * y) == last_digit(z)
```

```python
# Lösungsvorschlag 5 - mit Listen


def has_same_last_digit(x: int, y: int, z: int) -> bool:
    product_last_digit = [int(d) for d in str(x * y)][-1]
    z_last_digit = [int(d) for d in str(z)][-1]
    return product_last_digit == z_last_digit
```

```python
# Testcases
print(has_same_last_digit(5, 5, 105))  # True
print(has_same_last_digit(1, 1, 12))  # False
print(has_same_last_digit(15, 12, 100))  # True
print(has_same_last_digit(7, 3, 13))  # False
print(has_same_last_digit(10, 10, 50))  # True
```

## Aufgabe 5: Das Basler-Problem

Schreibe eine Funktion `sum_reciprocal_squares`, die für ein übergebenes `n` die Summe 

\begin{align}
\sum_{k=1}^{n} \frac{1}{k^2} = \frac{1}{1^2} + \frac{1}{2^2} + ... + \frac{1}{n^2}
\end{align}

berechnet. Die Funktion soll das Ergebnis als Fließkommazahl zurückgeben. 

### Anforderungen:
- Die Funktion `sum_reciprocal_squares(n: int) -> float` nimmt ein ganzzahliges Argument `n` entgegen und gibt die berechnete Summe zurück.

**Videoempfehlung: Für alle Neugierigen unter euch, die sich fragen, was es mit dem Basel-Problem auf sich hat, hier eine [Videoempfehlung](https://www.youtube.com/watch?v=d-o3eB9sfls) (engl., aber deutsche Untertitel verfügbar):** 

```python
# Lösungsvorschlag 1 - sum über Liste (List Comprehension)
def sum_reciprocal_squares(n: int) -> float:
    return sum([1 / (k * k) for k in range(1, n + 1)])  # sum() ist eine sehr nützliche Funktion
```

```python
# Lösungsvorschlag 2 - oder direkt mit Generator-Expression ohne Umweg über eine Liste
def sum_reciprocal_squares(n: int) -> float:
    return sum(1 / (k * k) for k in range(1, n + 1))
```

```python
# Lösungsvorschlag 3 - rekursiv
def sum_reciprocal_squares(n: int) -> float:
    if n == 0:  # int
        return 0.0  # float
    return 1 / (n * n) + sum_reciprocal_squares(n - 1)  # float
```

```python
# Lösungsvorschlag 4 - mit for-Schleife
def sum_reciprocal_squares(n: int) -> float:
    sum_ = 0.0  # 'sum' ist ein reserviertes keyword und sollte nicht gebraucht werden!
    for k in range(1, n + 1):
        sum_ += 1 / (k * k)  # == k**2
    return sum_
```

```python
# Lösungsvorschlag 5 - mit while-Schleife
def sum_reciprocal_squares(n: int) -> float:
    sum_ = 0.0
    k = 1
    while k <= n:
        sum_ += 1 / (k * k)
        k += 1
    return sum_
```

```python
print(sum_reciprocal_squares(1))  # 1.0
print(sum_reciprocal_squares(2))  # 1.25
print(sum_reciprocal_squares(5))  # 1.4636111111111112
print(sum_reciprocal_squares(10))  # 1.5497677311665408
print(sum_reciprocal_squares(100))  # 1.6349839001848923
print(sum_reciprocal_squares(1000))  # 1.6439345666815597
```

## Aufgabe 6: Perfekte Zahl (etwas kniffliger)

- Eine **perfekte Zahl** ist eine Zahl, bei der die Summe aller Teiler (außer sie selbst) wieder die Zahl ergibt.

- Schreibe eine Funktion `find_perfect_numbers`, die alle **perfekten Zahlen** zwischen `low` und `high` zurückgibt.

### Anforderungen:
- Die Funktion `find_perfect_numbers(low: int, high: int) -> list[int]` nimmt zwei Zahlen als Eingabeargumente entgegen (`low`, `high`) und gibt eine Liste der perfekten Zahlen im angegebenen Bereich zurück.


```python
# Lösungsvorschlag 1 - for-Schleife
def find_perfect_numbers(low: int, high: int) -> list[int]:
    perfect_numbers = []
    for number in range(low, high + 1):
        divisors_sum = sum(divisor for divisor in range(1, number) if number % divisor == 0)
        if divisors_sum == number:
            perfect_numbers.append(number)
    return perfect_numbers
```

```python
# Lösungsvorschlag 2 - verschachtelte while-Schleifen
def find_perfect_numbers(low: int, high: int) -> list[int]:
    perfect_numbers = []
    number = low
    while number <= high:
        divisors_sum = 0
        divisor = 1
        while divisor < number:
            if number % divisor == 0:
                divisors_sum += divisor
            divisor += 1
        if divisors_sum == number:
            perfect_numbers.append(number)
        number += 1
    return perfect_numbers
```

```python
# Lösungsvorschlag 3 - mit rekursiver Hilfsfunktion
def find_perfect_numbers(low: int, high: int) -> list[int]:
    def divisors_sum(number: int, divisor: int = 1) -> int:
        if divisor == number:
            return 0
        return (
            divisor + divisors_sum(number, divisor + 1)
            if number % divisor == 0
            else divisors_sum(number, divisor + 1)
        )

    perfect_numbers = [number for number in range(low, high + 1) if divisors_sum(number) == number]
    return perfect_numbers
```

```python
# Lösungsvorschlag 4 - List Comprehension
def find_perfect_numbers(low: int, high: int) -> list[int]:
    return [
        number
        for number in range(low, high + 1)
        if sum(divisor for divisor in range(1, number) if number % divisor == 0) == number
    ]
```

```python
# Lösungsvorschlag 5 - mit einer Generatorfunktion
def perfect_numbers(low: int, high: int) -> list[int]:
    def is_perfect(number: int) -> bool:
        return sum(divisor for divisor in range(1, number) if number % divisor == 0) == number

    return [number for number in range(low, high + 1) if is_perfect(number)]
```

```python
print(find_perfect_numbers(-1, 100))  # [6, 28]
print(find_perfect_numbers(0, 1000))  # [6, 28, 496]
print(find_perfect_numbers(1, 10000))  # [6, 28, 496, 8128]
print(find_perfect_numbers(1000, 10000))  # [8128]
```
