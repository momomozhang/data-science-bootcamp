# Goldener Schnitt


Wir nehmen an, dass wir zwei Längen $a$ und $b$ haben und $a$ länger ist als $b$, d.h. $a>b$.<br>

$a$ und $b$ befinden sich im Verhältnis des *Goldenen Schnitts*, wenn ihr Verhältnis (also $\frac{a}{b}$) das selbe ist, wie das Verhältnis zwischen der Summe $a+b$ und der größeren der beiden Längen (also $\frac{a+b}{a}$).

Das folgende Bild zeigt zwei Größen $a$ und $b$, die sich im *Goldenen Schnitt* befinden.




Quelle: Wikipedia, https://de.wikipedia.org/wiki/Goldener_Schnitt#/media/Datei:Golden_ratio_line_percentages.svg

Wir können obige Definition als Formel schreiben:

\begin{align}
\frac{a}{b}=\frac{a+b}{a}=: \phi
\end{align}

Wir haben direkt die Größe $\phi$ definiert für $\frac{a}{b}$, d.h. wenn das Verhältnis zweier Größen $\phi$ entspricht, befinden sie sich im *Goldenen Schnitt*.\
Im folgenden wollen wir versuchen $\phi$ zu bestimmen.


\begin{align}
&\frac{a}{b}=\frac{a+b}{a} \quad \big| \cdot a \\
\Longrightarrow \quad &\frac{a^2}{b}=a+b \quad \big| \cdot \frac{1}{b} \\
\Longrightarrow \quad &\frac{a^2}{b^2}=\frac{a}{b}+1 \quad \big| -\frac{a}{b}-1\\
\Longrightarrow \quad &\frac{a^2}{b^2} - \frac{a}{b} - 1 = 0 \\ 
\end{align}

Wenn wir jetzt die Definition $\phi:=\frac{a}{b}$ einsetzen, kommen wir auf eine Gleichung für $\phi$:

\begin{align}
\phi^2 - \phi - 1&=0 \\
\end{align}

Einsetzen in die Mitternachtsformel (p-q-Formel) ergibt als (positive) Lösung:
\begin{align}
\Longrightarrow \phi = \frac{1+\sqrt{5}}{2}
\end{align}
Das Ergebnis ist eine irrationale Zahl und ungefähr: 
\begin{align}
\phi =   1.6180339....
\end{align}

Wir können diese Annäherung an $\phi$ mit Python überprüfen.\
Wir setzen den Wert für $\phi$ in die quadratische Gleichung ein und kontrollieren das Ergebnis:

```python
phi = 1.6180339
print(phi**2 - phi - 1)
```

Die angezeigte Lösung ist geschrieben in wissenschaftlicher Notation.\
Das `e-07` am Ende der Zahl bedeutet $\cdot 10^{-7}$, also: 
\begin{align}
-1.9845079002323018\cdot10^{-7} \approx 0
\end{align},
da $10^{-7}$ schon sehr klein ist wird die Gleichung daher (annähernd) erfüllt und unsere obige Annäherung an das *wirkliche* (irrationale) $\phi$ ist gut genug.



# Fibonacci-Folge:

Die **Fibonacci Folge** ist eine berühmte mathematische Folge von Zahlen, definiert durch 

\begin{align}
F(n) &= F(n-1) + F(n-2) \\
F(0)&=1 \\
F(1)&=1
\end{align}

> <span style="color:green"> **_ERINNERUNG:_** </span> Vergleiche dazu auch die Lektion zur Rekursion. Dort wird genauer auf die Fibonacci-Folge eingegangen.

Wir können obige Definition rekursiv implementieren:

```python
def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)
```

Dadurch können wir uns die ersten Folgenglieder ausgeben lassen:

```python
for i in range(10):
    print(fib(i), end=",")
```

# Verbindung zwischen dem Goldenen Schnitt und der Fibonacci-Folge:

Auf den ersten Blick ist nicht erkennbar, wo (und wie) es eine Verbindung zwischen der **Fibonacci-Folge** und dem **Goldenen Schnitt** gibt.\
Man kann jedoch mathematisch zeigen, dass der Quotient zweier aufeinander folgender Glieder der **Fibonacci-Folge** sich dem Goldenen Schnitt annähert, d.h. mathematisch gilt:

\begin{align}
\frac{\text{fib}(n)}{\text{fib}(n-1)} \Longrightarrow \phi \quad \text{für } n \rightarrow \infty.
\end{align}
<br>

Hierbei bezeichnet fib(n) das n-te Folgenglied der Fibonacci-Folge. Wir wollen diese Ergebnis mit Python überprüfen.\
Zum Beispiel berechnen wir den Quotienten für $n=6$ 

```python
n = 6
g = fib(n) / fib(n - 1)
print(g)
```

Oder für $n=10$:

```python
n = 10
g = fib(n) / fib(n - 1)
print(g)
```

Das Ergebnis kommt für große $n$ immer näher an den Wert von $\phi$ heran!
