## Aufgabe 1: Finde die größere Zahl

Schreibe eine Funktion `find_larger_number`, die zwei Zahlen als Argumente (`num1`, `num2`) entgegennimmt und die größere der beiden Zahlen zurückgibt. Falls beide Zahlen gleich groß sind, soll eine der beiden Zahlen zurückgegeben werden.

### Anforderungen:
- Die Funktion nimmt zwei ganze Zahlen entgegen und gibt die größere Zahl zurück.
- Wenn `num1` und `num2` gleich groß sind, gibt die Funktion einfach `num1` oder `num2` zurück.


```python
# Schreibe deinen Code hier


def find_larger_number(num1, num2):
    return max(num1, num2)
```

```python
print(find_larger_number(3, 3))
print(find_larger_number(3, 5))
print(find_larger_number(-3, 3))
```

## Aufgabe 2: Finde die größte Zahl in einer Liste

Schreibe eine Funktion `find_largest_number`, die eine Liste von Zahlen als Argument (`numbers`) entgegennimmt und die größte Zahl in dieser Liste zurückgibt.

### Anforderungen:
- Die Funktion nimmt eine Liste ganzer Zahlen entgegen und gibt die größte Zahl zurück.
- Die Python-Funktionen `max`, `sort` und `sorted` dürfen für diese Aufgabe **nicht** verwendet werden.


```python
# Schreibe deinen Code hier


def find_largest_number(my_list: list[float]) -> float | None:
    if not my_list:
        return None
    else:
        largest = my_list[0]
        for num in my_list:
            if largest < num:
                largest = num
        return largest
```

```python
print(f"find_largest_number([1,9,-9]): {find_largest_number([1, 9, -9])}")
print(f"find_largest_number([-2,0,-1]): {find_largest_number([-2, 0, -1])}")
print(f"find_largest_number([1,9,200]): {find_largest_number([1, 9, 200])}")
print(f"find_largest_number([1,200,-9]): {find_largest_number([1, 200, -9])}")
print(f"find_largest_number([]): {find_largest_number([])}")
```

Schreibe eine Funktion `print_values`, die eine beliebige Anzahl von Argumenten annimmt und diese dann einzeln auf der Konsole ausgibt.

## Aufgabe 4: Summe aus beliebigen Zahlen

Schreibe eine Funktion `calculate_sum`, die eine beliebige Anzahl an Zahlen als Argumente entgegennimmt und die Summe aller dieser Zahlen zurückgibt.

### Anforderungen:
- Die Funktion `calculate_sum` nimmt eine beliebige Anzahl von Zahl-Argumenten entgegen.
- Die Summe der Argumente soll berechnet und zurückgegeben werden.
- Die Python-Funktion `sum` darf **nicht** verwendet werden.


```python
# Schreibe deinen Code hier


def calculate_sum(*nums):
    nums_sum = 0
    for num in nums:
        nums_sum += num
    return nums_sum
```

```python
print(calculate_sum(1, 9))
print(calculate_sum(1, 9, 22))
print(calculate_sum(-1, 9, 22))
```

## Aufgabe 6: Gerade Zahlen filtern und sortieren

Schreibe eine Funktion `filter_sort_even_numbers`, die aus einer übergebenen Liste von ganzen Zahlen alle geraden Zahlen extrahiert, diese sortiert und in einer neuen Liste zurückgibt.

### Anforderungen:
- Die Funktion nimmt eine Liste von **ganzen** Zahlen entgegen.
- Die zurückgegebene Liste soll nur gerade Zahlen enthalten, sortiert in aufsteigender Reihenfolge.


```python
# Schreibe deinen Code hier


def filter_sort_even_numbers(num_list: list[int]) -> list[int]:
    new_list = []
    for num in num_list:
        if num % 2 == 0:
            new_list.append(num)
    return sorted(new_list)
```

```python
print(filter_sort_even_numbers([1, 9, 12, -4, 11, 28, 72, 3, -2]))
print(filter_sort_even_numbers([]))
```

## Aufgabe 7: Berechnung des euklidischen Abstandes

Schreibe eine Funktion `calculate_euclidean_distance`, die die Koordinaten zweier Punkte in der Ebene entgegennimmt und den euklidischen Abstand zwischen diesen beiden Punkten berechnet und zurückgibt. Die Koordinaten der Punkte werden jeweils als Listen mit zwei Elementen übergeben, wobei das erste Element die x-Koordinate und das zweite Element die y-Koordinate darstellt.

Der Abstand zweier Punkte 
$p_1=
\begin{pmatrix}
x_1 \\ y_1
\end{pmatrix}
, \
p_2=
\begin{pmatrix}
x_2 \\ y_2
\end{pmatrix}
$
in der Ebene ist definiert durch

\begin{align}
        d = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2} .
\end{align}

### Anforderungen:
- Die Funktion nimmt zwei **Listen von Fließkommazahlen** entgegen, welche die Koordinaten von zwei Punkten darstellen (`p1` und `p2`), und gibt den euklidischen Abstand zwischen diesen Punkten als Fließkommazahl zurück.


```python
# Schreibe deinen Code hier


def calculate_euclidean_distance(p1: list[float], p2: list[float]) -> float:
    if len(p1) != 2 or len(p2) != 2:
        raise ValueError("Points must have exactly 2 coordinates")
    else:
        x1, y1 = p1
        x2, y2 = p2
        distance = ((x1 - x2) ** 2 + (y1 - y2) ** 2) ** 0.5
    return distance
```

```python
calculate_euclidean_distance([0, 0], [1, 1])
```

## Aufgabe 8: Primzahlprüfung

Schreibe eine Funktion `is_prime`, die eine Zahl als Argument entgegennimmt und `True` zurückgibt, wenn die Zahl eine Primzahl ist. Andernfalls gibt sie `False` zurück. Eine Primzahl ist eine Zahl, die nur durch 1 und sich selbst ohne Rest teilbar ist.<br> 
Anders ausgedrückt: eine Primzahl besitzt immer nur zwei Teiler, nie mehr, aber auch nicht weniger.

### Anforderungen:
- Die Funktion nimmt eine ganze Zahl entgegen und gibt einen Booleschen Wert zurück (`True` für eine Primzahl, `False` andernfalls).


```python
# Schreibe deinen Code hier


def is_prime(num: int) -> bool:
    if num <= 1:
        raise ValueError("Input value must be greater than 1")
    elif num == 2:
        return True
    elif num % 2 == 0:
        return False
    else:
        for i in range(3, int(num**0.5) + 1, 2):  # Check odd divisors only, up to sqrt(num)
            if num % i == 0:
                return False
        return True
```

```python
# print(f"0 is a prime number: {is_prime(0)}")
# print(f"1 is a prime number: {is_prime(1)}")
print(f"2 is a prime number: {is_prime(2)}")
print(f"3 is a prime number: {is_prime(3)}")
print(f"4 is a prime number: {is_prime(4)}")
print(f"5 is a prime number: {is_prime(5)}")
print(f"6 is a prime number: {is_prime(6)}")
print(f"7 is a prime number: {is_prime(7)}")
print(f"8 is a prime number: {is_prime(8)}")
```

## Aufgabe 9: Zahlen verdoppeln

Schreibe eine Funktion `double_numbers`, die eine Liste von Zahlen entgegennimmt und eine neue Liste zurückgibt, in der jede Zahl verdoppelt wurde.

### Anforderungen:
- Die Funktion nimmt eine Liste von **ganzen Zahlen** entgegen und gibt eine neue Liste mit den verdoppelten Werten zurück.


```python
# Schreibe deinen Code hier


def double_numbers(my_list: list[int]) -> list[int]:
    new_list = []
    for num in my_list:
        new_list.append(num * 2)
    return new_list
```

```python
print(double_numbers([2, -6, 10]))
print(double_numbers([0, 0, 0]))
print(double_numbers([]))
```

## Aufgabe 10: Fakultät berechnen

Schreibe eine Funktion `calculate_factorial`, die die Fakultät einer übergebenen Zahl berechnet. Die Fakultät einer Zahl \(n\) ist definiert als das Produkt aller ganzen Zahlen von 1 bis n. 

Die Fakultät einer Zahl $n$ ist definiert als

$$
\prod_{i=1}^n i= 1 \cdot 2\cdot ... \cdot i
$$


### Anforderungen:
- Die Funktion nimmt eine ganze Zahl **n** entgegen und gibt die Fakultät von **n** zurück.
- Es soll keine externe Bibliothek für die Berechnung verwendet werden.
- Beachte, dass die Fakultät von 0 als 1 definiert ist.
- Es gibt viele verschiedene Lösungsansätze. Insbesondere ist hier auch eine Rekursion zielführend.


```python
# Schreibe deinen Code hier


def calculate_factorial(n: int) -> int:
    if n < 0:
        raise ValueError("n must be a positive integer.")
    elif n <= 1:
        return 1
    else:
        return n * calculate_factorial(n - 1)
```

```python
# print(calculate_factorial(-1))
print(calculate_factorial(0))
print(calculate_factorial(1))
print(calculate_factorial(2))
print(calculate_factorial(3))
print(calculate_factorial(4))
```

## Aufgabe 11: Größter gemeinsamer Teiler

Implementiere die rekursive Funktion `gcd` (*engl: greatest common divisor*), die den größten gemeinsamen Teiler zweier Zahlen `x` und `y` findet. 

$$
\text{gcd}(x,y)=
\begin{cases}
x & \text{ falls } y=0 \\
\text{gcd}(y, x\%y) & \text{ falls } y>0
\end{cases}
$$


### Anforderungen:
- Die Funktion nimmt zwei ganze Zahlen entgegen und gibt ihren größten gemeinsamen Teiler zurück.
- Verwende folgende rekursive Definition:
  - Wenn `y` gleich `0` ist, ist der ggT gleich `x`.
  - Andernfalls ist der größte gemeinsame Teiler gleich `gcd(y, x % y)`.
  


```python
# Schreibe deinen Code hier


def gcd(x: int, y: int) -> int:
    if y == 0:
        return x
    else:
        return abs(
            gcd(y, x % y)
        )  # Mathematisch gesehen darf y bei der gcd-Berechnung auch negativ sein
```

```python
print(gcd(1, 0))
print(gcd(2, 3))
print(gcd(17, 51))
print(gcd(22, -55))
print(gcd(-22, 66))
print(gcd(-8, 2))
print(gcd(2, -8))
```

## Aufgabe 12: Gemeinsames Element zweier Listen finden

Schreibe eine Funktion `has_common_element`, die zwei Listen (`list1`, `list2`) als Argumente erwartet und `True` zurückgibt, wenn die Listen mindestens ein gemeinsames Element enthalten. Andernfalls soll `False` zurückgegeben werden.

### Anforderungen:
- Die Funktion nimmt zwei Listen entgegen und gibt `True` zurück, wenn mindestens ein Element in beiden Listen vorhanden ist. Andernfalls gibt sie `False` zurück.


```python
# Schreibe deinen Code hier


def has_common_element(list1: list, list2: list) -> bool:
    set1 = set(list1)
    set2 = set(list2)
    intersection = set1 & set2
    return len(intersection) > 0
```

```python
print(has_common_element([1, "a", "b"], [2, 9, "c"]))
print(has_common_element([1, "a", "b"], [2, 9, "a"]))
print(has_common_element([1, "a", "b"], []))
```

## Aufgabe 13: Element aus Liste entfernen

Schreibe eine Funktion `remove_element`, die eine Liste und einen Index entgegennimmt und die Liste ohne das Element an diesem Index zurückgibt. <br>Falls der Index außerhalb des gültigen Bereichs liegt, soll die gesamte Liste zurückgegeben werden.

### Anforderungen
- Die Funktion `remove_element` erwartet eine Liste und einen Index als Argumente.
- Wenn der Index innerhalb des gültigen Bereichs der Liste liegt, wird die Liste ohne das Element an diesem Index zurückgegeben.
- Liegt der Index außerhalb des gültigen Bereichs, wird die gesamte Liste unverändert zurückgegeben.


```python
# Schreibe deinen Code hier


def remove_element(my_list: list, index: int) -> list:
    if 0 <= index < len(my_list):
        return my_list[:index] + my_list[index + 1 :]
    else:
        return my_list
```

```python
print(remove_element([1, "a", "b"], 2))
# print(remove_element([1, "a", "b"], -1))
print(remove_element([1, "a", "b"], 0))
print(remove_element([], 0))
print(remove_element([[1, "a", "b"], 1, (1, 9, -8)], 2))
```
