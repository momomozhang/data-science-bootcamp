# Objekt-Orientierte Programmierung (OOP)

## Klassen in Python

**Klassen** werden verwendet um Daten _und_ Funktionalität zu bündeln. Wir können unsere eigenen Klassen schreiben, um einen neuen *Datentyp* zu definieren. Klassen können als *Blaupausen* für Objekte in der Realität angesehen werden, für die wir ein allgemeines Verhalten definieren können/wollen. Ausgehend von dieser Klasse können wir konkrete Objekte davon erzeugen, sogenannte **Instanzen** der Klasse. Der Prozess eine Instanz einer Klasse zu erzeugen wird als *Instanziierung* bezeichnet.

> <span style="color:red"> **_WICHTIG:_** </span>  Eine Klasse ist eine Vorlage, die beschreibt, welche Eigenschaften (Attribute) und Funktionen (Methoden) Instanzen dieser Klasse haben können.

Mit Klassen können Programme modularer und skalierbarer gestaltet werden, da man benutzerdefinierte Datentypen erstellen und diese im Code verwenden kann. Wird eine neue Klasse definieren, erstellen man im Grunde einen neuen Datentyp.

In Python haben wir schon unterschiedliche [Klassen](https://docs.python.org/3/tutorial/classes.html) gesehen, z.B. die Klasse *String*. 
Die folgende Zeile instanziiert eine Instanz der Klasse *String*:

```python
s = "Hallo"
```

Tatsächlich ist die Instanz `s` vom Typen `string`:

```python
print(type(s))
```

Wir haben schon gesehen, dass wir über diese Instanz nicht nur Zugriff auf die Daten (in diesem Fall das Wort "Hallo") haben, sondern auch über Funktionalität verfügen, z.B. können wir die Daten verändern und in Großbuchstaben umwandeln:


```python
s.upper()
```

Die Methode `upper` steht nur einer Instanz der Klasse *String* zur Verfügung. Wenn wir z.B. eine Instanz der Klasse *Integer* haben, gibt es diese Funktion nicht. 

```python
a = 3
print(type(a))
```

```python
a.upper()  # Gibt einen Fehler
```

> <span style="color:red"> **_WICHTIG:_** </span> Je nach Datentyp stehen unterschiedliche Funktionalitäten zur Verfügung.


Wir wollen in dieser Lektion als erstes eine (sehr abstrakte und sehr kleine) Klasse `MyFirstClass` erzeugen um uns an die Syntax und (was schwieriger ist) das Konzept von Klassen zu gewöhnen. Diese Klasse ist in der Lage, Daten zu speichern und wir statten sie zusätzlich noch mit einer (kleinen) Funktionalität aus.
Sobald dies funktioniert, wollen wir als praktisches Anwendungsbeispiel eine neue (konkretere) Klasse `Client` erzeugen, die einen Kunden aus einer Kundenkartei darstellt.

## Unsere erste Klasse `MyFirstClass`


```python
class MyFirstClass:
    a = 3

    def func(self):
        print("Hello")
```

> <span style="color:red"> **_WICHTIG:_** </span> Um unseren Code einheitlich und lesbar zu machen, können wir versuchen einige Konventionen zur Benennung von Objekten zu verwenden:
> - Variablen & Funktionen: Kleinbuchstaben und mit Unterstichen (*snake_case*), z.B.:
    - `add_two_numbers()`
    - `my_first_variable = 3`

> - Klassen: In Großbuchstaben ohne Unterstrich (*CamelCase*), z.B:
    - `MyFirstClass`
    - `Car`
    - `CarEngine`



Die erste Zeile signalisiert Python, dass wir eine Klasse namens `MyFirstClass` definieren wollen. Alles, was nun eingerückt folgt, gehört zur Klasse `MyFirstClass`. Hier können wir nun auf gewohnt Weise Variablen und Funktionen definieren.

Die Klasse `MyFirstClass` hat $2$ Attribute: eine Variable `a` und eine Funktion `func`. 

> <span style="color:blue"> **_INFO:_** </span> Für Attribute von Klassen, die Funktionen sind, sagt man auch *Methoden*. Im obigen Beispiel ist `func` damit eine Methode der Klasse `MyFirstClass`. Alles was wir über Funktionen gelernt haben, gilt auch für Methoden. 

Die Methode `func` der Klasse `MyFirstClass` macht nichts weiter als `Hello` auf den Bildschirm auszugeben.

> <span style="color:blue"> **_INFO:_** </span> Auf den Parameter `self` in der Methode `func` werden wir später in diesem Notebook eingehen.

## Erzeugen von Instanzen einer Klasse

Wir können nun - genau wie bei den anderen Datentypen - Instanzen vom unserer neuen Klasse (Datentyp) erzeugen:

```python
x = MyFirstClass()
y = MyFirstClass()
```

Die erste Zeile erzeugt eine neue Instanz der Klasse `MyFirstClass` und weist sie der Variablen `x` zu.
Die zweite Zeile erzeugt eine weitere Instanz der Klasse und weist sie der Variablen `y` zu. 
Wir haben nun 2 Instanzen der Klasse erzeugt. Es können **beliebig viele Instanzen** einer Klasse erstellt werden. Wie gewohnt können wir uns über die in Python eingebaute Funktion `type()` den Typen der Variablen `x`, bzw. `y` angeben lassen:

```python
print(type(x))
print(type(y))
```

Wir sehen (unter anderem), dass die Variablen zur Klasse `MyFirstClass` gehören. Über die Funktion [`isinstance()`](https://docs.python.org/3/library/functions.html#isinstance) könnten wir, wenn wir wollen, eine boolsche Abfrage machen:

```python
isinstance(x, MyFirstClass)
```

Das Ergebnis ist `True`, d.h. `x` ist eine Instanz von `MyFirstClass`.

## Attribute und Methoden der Klassen abfragen

Wir können über die Instanzen nun auf die Daten und Methoden der Klasse zugreifen, indem wir zwischen dem Namen der Instanz und dem Attribut einen Punkt `.` setzen:

```python
x.a  # Greift auf das Attribut `a` der Instanz `x` zu
```

```python
x.func()  # Führt die Methode `func` der Instanz `x` der Klasse aus
```

```python
y.func()  # Führt die Methode `func` der Instanz `y` der Klasse aus
```

Zunächst unterscheiden wir unsere Variablen zwischen **Klassenvariablen** und **Instanzvariablen**.

## Klassenvariablen 

Die Variable `a` in `MyFirstClass` ist eine **Klassenvariable**. Ihr Wert wird von *allen* Instanzen der Klasse geteilt. Wenn wir den Wert ändern, ändert er sich für **alle** Instanzen der Klasse:

```python
MyFirstClass.a = 10
```

Wir sehen hier in der ersten Zeile, dass wir hier *nicht* über den Namen der Instanzvariable, sondern über den *allgemeinen* Namen der Klasse auf die Variable `a` zugreifen! Unsere Instanzen haben nun den neuen Wert für `a`:

```python
print(x.a)
print(y.a)
```

Um eine Klassenvariable zu definieren, muss sie innerhalb der Klassendefinition, aber außerhalb der Methoden definiert werden, siehe auch im Beispiel von `MyFirstClass`.

## Instanzvariablen 

Hätten wir nur Klassenvariablen zur Verfügung, wären Klassen nicht besonders wertvoll. Wir könnten zwar unterschiedliche Instanzen erzeugen, aber diese hätten immer den gleichen Wert, da Klassenvariablen - wie oben gesehen - für alle Instanzen gelten.

Meistens wollen wir in der Lage sein, für unterschiedliche Instanzen unterschiedliche Werte zu speichern. Hier kommen *Instanzvariablen* ins Spiel. **Instanzvariablen** sind einzigartig für jede Instanz einer Klasse. Innerhalb einer Klasse können wir auf Instanzvariablen mithilfe des Schlüsselwortes `self` zugreifen. `self` verweist immer auf die aktuelle Instanz der Klasse. Das folgende Kapitel zeigt, wie wir eine Klasse mit Instanzvariablen anlegen können:

***

## Die __init__()  Methode

Jede Klasse, die wir definieren, hat eine Methode namens `__init__`, welche _automatisch_ von Python aufgerufen wird, wenn eine Instanz der Klasse erzeugt wird. Die `__init__`-Methode wird normalerweise verwendet, um die Eigenschaften eines Objekts zu initialisieren und den Zustand des Objekts beim Erstellen zu definieren.<br>


> <span style="color:blue"> **_INFO:_** </span> Da `__init__` beim Erzeugen (Konstruieren) einer Instanz einer Klasse aufgerufen wird, bezeichnet man die Funktion auch als _Konstruktor_ .

> <span style="color:red"> **_WICHTIG:_** </span> `__init__` **muss** mit 2 führenden und 2 endenden Unterstrichen geschrieben werden. 

**Beispiel**

Im nächsten Beispiel definieren wir eine neue Klasse `Dog`, die in der Instanzvariablen `name`, den jeweiligen Namen des Hundes speichern soll. Dieser Name kann unterschiedlich von Instanz (Hund) zu Instanz (Hund) sein.

```python
class Dog:
    # Das ist eine Klassenvariable, die von allen Instanzen geteilt wird
    # ALLE Hunde haben 4 Beine
    number_of_legs = 4

    def __init__(self, name):
        self.dogname = name  # dogname ist eine Instanzvariable.
        # Jede Instanz der Klasse kann einen
        # unterschiedlichen Wert stehen haben
        # JEDER Hund hat einen unterschiedlichen Namen
```

Der Klasse `Hund` wurde zusätzlich noch eine **Klassenvariable** `number_of_legs` gegeben. Das macht hier Sinn, wir gehen davon aus, dass *alle* Hunde $4$ Beine haben.<br>
Wir erzeugen eine Instanz vom Typen `Dog`. Diese Instanz erhält den Namen "Hasso".

```python
dog1 = Dog("Hasso")  # Python ruft die Methode __init__
# mit dem Argument "Hasso" auf
```

Wir sehen, dass wir hier nun beim Erzeugen der Instanz das Argument `"Hasso"` mitgeben. Diese Argument entspricht dem Argument `name` in der Definition der Funktion `__init__(self,name)` oben.  Dieser übergebene Wert wird in der Instanzvariablen `self.dogname` gespeichert. Durch das Verwenden von `self` steht die Variable allen Methoden der Klasse zur Verfügung. Das werden wir später noch sehen, wenn wir die Klasse um weitere Funktionalität erweitern (siehe unten).
Für die erzeugte Instanz `dog1` können wir nun die Werte abfragen:

```python
print(dog1.number_of_legs)
print(dog1.dogname)
```

Wir "erzeugen" nun einen weiteren Hund mit einem anderen Namen `"Ted"`:

```python
dog2 = Dog("Ted")
```

```python
print(dog2.number_of_legs)  # Klassenvariable ist universell für alle Instanzen der Klasse `Dog`
print(dog2.dogname)  # Instanzvariable ist zugänglich zur eigenen Instanz durch `self`.
```

Wir sehen, dass `dog1` und `dog2` unterschiedliche Namen haben. Das wollten wir erreichen!<br>
Wir wollen nun die Funktionalität der Klasse `Dog` erweitern und noch eine Methode zur Verfügung stellen, in der wir auf den jeweiligen Namen zugreifen können. Diese soll `bellen` heißen und eine Ausgabe auf den Bildschirm ausgeben, in dem der Name des Hundes angezeigt wird, der bellt.

```python
class Dog:
    number_of_legs = 4  # Das ist eine Klassenvariable, die von allen Instanzen geteilt wird

    def __init__(self, name):
        self.dogname = name  # name ist eine Instanzvariable.
        # Jede Instanz der Klasse kann einen
        # unterschiedlichen Wert stehen haben

    def bellen(self):
        print(self.dogname, "bellt!")
```

Wie bei `__init__` müssen wir der Funktion `bellen` ein Argument `self` mitgeben. Dies müssen wir innerhalb von Klassen *immer* machen. Das Argument `self` bezieht sich, wie oben schon angedeutet, immer auf die **Instanz**, welche die jeweilige Methode aufruft. So kann Python wissen, *welcher* Hund bellt!

Schauen wir uns das Ergebnis an:<br>
Wir "erzeugen" wieder zwei Hunde...

```python
dog1 = Dog("Hasso")
dog2 = Dog("Balu")
```

...und lassen sie bellen!

```python
dog1.bellen()
```

```python
dog2.bellen()
```

Es wird erkannt, welcher Hund bellt!

> <span style="color:red"> **_WICHTIG:_** </span> Durch Klassen können mehrere Objekte mit denselben Attributen und Methoden erstellt werden, wobei jedes Objekt seine eigenen Werte für die Attribute haben kann.

## Beispielklasse: `Client`

Die Beispiele bisher waren etwas abstrakt, wir wollen uns nun ein konkreteres Beispiel überlegen.

Angenommen, wir arbeiten in einer Firma mit einer Kundenkartei. Jeder Kunde hat gewisse Eigenschaften, dargestellt in folgender Tabelle:

|Eigenschaft|
|---:|
|Vorname|
|Nachname|
|Anschrift|
|Geburtstag| 

Dies sind daher die **Instanzvariablen** unserer Klasse `Kunde`. 
Wir wollen sie daher der Funktion `__init__` übergeben:


```python
class Client:
    def __init__(self, firstname, lastname, address, birthday):
        self.firstname = firstname
        self.lastname = lastname
        self.address = address
        self.birthday = birthday
```

Wir können nun einen Kunden "erzeugen":

```python
Client1 = Client("Max", "Mustermann", "Berlin", "01.02.1975")
print(Client1.firstname)
print(Client1.lastname)
print(Client1.address)
print(Client1.birthday)
```

Wir würden gerne für die einzelnen Kunden eine Funktion definieren, die uns die Informationen über den Kunden schön ausgibt. Dafür ergänzen wir die Klassendefinition:

```python
class Client:
    def __init__(self, firstname, lastname, address, birthday):
        self.firstname = firstname
        self.lastname = lastname
        self.address = address
        self.birthday = birthday

    def show_info(self):
        print("Name des Kunden:", self.firstname, self.lastname)
        print("Wohnort:", self.address)
        print("Geburtstag:", self.birthday)
```

Nun können wir uns die Information über einen Kunden sehr kompakt ausgeben lassen:

```python
Client1 = Client("Max", "Mustermann", "Berlin", "01.02.1975")
Client2 = Client("Sandra", "Schmidt", "Köln", "14.03.1960")
Client1.show_info()
Client2.show_info()
```

Es ist möglich, dass wir feststellen, dass uns die bisherige Funktionalität nicht reicht. Möglicherweise wollen wir noch eine Methode zur Verfügung stellen, womit wir den Wohnort eines Kunden einfach ändern können. Wir ergänzen die Klassendefinition um eine weitere Methode:

```python
class Client:
    def __init__(self, firstname, lastname, address, birthday):
        self.firstname = firstname
        self.lastname = lastname
        self.address = address
        self.birthday = birthday

    def show_info(self):
        print("Name des Kunden:", self.firstname, self.lastname)
        print("Wohnort:", self.address)
        print("Geburtstag:", self.birthday)

    def change_address(self, new_address):
        print(
            "Ändere Adresse des Kunden",
            self.firstname,
            self.lastname,
            "von",
            self.address,
            "zu",
            new_address,
        )
        self.address = new_address
```

```python
Client1 = Client("Max", "Mustermann", "Berlin", "01.02.1975")
```

```python
Client1.show_info()
```

```python
Client1.change_address("Hamburg")
```

```python
Client1.show_info()
```

Wir sehen, dass die Adresse sich geändert hat!
Das einfache Ergänzen von Funktionalität und Attributen zu Klassen (wie wir es gerade gesehen haben) ist ein weiteres Argument für den Gebrauch von Klassen. Normalerweise haben wir nur *eine* Klassendefinition, die wir immer weitere ergänzen können. In diesem Notebook sind es mehrere um den Gedankenprozess nachvollziehen zu können, aber *normalerweise* würden wir die Definition nur *einmal* schreiben.

Die "Blaupause" unserer Klasse `Client` sieht momentan wie folgt aus:
