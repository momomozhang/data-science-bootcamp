Wie wir schon gesehen haben, ist es oft von Bedeutung, welchen (Daten)_Typ_ eine Variable hat. Dies definiert nämlich die Funktionen, die wir für diese Variable aufrufen können. Z.B. hat jede Variable vom Typ Liste eine Funktion `append` mit der etwas ans Ende der Liste angehängt werden kann. Strings steht diese Funktionalität _nicht_ zur Verfügung, dafür aber z.B. die Funktion `lower` mit der jeder Buchstabe klein geschrieben werden kann. 

> <span style="color:red"> **_WICHTIG:_** </span> Der Typ einer Variablen definiert die Funktionen, die man mit dieser Variablen aufrufen kann.

> <span style="color:blue"> **_INFO:_** </span> Funktionen, die an einen Datentypen "gekoppelt" sind, nennt man auch _Methoden_. Z.B. ist `append` eine _Methode_ des Datentyps Liste.

In diesem Notebook wollen wir betrachten, welche Möglichkeiten wir in Python haben, den Typ einer Variable zu erhalten. Außerdem wollen wir uns noch anschauen, wie man Datentypen in andere Datentypen umwandeln kann, z.b. den String `"3"` in die Zahl `3`.


# isinstance

Mit der Funktion [`isinstance(object, type)`](https://docs.python.org/3/library/functions.html#isinstance)  können wir überprüfen, ob ein Objekt einen speziellen Datentyp hat. Diese Funktion erwartet zwei Argumente.

**Argumente:**
- `object`: Das Objekt, welches wir überprüfen wollen.
- `type`: Der Datentyp auf den getestet werden soll.
    
Die Funktion gibt einen boolschen Wert (`True`, `False`) zurück.    
    
**Rückgabewert:**
     
- `True`, falls `object` vom Typ `type` ist.
- `False` ansonsten.

## Beispiele:

Wir erzeugen eine Liste `my_list`.

```python
my_list = [1, 2, 3]
```

Mit `isinstance` können wir nun überprüfen, ob die Variable `my_list` den Datentyp Liste hat: 

```python
isinstance(my_list, list)
```

Wie erwartet, ist `my_list` eine Liste.

```python
isinstance(my_list, str)
```

`my_list` ist kein String.

Wir erzeugen ein leeres Wörterbuch `my_dict` und prüfen, dass es wirklich ein Wörterbuch ist:

```python
my_dict = {}
isinstance(my_dict, dict)
```

# type()

Wie wir schon gesehen haben, können wir uns den Typ einer Variablen auch durch die Funktion [`type`](https://docs.python.org/3/library/functions.html#type) ausgeben lassen. Diese Funktion erwartet als Argument eine Variable und liefert den Typen der Variable zurück. 

## Beispiel:

```python
s = "Hallo"
print("Der Typ der Variablen s ist", type(s))
```

Wir sehen, dass die Variable `s` vom Typ `string` ist.

# Casting (Typenumwandlung)

Wir können den Datentyp einer Variablen ändern, indem wir sie in einen anderen Datentyp umwandeln. Dieser Prozess wird in der Informatik **"casting"** genannt. In Python stehen mehrere Funktionen zur Verfügung (_built-in functions_), je nachdem welcher Datentyp gewünscht ist:

- [`int(object)`](https://docs.python.org/3/library/functions.html#int)
- [`float(object)`](https://docs.python.org/3/library/functions.html#float)
- [`str(object)`](https://docs.python.org/3/library/functions.html#func-str)
- [`bool(object)`](https://docs.python.org/3/library/functions.html#bool)
- [`complex(object)`](https://docs.python.org/3/library/functions.html#complex)

Der Name der Funktion zeigt an, _in_ welchen Datentyp das Objekt `object` verwandelt (gecasted) wird (falls möglich).

**Argument:**
- `object`: Das Objekt, welches umgewandelt werden soll.

**Rückgabewert:**
- Das umgewandelte Objekt.


## Beispiele: `int()`

```python
x = 1
y = int(7.8)
z = int("5")

print(x)
print(y)
print(z)
```

Wir sehen das der Float Wert `7.8` in die ganze Zahl `7` umgewandelt wird. Python schneidet automatisch die Nachkommastellen ab, sodass das Resultat wirklich ein Integer ist. Auch ein String (`"5"`) kann in eine Zahl (`5`) umgewandelt werden. Zur Kontrolle können wir mit `isinstance` überprüfen, dass das Resultat wirklich ein Integer ist:

```python
print(isinstance(x, int))
print(isinstance(y, int))
print(isinstance(z, int))
```

Auch Wahrheitswerte können in Integer umgewandelt werden. Die Konvention in Python ist, dass `True` der Zahl `1` entspricht und `False` der Zahl `0`:

```python
print(int(True))
```

```python
print(int(False))
```

## Beispiele: `float()`

```python
x = float(1)
y = 7.8
z = float("3.5")

print(x)
print(y)
print(z)
```

`float(object)` wandelt das Objekt `object` in eine Dezimalzahl um. Im obigen Beispiel sehen wir, dass aus dem Integer `1` die Dezimalzahl `1.0` wird. Der String `"3.5"` kann ohne Probleme in die Dezimalzahl `3.5` umgewandelt werden.

```python
print(isinstance(x, float))
print(isinstance(y, float))
print(isinstance(z, float))
```

## Beispiele: `str()`

Zahlen können auch in Zeichenketten umgewandelt werden. Dafür benötigt man den Befehl `str()`:

```python
x = str(2)
y = str(5.0)
z = str(4 + 6)

print(x)
print(y)
print(z)
```

Die Variable `x` ist jetzt ein String und **keine** Zahl! Das erkennen wir z.B. daran, wenn wir versuchen mit der Variable zu rechnen:

```python
x + x
```

Da `x` ein String ist, ist das Ergebnis nicht die Zahl `4`, sondern ein neuer String, der die Strings `"2"` und `"2"` zusammenfügt.

## Beispiele: ` bool()`


Man kann auch Zahlen in Wahrheitswerte umwandeln. Hier ist die Konvention, dass jede Zahl **außer** der 0, `True` ist:

```python
x = bool(3)
y = bool(2.5)
z = bool(0)

print(x)
print(y)
print(z)
```

```python
print(isinstance(x, bool))
print(isinstance(y, bool))
print(isinstance(z, bool))
```
