### 1. Importiere das Numpy-Paket
**Aufgabe:** Importiere Numpy unter dem Namen `np`.

**Tipp:** Das machen fast alle so. Es spart Tipparbeit und ist Standard.

```python
import numpy as np
```

---

### 2. Erstelle einen Null-Vektor der Größe 10
**Aufgabe:** Erstelle ein Array mit 10 Nullen.

**Tipp:** Verwende die Funktion `np.zeros`.

```python
null_array = np.zeros(10)
null_array
```

---

### 3. Null-Vektor mit einer 1 an Position 5
**Aufgabe:** Erstelle einen Null-Vektor mit 10 Elementen, wobei das 5. Element eine 1 ist.

**Tipp:** Nutze Indexierung, um einzelne Elemente zu ändern.

```python
null_array[4] = 1
null_array
```

---

### 4. Vektor mit Werten von 10 bis 49 erstellen
**Aufgabe:** Erstelle einen Vektor mit Zahlen von 10 bis 49 (einschließlich).

**Tipp:** `np.arange` ist dein Freund.

```python
my_array = np.arange(10, 50)
my_array
```

---

### 5. Vektor umkehren
**Aufgabe:** Drehe die Reihenfolge eines Vektors um.

**Tipp:** Mit Python-Slicing geht das super einfach: `[::-1]`.

```python
my_array[::-1]
```

---

### 6. 3x3-Matrix mit Werten von 0 bis 8 erstellen
**Aufgabe:** Erstelle eine Matrix (3x3) mit Werten von 0 bis 8.

**Tipp:** Nutze `np.arange` und `reshape`.

```python
np.arange(0, 9).reshape(3, 3)
```

---

### 7. 3x3x3-Array mit zufälligen Werten erstellen
**Aufgabe:** Erstelle ein 3x3x3-Array mit zufälligen Zahlen.

**Tipp:** `np.random.random` erstellt Zufallszahlen zwischen 0 und 1.

```python
np.random.random((3, 3, 3))
```

---
### 8. Mittelwert eines Zufallsvektors berechnen
**Aufgabe:** Erstelle einen Zufallsvektor der Größe 30 und berechne dessen Mittelwert.

**Tipp:** `np.mean` hilft dir hier weiter.

```python
np.mean(np.random.random(30))
```
