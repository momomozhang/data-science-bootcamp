## Aufgabe 7: Kombiniere Filterbedingungen
Filtere Mehrfachbestellungen der Produkte mit einem Preis > 50. 

<details>
  <summary>Hinweise anzeigen</summary>
    
#### Hinweise:
- In Pandas können mehrere Bedingungen mit **`&` (UND-Operator)** kombiniert werden.
- Die Bedingung muss in **Klammern** gesetzt werden:
  - **`(df['Spalte'] > Wert) & (df['Spalte'] > Wert)`** für UND-Bedingungen.
  - Falls eine ODER-Bedingung (`|`) gewünscht ist, könnte alternativ:
    - **`(df['Spalte'] > Wert) | (df['Spalte'] > Wert)`** verwendet werden.
- Dieser Filter wählt nur Bestellungen aus, die beide Kriterien gleichzeitig erfüllen.
- Falls keine Einträge gefunden werden:
  - Könnte der **Preisfilter zu hoch** oder die **Mengenbedingung zu streng** sein.
  - Eine **alternative Kombination** von Werten könnte sinnvoll sein.
- Falls fehlende Werte (`NaN`) in einer der Spalten vorkommen, könnten diese mit **`dropna(subset=['Spalte'])`** entfernt werden.
- Falls eine breitere Filterung gewünscht ist, könnten **alternative Preis- oder Mengenbereiche** getestet werden.

    </details>
