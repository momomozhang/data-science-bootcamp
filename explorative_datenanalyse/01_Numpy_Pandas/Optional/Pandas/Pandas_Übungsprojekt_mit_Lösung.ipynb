# Datenbereinigung und Sortierung mit Pandas

## Laden des Datensatzes
Lade den Datensatz und untersuche seine Struktur.

#### Schritte:
1. Importiere die benötigten Bibliotheken.
2. Lade die Datei `Sales_April_2019.csv` mit Pandas.
3. Zeige die ersten 10 Zeilen der Daten an.
4. Überprüfe die Datentypen der Spalten und wandele gegebenfalls um.


```python
# Bibs Laden
import pandas as pd

# Daten in einen DataFrame laden
sales_data = pd.read_csv("Data/Sales_April_2019.csv")
```

```python
# Erste 10 Zeilen anzeigen
sales_data.head(10)
```

```python
# Spalteninformationen
sales_data.info()
```

```python
# Spalten in die richtigen Datentypen umwandeln
sales_data["Quantity Ordered"] = pd.to_numeric(
    sales_data["Quantity Ordered"], errors="coerce"
)  # Umwandlung in numerisch
sales_data["Price Each"] = pd.to_numeric(
    sales_data["Price Each"], errors="coerce"
)  # Umwandlung in numerisch
sales_data["Order Date"] = pd.to_datetime(
    sales_data["Order Date"], errors="coerce"
)  # Umwandlung in Datumsformat

## Warnung ignorieren (erscheint, weil "Order Date" urpsrünglich nicht einheitlich formatiert ist...
## ...wird im nächsten Kapitel behandelt)
## Order ID besteht zwar aus Zahlen, hat aber keien numerische Bedeutung
```

## Aufgabe 1: Entferne alle Zeilen mit fehlenden Werten
Fehlende Werte in irgendeiner Spalte können Analysen verfälschen, daher werden sie entfernt.
Vergleiche die Anzahl der Zeilen vor und nach dem Entfernen und gebe erneut die ersten 10 Zeilen aus.

```python
# Entfernen von Zeilen mit fehlenden Werten
print(f"Anzahl Zeilen vor dem Entfernen von NA`s: {len(sales_data)}")
sales_data.dropna(inplace=True)
print(f"Anzahl der verbleibenden Zeilen nach dem Entfernen: {len(sales_data)}")
```

```python
sales_data.head(10)
```

## Aufgabe 2: Sortiere nach Bestelldatum
Sortiere die Daten chronologisch nach der Spalte "Order Date", um Trends oder Muster leichter erkennen zu können und gebe die ersten 5 und die letzen Zeilen aus.

```python
# Daten nach der Spalte "Order Date" sortieren
sales_data.sort_values(by="Order Date", inplace=True)
```

```python
sales_data.head(5)
```

```python
sales_data.tail(5)
```

## Aufgabe 3: Filtere große Mengen
Filtere alle Einträge, bei denen die Menge größer als 5 ist. Dies ermöglicht eine Analyse von Bestellungen mit hohen Mengen.

```python
# Filtern der Daten, bei denen "Quantity Ordered" > 5 ist
filtered_data = sales_data[sales_data["Quantity Ordered"] > 5]

print("Gefilterte Daten (Menge > 5):")
filtered_data
```

## Aufgabe 4: Sortiere alphabetisch nach Produkt.

```python
# Daten alphabetisch nach der Spalte "Product" sortieren

print("\nDaten sortiert nach Produktnamen:")
sales_data.sort_values(by="Product")
```

## Aufgabe 5: Filtere nach Preisbereich
Filtere alle Einträge, bei denen der Preis zwischen 10 und 100 liegt und sortiere die Daten nach dem Preis. Dies hilft, Bestellungen in einem bestimmten Preisbereich zu analysieren. Welche Produkte liegen in diesem Preisbereich?

```python
# Filtern der Daten, bei denen der Preis zwischen 10 und 100 liegt

price_filtered_data = sales_data[
    (sales_data["Price Each"] >= 10) & (sales_data["Price Each"] <= 100)
]
print("\nGefilterte Daten (Preis zwischen 10 und 100):")
price_filtered_data.sort_values(by="Price Each")  # sortieren
```

```python
# Übersicht der Produkte
unique_products10to100 = price_filtered_data["Product"].unique()
print(unique_products10to100)
```

## Aufgabe 6: Filtere nach Bestelldatum
Filtere alle Bestellungen, die im April 2019 in der ersten Woche getätigt wurden. Dieser Filter ermöglicht die Analyse von Bestellungen in einem bestimmten Zeitraum:


```python
# Filtern der Daten nach Bestellungen, die im April 2019 getätigt wurden

first_week_april = sales_data[
    (sales_data["Order Date"].dt.month == 4) & (sales_data["Order Date"].dt.day <= 7)
]

print("\nGefilterte Daten (1. Aprilwoche 2019):")
first_week_april
```

## Aufgabe 7: Kombiniere Filterbedingungen
Filtere Bestellungen, die in größerer Menge bestellt wurden und einen Preis > 50 enthalten. Dies kombiniert mehrere Bedingungen, um spezifische Bestellungen zu analysieren:

```python
# Filtern der Daten nach Mengenbestellung und Preis > 50
combined_filter_data = sales_data[
    (sales_data["Quantity Ordered"] > 1) & (sales_data["Price Each"] > 50)
]
print("\nGefilterte Daten (Mengenbestellung und Preis > 50):")
combined_filter_data
```
