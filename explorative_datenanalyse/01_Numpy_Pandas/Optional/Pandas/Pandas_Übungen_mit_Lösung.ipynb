In diesem Notebook wollen wir für einen Datensatz (`"Employee-Overview.csv"` - siehe Kursbereich) einige Fragen von Interesse beantworten. Dieser Datensatz enthält die (fiktiven) Mitarbeiterdaten einer (ebenfalls fiktiven) Firma. Benutze im folgenden die Bibliothek `pandas`.

```python
import pandas as pd
```

## 1. Aufgabe: Der Datensatz 

- Lies die csv-Datei `Employee-Overview.csv` in ein `DataFrame` ein. Mache dich mit den Daten vertraut. Beantworte folgende Fragen:

1. Wieviele Mitarbeiter hat das Unternehmen?
2. Wieviele Bereiche (`Dept_Name`) hat das Unternehmen? Wie lauten deren Namen?
3. Wieviele Mitarbeiter arbeiten in den verschiedenen Bereichen?

> <span style="color:blue"> **_INFO:_** </span> Bei letzterer Aufgabe könnte der Befehl [`groupby`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html) helfen.

```python
# Schritt 1: Laden der CSV-Datei in ein DataFrame
# Die CSV-Datei enthält die Mitarbeiterdaten einer fiktiven Firma.
# Sie wird in ein pandas-DataFrame geladen, um sie einfacher analysieren zu können.
file_path = "Data/Employee-Overview.csv"
df = pd.read_csv(file_path)  # CSV-Datei lesen und in ein DataFrame laden
```

```python
df.info()
```

```python
df.head()
```

```python
df.describe(include="all")
```

Die Spalten sind nicht ideal formatiert. Mitarbeiter_ID sollte beispielsweise keien numerische Spalte sein. Eine ordentliche Formatierung ist für eine schnelle Übersicht sehr hilfsvoll.

```python
# Aufgabe 1: Wie viele Mitarbeiter hat das Unternehmen?

total_employees = df["Mitarbeiter_ID"].nunique()  # Anzahl der Zeilen im DataFrame
print(f"Das Unternehmen hat insgesamt {total_employees} Mitarbeiter.")

# oder so:
# Jede Zeile repräsentiert einen Mitarbeiter, daher ist die Anzahl der Zeilen gleich der Anzahl der Mitarbeiter.
total_employees = len(df)
print(f"\nDas Unternehmen hat insgesamt {total_employees} Mitarbeiter.")

# Aufgabe 2: Wie viele Bereiche (Dept_Name) hat das Unternehmen? Wie lauten deren Namen?
# 1. Die Anzahl der eindeutigen Bereiche (Departments) wird mit `nunique()` ermittelt.
# 2. Die Namen der Bereiche werden mit `unique()` abgerufen, das alle einzigartigen Werte einer Spalte zurückgibt.
unique_departments = df["Dept_Name"].nunique()  # Anzahl der eindeutigen Bereiche
department_names = df["Dept_Name"].unique()  # Namen der Bereiche
print(f"\nDas Unternehmen hat {unique_departments} Bereiche.")
print("Die Namen der Bereiche sind:")
for name in department_names:  # Schleife zum Ausgeben der Namen
    print(f"- {name}")

# Aufgabe 3: Wie viele Mitarbeiter arbeiten in den verschiedenen Bereichen?
# Mit der Methode `groupby('Dept_Name')` werden die Daten nach Bereichen gruppiert.
# `size()` zählt die Anzahl der Zeilen (Mitarbeiter) pro Gruppe (Bereich).
employees_per_department = df.groupby("Dept_Name").size()
print("\nAnzahl der Mitarbeiter pro Bereich:")
for department, count in employees_per_department.items():  # Schleife zur Ausgabe der Ergebnisse
    print(f"- {department}: {count} Mitarbeiter")
```

## 2. Aufgabe: Manager & Mitarbeiter

- Wie viele Manager gibt es? Finde dazu die Anzahl von `Führungskräften` in der Spalte `Beschaeftigung`. 
- Stelle die Anzahl an `Führungskräften` und `Mitarbeitern` gegenüber.

```python
# Aufgabe 2: Manager und Mitarbeiter analysieren

# Ziel:
# - Finde die Anzahl der Manager (Führungskräfte) in der Spalte 'Beschaeftigung'.
# - Vergleiche die Anzahl der Führungskräfte mit der Anzahl der Mitarbeiter.

# Schritt 1: Gruppierung der Daten nach 'Beschaeftigung'
# Die Spalte 'Beschaeftigung' enthält Informationen darüber, ob eine Person
# eine Führungskraft ('Führungskraft') oder ein normaler Mitarbeiter ('Mitarbeiter') ist.
# Wir verwenden die Methode `value_counts()`, um die Häufigkeiten zu zählen.
employment_counts = df["Beschaeftigung"].value_counts()

# Schritt 2: Anzahl der Führungskräfte und Mitarbeiter extrahieren
# `value_counts` gibt uns die Anzahl für jede Kategorie. Hier filtern wir die Werte heraus.
num_managers = employment_counts.get("Führungskraft", 0)  # Anzahl der Führungskräfte
num_employees = employment_counts.get("Mitarbeiter", 0)  # Anzahl der Mitarbeiter

# Schritt 3: Ergebnisse anzeigen
print(f"Anzahl der Führungskräfte: {num_managers}")
print(f"Anzahl der Mitarbeiter: {num_employees}")

# Schritt 4: Gegenüberstellung der Anzahl
# Berechnung des Anteils der Führungskräfte an der Gesamtanzahl
total_people = num_managers + num_employees  # Gesamtanzahl der Personen
if total_people > 0:  # Vermeidung einer Division durch Null
    manager_percentage = (num_managers / total_people) * 100
    employee_percentage = (num_employees / total_people) * 100
    print(f"Prozentsatz Führungskräfte: {manager_percentage:.4f}%")
    print(f"Prozentsatz Mitarbeiter: {employee_percentage:.4f}%")
else:
    print("Keine Daten zu Führungskräften oder Mitarbeitern vorhanden.")
```

## 3. Aufgabe: Männer & Frauen

- Stelle das Gehalt der Frauen und Männer gegenüber: Was verdienen Männer im Schnitt? Was Frauen?

```python
if "Geschlecht_Mitarbeiter" in df.columns and "Mitarbeiter_Gehalt" in df.columns:
    df["Mitarbeiter_Gehalt"] = pd.to_numeric(
        df["Mitarbeiter_Gehalt"], errors="coerce"
    )  # Sicherstellen, dass Gehälter numerisch sind

    # Durchschnittsgehalt berechnen
    avg_salary_men = df[df["Geschlecht_Mitarbeiter"] == "M"]["Mitarbeiter_Gehalt"].mean()
    avg_salary_women = df[df["Geschlecht_Mitarbeiter"] == "F"]["Mitarbeiter_Gehalt"].mean()
    print(f"\nDurchschnittsgehalt der Männer: {avg_salary_men:.2f}")
    print(f"Durchschnittsgehalt der Frauen: {avg_salary_women:.2f}")
```

## 4. Aufgabe: Gehaltserhöhung

- Die Spalte `Gehaltserhoehung` gibt an, ob die Person eine Gehaltserhöhung bekommt. Sollte hier "JA" stehen, berechne einen Aufschlag von 10% auf das aktuelle Gehalt. 

> <span style="color:blue"> **_INFO:_** </span> Der Befehl [`apply`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html) könnte hier helfen. Damit kann man eine Funktion auf _jeden_ Eintrag einer Spalte anwenden.

```python
if "Gehaltserhoehung" in df.columns and "Mitarbeiter_Gehalt" in df.columns:
    df["Mitarbeiter_Gehalt"] = pd.to_numeric(
        df["Mitarbeiter_Gehalt"], errors="coerce"
    )  # Sicherstellen, dass Gehälter numerisch sind

    def calculate_raise(row):
        if row["Gehaltserhoehung"] == "JA":
            return row["Mitarbeiter_Gehalt"] * 1.1  # 10% Aufschlag
        return row["Mitarbeiter_Gehalt"]

    df["Bevorstehendes_Gehalt"] = df.apply(calculate_raise, axis=1)
    print("\nGehälter wurden aktualisiert. Hier sind die ersten Einträge:")
    print(df[["Mitarbeiter_Gehalt", "Gehaltserhoehung", "Bevorstehendes_Gehalt"]].head())
```

## 5. Aufgabe: Alter & Beschäftigung

- Wie ist der Name des jüngsten Mitarbeiters des Unternehmens?

- Wer arbeitet am schon am längsten im Unternehmen?



```python
if "Bday_Mitarbeiter" in df.columns:
    df["Bday_Mitarbeiter"] = pd.to_datetime(df["Bday_Mitarbeiter"], errors="coerce")
    jung = df.loc[df["Bday_Mitarbeiter"].idxmax()]
    print(
        f"\nDer jüngste Mitarbeiter ist {jung['MA_Vorname']} {jung['MA_Nachname']}, geboren am {jung['Bday_Mitarbeiter'].date()}."
    )

if "Hire_Date_Mitarbeiter" in df.columns:
    df["Hire_Date_Mitarbeiter"] = pd.to_datetime(df["Hire_Date_Mitarbeiter"], errors="coerce")
    alt = df.loc[df["Hire_Date_Mitarbeiter"].idxmin()]
    print(
        f"\nDer Mitarbeiter mit der längsten Betriebszugehörigkeit ist {alt['MA_Vorname']} {alt['MA_Nachname']}, eingestellt am {alt['Hire_Date_Mitarbeiter'].date()}."
    )
```
