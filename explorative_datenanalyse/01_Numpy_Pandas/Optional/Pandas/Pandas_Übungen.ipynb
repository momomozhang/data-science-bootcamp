## 1. Aufgabe: Der Datensatz 

- Lies die csv-Datei `Employee-Overview.csv` in ein `DataFrame` ein. Mache dich mit den Daten vertraut. Beantworte anschließend folgende Fragen:

### 1: Wie viele Mitarbeiter hat das Unternehmen?

#### Hinweise:
- Jede Zeile im DataFrame repräsentiert einen Mitarbeiter.
- Um die Anzahl der Mitarbeiter zu ermitteln, gibt es zwei Ansätze:
  1. **`nunique()`**: Diese Methode gibt die Anzahl der **eindeutigen Werte** in einer Spalte zurück. Sie kann verwendet werden, wenn es eine Spalte mit **Mitarbeiter-IDs** gibt.
  2. **`len(df)`**: Diese Methode gibt die **Gesamtanzahl der Zeilen** im DataFrame zurück, falls jede Zeile einem einzelnen Mitarbeiter entspricht.

---

### 2: Wie viele Bereiche (Dept_Name) hat das Unternehmen? Wie lauten deren Namen?

#### Hinweise:
- Die Spalte `'Dept_Name'` enthält die Namen der Abteilungen.
- Zwei wichtige Funktionen für diese Aufgabe sind:
  1. **`nunique()`**: Diese Methode zählt die Anzahl der **einzigartigen Werte** in einer Spalte und gibt die Anzahl der unterschiedlichen Bereiche zurück.
  2. **`unique()`**: Diese Methode gibt eine **Liste aller einzigartigen Werte** in der Spalte zurück, also die Namen der vorhandenen Bereiche.

---

### 3: Wie viele Mitarbeiter arbeiten in den verschiedenen Bereichen?

#### Hinweise:
- Um die Anzahl der Mitarbeiter pro Bereich zu berechnen, kann man die Daten nach `'Dept_Name'` gruppieren und die Anzahl der Mitarbeiter in jeder Gruppe zählen.
- Dafür eignet sich die Methode **`groupby()`**:
  1. **`groupby('Dept_Name')`**: Diese Methode gruppiert den DataFrame nach der Spalte `'Dept_Name'`.
  2. **`size()`**: Diese Methode zählt die Anzahl der Einträge in jeder Gruppe, also die Anzahl der Mitarbeiter pro Bereich.

- Das Ergebnis kann in einer **Schleife** ausgegeben werden, um die Abteilungen und die jeweilige Mitarbeiteranzahl übersichtlich darzustellen.

> <span style="color:blue"> **_INFO:_** </span> Der Befehl [`groupby`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html) kann oft helfen.

## 2. Aufgabe: Manager & Mitarbeiter

- Wie viele Manager gibt es? Finde dazu die Anzahl von `Führungskräften` in der Spalte `Beschaeftigung`. 
- Stelle die Anzahl an `Führungskräften` und `Mitarbeitern` gegenüber.

### Schritt 1: Gruppierung der Daten nach 'Beschaeftigung'

#### Hinweise:
- Die Spalte `'Beschaeftigung'` enthält Informationen darüber, ob eine Person eine **Führungskraft** oder ein **normaler Mitarbeiter** ist.
- Um die Anzahl der Personen in jeder Gruppe zu zählen, wird die Methode **`value_counts()`** verwendet.
- Diese Methode gibt eine Übersicht darüber, wie oft jede Kategorie in der Spalte vorkommt.

---

### Schritt 2: Anzahl der Führungskräfte und Mitarbeiter extrahieren

#### Hinweise:
- Die Funktion **`value_counts()`** gibt eine Häufigkeitsverteilung der Werte in einer Spalte zurück.
- Um gezielt Werte abzurufen, kann die Methode **`.get('Kategorie', 0)`** genutzt werden:
  - `'Kategorie'` steht für den gewünschten Wert (z. B. `'Führungskraft'` oder `'Mitarbeiter'`).
  - Falls eine Kategorie nicht existiert, wird mit `0` ein Standardwert zurückgegeben, um Fehler zu vermeiden.

---

### Schritt 3: Gegenüberstellung der Anzahl

#### Hinweise:
- Der **Gesamtanteil** der Führungskräfte und Mitarbeiter wird berechnet.
- Der **prozentuale Anteil** wird mit der Formel:
$$
  \text{Prozentwert} = \left(\frac{\text{Anzahl}}{\text{Gesamtzahl}}\right) \times 100
$$
  berechnet.
- Eine **Überprüfung auf Nullwerte** ist wichtig, um eine Division durch Null zu vermeiden.
- Falls keine Daten vorhanden sind, sollte eine entsprechende Meldung ausgegeben werden.

---

#### Zusammenfassung:
- **`value_counts()`** zählt die Häufigkeiten der Kategorien.
- **`.get('Kategorie', 0)`** ruft Werte ab und stellt sicher, dass kein Fehler auftritt.
- Die Berechnung des **prozentualen Anteils** ermöglicht eine bessere Analyse der Verteilung.

## 4. Aufgabe: Gehaltserhöhung

- Die Spalte `Gehaltserhoehung` gibt an, ob die Person eine Gehaltserhöhung bekommt. Sollte hier "JA" stehen, berechne einen Aufschlag von 10% auf das aktuelle Gehalt. 

> <span style="color:blue"> **_INFO:_** </span> Der Befehl [`apply`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html) könnte hier helfen. Damit kann man eine Funktion auf _jeden_ Eintrag einer Spalte anwenden.

### Ziel:
- Prüfe, welche Personen eine **Gehaltserhöhung** erhalten (Spalte `'Gehaltserhoehung'`).
- Falls die Spalte `'Gehaltserhoehung'` den Wert **"JA"** enthält, berechne einen **Aufschlag von 10%** auf das aktuelle Gehalt.
- Speichere das neue Gehalt in einer zusätzlichen Spalte **`Bevorstehendes_Gehalt`**.

---

### Schritt 1: Sicherstellen, dass die Gehälter numerisch sind

#### Hinweise:
- Falls die Spalte `'Mitarbeiter_Gehalt'` als Text gespeichert ist, muss sie in einen numerischen Typ umgewandelt werden.
- Die Methode **`pd.to_numeric()`** konvertiert Werte in eine numerische Darstellung.
- Der Parameter **`errors='coerce'`** sorgt dafür, dass nicht-numerische Werte in `NaN` umgewandelt werden, um Berechnungsfehler zu vermeiden.

---

### Schritt 2: Berechnung der Gehaltserhöhung mit `apply()`

#### Hinweise:
- Die Methode **`.apply()`** ermöglicht es, eine benutzerdefinierte Funktion auf jede Zeile einer Spalte oder des gesamten DataFrames anzuwenden.
- Die Funktion überprüft:
  - Falls der Wert in `'Gehaltserhoehung'` **"JA"** ist, wird das aktuelle Gehalt **um 10% erhöht**.
  - Falls nicht, bleibt das Gehalt unverändert.
- **Wichtig:** `apply()` muss mit **`axis=1`** verwendet werden, da die Berechnung auf Zeilenebene erfolgt.

---

### Schritt 3: Speichern des neuen Gehalts

#### Hinweise:
- Die berechneten Werte werden in einer neuen Spalte **`Bevorstehendes_Gehalt`** gespeichert.
- Dadurch bleibt das ursprüngliche Gehalt in der Spalte `'Mitarbeiter_Gehalt'` unverändert.
- Eine Vorschau der ersten Werte kann mit **`.head()`** angezeigt werden, um das Ergebnis zu überprüfen.

---

### Schritt 4: Interpretation der Ergebnisse

#### Fragen zur Analyse:
- Wie viele Mitarbeiter erhalten eine Gehaltserhöhung?
- Wie groß ist der durchschnittliche Unterschied zwischen **altem und neuem Gehalt**?
- Gibt es auffällige Muster bei den Erhöhungen, z. B. Abteilungen mit vielen oder wenigen Erhöhungen?

---

### Zusammenfassung:
- **`pd.to_numeric(errors='coerce')`** stellt sicher, dass die Gehälter numerisch sind.
- **`.apply()`** wird verwendet, um eine Funktion auf jede Zeile anzuwenden.
- **Ein bedingter Gehaltsaufschlag von 10%** wird nur für Mitarbeiter mit einer Gehaltserhöhung berechnet.
- Die Ergebnisse werden in einer neuen Spalte gespeichert, ohne das Originalgehalt zu verändern.

## 5. Aufgabe: Alter & Beschäftigung

- Wie ist der Name des jüngsten Mitarbeiters des Unternehmens?

- Wer arbeitet am schon am längsten im Unternehmen?

---

### Schritt 1: Konvertierung der Datumswerte in das richtige Format

#### Hinweise:
- Falls die Spalten **`'Bday_Mitarbeiter'`** (Geburtsdatum) und **`'Hire_Date_Mitarbeiter'`** (Einstellungsdatum) als **Text** gespeichert sind, müssen sie in **Datumswerte** umgewandelt werden.
- Die Methode **`pd.to_datetime()`** konvertiert Text in das Datetime-Format.
- Der Parameter **`errors='coerce'`** sorgt dafür, dass ungültige Datumswerte als `NaT` (Not a Time) gespeichert werden, um Fehler zu vermeiden.

---

### Schritt 2: Jüngsten Mitarbeiter ermitteln

#### Hinweise:
- Der jüngste Mitarbeiter ist derjenige mit dem **höchsten Geburtsdatum**.
- Die Methode **`idxmax()`** gibt den Index des größten Werts einer Spalte zurück.
- Mit **`.loc[]`** kann dann die gesamte Zeile des jüngsten Mitarbeiters abgerufen werden.

---

### Schritt 3: Mitarbeiter mit der längsten Betriebszugehörigkeit ermitteln

#### Hinweise:
- Der Mitarbeiter, der **am längsten** im Unternehmen ist, hat das **älteste Einstellungsdatum**.
- Die Methode **`idxmin()`** gibt den Index des **kleinsten Werts** einer Spalte zurück.
- Mit **`.loc[]`** kann dann die gesamte Zeile des Mitarbeiters mit der längsten Betriebszugehörigkeit abgerufen werden.

---

### Schritt 4: Interpretation der Ergebnisse

#### Fragen zur Analyse:
- Gibt es **Muster** zwischen Alter und Betriebszugehörigkeit?
- Wie **groß ist der Unterschied** zwischen dem jüngsten und dem am längsten beschäftigten Mitarbeiter?
- Sind die **ältesten Mitarbeiter** auch die am längsten beschäftigten?

---

### Zusammenfassung:
- keine Panik: **Zeitreihen** werden wir im Laufe des Moduls noch ausführlicher behandeln, inkl. datetime Funktion.
- **`pd.to_datetime(errors='coerce')`** stellt sicher, dass Datumswerte korrekt verarbeitet werden.
- **`idxmax()`** hilft, den jüngsten Mitarbeiter zu finden.
- **`idxmin()`** bestimmt den am längsten beschäftigten Mitarbeiter.
- Die Ergebnisse können genutzt werden, um Unternehmensdemografie und Erfahrungswerte besser zu verstehen.
