### **1. Erstellen eines Umsatzarrays:**

Erstelle ein NumPy-Array mittels Zufallsgenerator, das die Umsätze (in Euro) der letzten 12 Monate für ein Produkt enthält. Verwende hierzu einen Wertebereich der ganzen Zahlen von **12.000 bis 25.000**.

Berechne den durchschnittlichen Monatsumsatz und finde heraus, welcher Monat den höchsten Umsatz erziehlt hat.

Gib dann noch alle Monate, wo der Umsatz höher war als der ausgerechnete Mittelwert.

<details>
  <summary>Hinweise anzeigen</summary>
    
**Hinweise:**  
✅ **`np.random.seed(42)` setzen**, um bei jeder Ausführung die gleichen Zufallszahlen zu erhalten.  
✅ **Nutze `np.random.randint(start, stop, size)`**, um ein Array mit zufälligen Zahlen zu erzeugen. Die obere Grenze `stop` ist **exklusiv**.  
✅ **Nutze `np.mean()`**, um den **durchschnittlichen Monatsumsatz** zu berechnen.  
✅ **Verwende `np.argmax()`**, um den **Index des höchsten Umsatzes** zu ermitteln. Da NumPy **nullbasiert** indiziert (Monate 0–11), muss **1 addiert** werden, um den Monat korrekt auszugeben.  
✅ **Verwende `np.where()`**, um die **Monate zu filtern**, an denen der Umsatz **größer als der Durchschnitt** war.  
</details>

```python
# Erstellen des NumPy-Arrays mit zufälligen Umsätzen
umsatze = np.random.randint(12000, 25000, size=12)
```

```python
# Berechne den durchschnittlichen Monatsumsatz.
avg_umsatze = umsatze.mean()
print(f"Der durchschnittliche Monatsumsatz ist {round(avg_umsatze, 2)}€.")

# Berechne den Monat mit dem höchsten Umsatz
max_umsatze_monat = np.argmax(umsatze)
print(f"Der Monat mit dem höchsten Umsatz ist {max_umsatze_monat + 1}. Monat des Jahres.")
```

```python
# Gib die Monate mit höher Umsatz als der Duchschnitt aus
# Hinweis -> WHERE liefert ein Tupel zurück
months_above_average = np.where(umsatze > avg_umsatze)[0] + 1
print(f"Monate mit überdurchschnittlichem Umsatz: {months_above_average}")
```

### **2. Verkaufspreisanalyse**  
Ein Händler bietet Produkte in **verschiedenen Preisklassen** an. Erstelle ein Array mit den Verkaufspreisen:  
`[50, 75, 100, 125, 150, 175, 200]`  

Berechne anschließend die **neuen Preise nach einem Rabatt von 10%** und **runde auf die nächsten ganzen Zahlen**.  

<details>
  <summary>Hinweise anzeigen</summary>
    
**Hinweise:**  
✅ **Nutze `np.arange(start, stop, step)`**, um die Preisreihe systematisch zu generieren.  
✅ **Multipliziere die Preise mit `0.9`**, um den **10%-Rabatt** anzuwenden.  
✅ **Nutze `np.round()`**, um die Werte auf die **nächsten ganzen Zahlen** zu runden.  
</details>

---


```python
# Originalpreise

original_preise = np.array([50, 75, 100, 125, 150, 175, 200])
# Preise nach 10% Rabatt
neu_preise = np.round(original_preise * 0.9).astype(int)
print(f"Neue Preise nach 10% Rabatt: {neu_preise}€")
```

### **3. Bestandsanalyse**  
Ein Array speichert die **Lagerbestände verschiedener Artikel**:  
`[100, 200, 150, 80, 300, 400, 250]`  

**Aufgaben:**  
1. Berechne die **Gesamtlagerkosten**, wenn **die Kosten pro Einheit 5 Euro** betragen.  
2. Identifiziere die **Artikel**, deren **Lagerkosten unter 500 Euro** liegen.  


<details>
  <summary>Hinweise anzeigen</summary>
    
**Hinweise:**  
✅ **Nutze `np.sum(array) * unit_cost`**, um die **Gesamtlagerkosten** zu berechnen.  
✅ **Multipliziere das Bestandsarray mit `unit_cost`**, um die Lagerkosten pro Artikel zu erhalten.  
✅ **Verwende `np.where(array < 500)`**, um die **Artikel mit Lagerkosten unter 500 Euro** zu finden.  
✅ **Da NumPy nullbasiert indiziert (Artikel 0–6), sollte `+1` addiert werden, um die tatsächliche Artikelnummer auszugeben.**  

</details>


---



```python
# Stocks und Unit Cost definieren
stocks = np.array([100, 200, 150, 80, 300, 400, 250])
unit_storage_cost = 5

# Total Cost ausrechnen
article_storage_cost = stocks * unit_storage_cost
total_storage_cost = article_storage_cost.sum()

print(f"Der Gesamtlagerkosten ist {total_storage_cost}€.")

# Low Stock Items rausfinden
low_storage_cost_items = np.where(article_storage_cost < 500)[0] + 1
print(f"Artikel mit Lagerkosten unter 500€: Artikel {low_storage_cost_items}")
```

### **4. Preisoptimierung**  
Ein Array speichert die **Verkaufszahlen eines Produkts** bei **unterschiedlichen Preisen**:  

- **Preise:** `[10, 20, 30, 40, 50]`  
- **Verkäufe:** `[1000, 800, 600, 400, 200]`  

**Aufgaben:**  
1. Berechne den **Umsatz für jede Preisstufe**.  
2. Bestimme den **optimalen Preis**, der den **höchsten Umsatz generiert**.  

<details>
  <summary>Hinweise anzeigen</summary>
    
**Hinweise:**  
✅ **Berechne den Umsatz pro Preisstufe** mit `revenues = prices * sales`.  
✅ **Nutze `np.argmax(revenues)`**, um den Index des maximalen Umsatzes zu finden.  
✅ **Verwende `np.max(revenues)`, um den höchsten Umsatz direkt zu erhalten**.  

</details>

---


```python
# Prices und Sales definieren
prices = np.array([10, 20, 30, 40, 50])
projected_sales = np.array([1000, 800, 600, 400, 200])

# Revenues ausrechnen
projected_revenue = prices * projected_sales

# Optimalen Preis und Maximales Revenue ausrechnen
index = np.argmax(projected_revenue)
print(
    f"Der Umsatz ist mit {projected_revenue[index]}€ am höchsten, wenn der Preis auf {prices[index]} festgelegt ist."
)
```
