## Matrizen multiplizieren

Eine häufige Operation, die wir berechnen müssen ist die Multiplikation zweier Matrizen. Siehe z.B. [hier](https://de.wikipedia.org/wiki/Matrizenmultiplikation).

Über den Befehl [`np.dot(X,Y)`](https://numpy.org/doc/stable/reference/generated/numpy.dot.html) können wir zwei Matrizen (Arrays) `X` und `Y` multiplizieren.

**Parameter**:

`X, Y`: Array, bzw. Matrix

**Resultat**:

Die Matrix $\mathbf{X} \cdot \mathbf{Y}$

### Beispiel 1

Wir berechnen das Ergebnis von 

\begin{align}
\begin{pmatrix}
1 & 2 \\
3 & 4 \\
\end{pmatrix}
\cdot 
\begin{pmatrix}
0 & 1 \\
3 & 4 \\
\end{pmatrix}
\end{align}

```python
import numpy as np
```

```python
X = np.array([[1, 2], [3, 4]])

Y = np.array([[0, 1], [3, 4]])

print("Matrix X:\n", X)
print("Matrix Y:\n", Y)

Z = np.dot(X, Y)
print("Matrix X*Y:\n", Z)
```

> <span style="color:red"> **_WICHTIG:_** </span> Beachte, dass die Multiplikation mittels `.dot` etwas anderes ist, als `X*Y`. Bei `X*Y` wird die Multiplikation nur *elementweise* ausgeführt. Das ist etwas anderes als die *mathematische* Multiplikation zweier Matrizen durch `.dot`.

```python
X * Y
```

## Lineare Gleichungen lösen

Auch [lineare Gleichungen](https://de.wikipedia.org/wiki/Lineares_Gleichungssystem) können wir problemlos mit **Numpy** lösen.
Nehmen wir an, wir haben folgendes (lineares) Gleichungssystem gegeben:

\begin{align}
1\cdot x_1 + 2\cdot x_2 & = 5 \\
3 \cdot x_1 + 4 \cdot x_2 & = 11
\end{align}

**Welche Werte muss $x_1$ und $x_2$ annehmen, damit beide Gleichungen erfüllt sind?**<br>
<br>
Wir können solche Gleichungen immer auch als eine Multiplikation zwischen einer Matrix und einem Vektor schreiben:

\begin{align}
\begin{pmatrix}
1 & 2 \\ 
3 & 4 \\
\end{pmatrix}
\cdot
\begin{pmatrix}
x_1 \\
x_2
\end{pmatrix}
&= 
\begin{pmatrix}
5 \\
11
\end{pmatrix}
\\
\mathbf{A} \cdot 
\begin{pmatrix}
x_1 \\
x_2
\end{pmatrix}
&= \mathbf{b}
\end{align} 
In dieser Form können wir die Gleichung lösen mit dem Befehl

[`np.linalg.solve(A,b)`](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html)

### Beispiel 2

Wir wollen das obige Beispiel lösen. Dafür definieren wir als erstes die Matrix $\mathbf{A}$ und den Vektor $\mathbf{b}$:

```python
A = np.array([[1, 2], [3, 4]])

b = np.array([[5], [11]])
```

Nun können wir die Werte der Funktion übergeben und erhalten die Lösungen für $x_1$ und $x_2$:

```python
y = np.linalg.solve(A, b)
print(y)
```

Das bedeutet:
\begin{align}
x_1=1,\quad x_2=2
\end{align}
<br>Und tatsächlich erfüllen diese Werte die obige Gleichung:<br>

\begin{align}
1\cdot 1 + 2\cdot 2 & = 5 \\
3 \cdot 1 + 4 \cdot 2 & = 11
\end{align}

## Matrix invertieren

Die Einheitsmatrix $\mathbf{I}$ ist die Matrix mit nur 1en auf der Diagonale. Z.b. die zweidimensionale Einheitsmatrix:

\begin{align}
\begin{pmatrix}
1 & 0 \\
0 & 1 \\
\end{pmatrix}
\end{align}

Oft stellt sich in der Mathematik folgende Frage:

**Wenn ich eine Matrix $\mathbf{A}$ gegeben habe, mit welche Matrix muss man $\mathbf{A}$ multiplizieren, damit das Ergebnis die Einheitsmatrix ist?**

Also:

\begin{align}
\mathbf{A} \cdot \mathbf{B} = \begin{pmatrix}
1 & 0 \\
0 & 1 \\
\end{pmatrix}
\end{align}

**Was ist $\mathbf{B}?$**

$\mathbf{B}$ ist die *inverse* Matrix von $\mathbf{A}$. `numpy` bietet uns die Funktion [`inv`](https://numpy.org/doc/stable/reference/generated/numpy.linalg.inv.html) um die inverse Matrix einer Matrix zu berechnen.

Diese Funktion erhält als Argument eine Matrix $\mathbf{A}$ und liefert die Inverse Matrix von $\mathbf{A}$ zurück, so dass gilt: 

\begin{align}
\mathbf{A} \cdot inv(\mathbf{A}) = \begin{pmatrix}
1 & 0 \\
0 & 1 \\
\end{pmatrix}
\end{align}

### Beispiel 3

Wir wollen die inverse Matrix von 
\begin{align}
A= \begin{pmatrix}
1 & 2 \\
3 & 4 \\
\end{pmatrix}
\end{align}
berechnen:

```python
# A definieren
A = np.array([[1, 2], [3, 4]])
```

```python
# Die Funktion inv befindet sich im Modul linalg (LINeare ALGebra)
from numpy.linalg import inv
```

```python
# B ist die inverse Matrix von A
B = inv(A)
B
```

**Es gilt nun**

\begin{align}
\mathbf{A} \cdot inv(\mathbf{A}) =
\mathbf{A} \cdot \mathbf{B} =
\begin{pmatrix}
1 & 2 \\
3 & 4 \\
\end{pmatrix} 
\cdot 
\begin{pmatrix}
-2 & 1 \\
1.5 & -0.5 \\
\end{pmatrix}
=
\begin{pmatrix}
1 & 0 \\
0 & 1 \\
\end{pmatrix}
\end{align}


```python
A.dot(B)
```

Der untere linke Wert bedeutet 
\begin{align}
8.8817842 \cdot 10^{-16}
\end{align}
ist also praktisch $0$. Dies liegt an der endlichen Genauigkeit von Python, dass hier nicht *exakt* $0$ als Ergebnis erscheint!

## Diagonalelemente und Spur

Zwei weitere Befehle die häufig von Nutzen sein können, sind [`diag`](https://numpy.org/doc/stable/reference/generated/numpy.diag.html) und [`trace`](https://numpy.org/doc/stable/reference/generated/numpy.trace.html).

Der erste Befehl liefert uns die Diagonalelemente einer Matrix zurück und der zweite Befehl die Summe selbiger.


```python
from numpy import diag, trace

A = np.array([[1, 2], [3, 4]])

print("Die Matrix A")
print(A)

print("Die Elemente auf der Diagonale:")
print(diag(A))

print("Die Summe der Diagonalelemente:")
print(trace(A))
```
