## 2. Erstellung von Datenstrukturen

### 2.1 Erstellung einer Series
Eine **Series** ist wie eine Liste, aber mit einem Index. Damit kannst du spezifische Elemente viel einfacher ansprechen.

````python
data = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])
print(data)
````

Eine **`Series`** ist eine **1-dimensionale Datenstruktur** aus der Pandas-Bibliothek, die aus zwei Komponenten besteht:

- **Werte (Values):** z. B. Zahlen, Strings, Datumsangaben, etc.
- **Index (Labels):** jeder Wert hat ein Label zur eindeutigen Identifikation

👉 Du kannst sie dir vorstellen wie eine „smarte Liste mit Namen“.

#### ✅ Vorteile gegenüber Python-Listen

| Feature                        | Python Liste (`list`)       | Pandas Series                  |
|-------------------------------|------------------------------|--------------------------------|
| **Indexierung**               | Nur automatisch (0, 1, 2, …) | Beliebig (z. B. Namen, Daten)  |
| **Datenanalyse**              | Eingeschränkt                | Voller Funktionsumfang         |
| **Mathematische Operationen** | Nur mit Schleifen            | Vektorisiert (sehr schnell!)   |
| **Kompatibilität mit Pandas** | Nein                         | Ja – Series = DataFrame-Spalte |

##### 🔍 Beispiel: Mathematische Operation

````python
liste = [1, 2, 3]
serie = pd.Series([1, 2, 3])

# Liste braucht eine Schleife:
[x + 1 for x in liste]  # [2, 3, 4]

# Series rechnet direkt:
serie + 1  # [2, 3, 4]
````

#### Vergleich: Series vs. DataFrame

| Merkmal           | Series                         | DataFrame                                 |
|------------------|--------------------------------|-------------------------------------------|
| Dimension         | 1D                             | 2D (mehrere Spalten)                      |
| Datenstruktur     | Werte + Index                  | Spaltennamen + Zeilenindex                |
| Aufbau            | Wie eine Spalte                | Wie eine Tabelle                          |
| Nutzung           | Einzelne Datenreihe            | Mehrere Datenreihen vergleichen/analyisieren |
| Zugriff           | `serie[index]`                 | `df["spalte"]`, `df.loc[row]`, etc.       |

##### 🔍 Beispiel:


# Series

````python
temperatur = pd.Series([22, 24, 21], index=["Mo", "Di", "Mi"])
display(temperatur)
````

# DataFrame

````python
wetter = pd.DataFrame({"Temperatur": [22, 24, 21], "Regen": [0, 5, 10]}, index=["Mo", "Di", "Mi"])
````

#### Zusammenfassung

| Du brauchst…                                       | Nutze…          |
|----------------------------------------------------|-----------------|
| Eine einfache Liste von Werten                    | `list`          |
| Werte mit eindeutigen Labels (z. B. Zeit, Namen)   | `pd.Series`     |
| Eine Tabelle mit mehreren Spalten & Zeilen         | `pd.DataFrame`  |
| Mathematische Berechnungen ohne Schleifen         | `pd.Series`     |
| Datenanalyse, Statistik, Visualisierung            | `pd.Series` / `pd.DataFrame` |

#### Statistik & Analyse mit Series

````python
serie = pd.Series([10, 20, 30, 40, 50])

print(serie.mean())  # Durchschnitt
print(serie.std())  # Standardabweichung
print(serie.describe())  # Überblick über die Verteilung
````

## 📊 Schritt 3: Zusammenfassende Statistiken

### 🟢 `df.describe()` – Statistischer Überblick

````python
df.describe()
````

````python
### Der folgende Code muss an dieser Stelle nicht gelernt/nachvollzogen werden. Es geht hier nur um das Ergebnis (Plot).

import matplotlib.pyplot as plt
import numpy as np

# Beispiel-Daten
np.random.seed(0)
daten = np.random.normal(loc=37, scale=1.5, size=40)
daten.sort()

# Quartile berechnen
q1 = np.percentile(daten, 25)
q2 = np.percentile(daten, 50)
q3 = np.percentile(daten, 75)

# Farben je Quartil
farben = []
for wert in daten:
    if wert <= q1:
        farben.append("skyblue")
    elif wert <= q2:
        farben.append("dodgerblue")
    elif wert <= q3:
        farben.append("lightgreen")
    else:
        farben.append("green")

# Leichtes vertikales Jittern, identisch für Punkt & Zahl
np.random.seed(42)
jitter = np.random.uniform(0, 0.1, size=len(daten))
y = 1 + jitter

# Plot
plt.figure(figsize=(16, 6))
plt.scatter(daten, y, c=farben, s=200, alpha=0.8, edgecolor="black")

# Nummerierung – folgt exakt dem Jitter
for i, (x, y_val) in enumerate(zip(daten, y)):
    plt.text(x, y_val + 0.003, str(i + 1), ha="center", va="bottom", fontsize=8)

# Quartil-Grenzen einzeichnen
plt.axvline(q1, color="dodgerblue", linestyle="--", label="Q1 (25%)")
plt.axvline(q2, color="red", linestyle="--", label="Median (Q2 / 50%)")
plt.axvline(q3, color="green", linestyle="--", label="Q3 (75%)")

plt.title("Quartile")
plt.xlabel("Messwert")
plt.yticks([])
plt.legend()
plt.tight_layout()
plt.show()
````

➡️ Liefert dir viele wichtige **statistische Kennzahlen** für **numerische Spalten**:

| Kennzahl        | Bedeutung                                         |
|-----------------|----------------------------------------------------|
| `count`         | Anzahl der Werte (z. B. 9 Zeilen)                  |
| `mean`          | Mittelwert (Durchschnitt)                          |
| `std`           | Standardabweichung (wie stark streuen die Werte?)  |
| `min` / `max`   | Kleinster / größter Wert                           |
| `25%`, `50%`, `75%` | Quartile – helfen dir, Verteilungen zu verstehen |

🧠 Praktisch bei Datensätzen mit **vielen Zahlen** (Preise, Bewertungen, Mengen etc.)

## 4. Auswahl und Indexierung von Daten

### 4.1 Auswahl von Spalten

````python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 40],
    "City": ["New York", "Los Angeles", "Chicago", "Houston"],
}
df = pd.DataFrame(data)
print(df)

print(df["Name"])  # Einzelne Spalte
print(df[["Name", "City"]])  # Mehrere Spalten
````

**Erklärung:**
- Mit `df['Name']` greifst du auf die Spalte "Name" zu.
- Willst du mehrere Spalten sehen, übergib eine Liste der Spaltennamen in doppelten eckigen Klammern.

### 4.2 Auswahl von Zeilen

````python
print(df.iloc[0])  # Zugriff über Position
print(df.loc[0])  # Zugriff über Index
````

Die Wahl zwischen iloc und loc hängt davon ab, wie du auf deine Daten zugreifen möchtest:
- Bearbeiten oder Anzeigen basierend auf Position: Verwende iloc.
- Bearbeiten oder Anzeigen basierend auf einem Label oder benutzerdefinierten Index: Verwende loc.

Unterschied zwischen iloc und loc
- Position (iloc): Greift unabhängig vom Index auf die Zeilenposition zu.
Beispiel: Funktioniert auch, wenn der Index des DataFrames angepasst wurde.
- Index (loc): Greift basierend auf dem Index-Label zu. Beispiel: Wenn der Index geändert wurde.

**Motivation:** Ob du spezifische Zeilen bearbeiten oder nur anzeigen willst – Pandas bietet dir dazu einfache Werkzeuge.

### 4.3 Boolean Indexing

````python
print(df[df["Age"] > 30])
````

### 🐼 Pandas Tutorial: Unterschiedliche Zugriffsformen in DataFrames

#### 🧾 Überblick

Beim Arbeiten mit Pandas-DataFrames gibt es **verschiedene Möglichkeiten**, auf Daten zuzugreifen. Die wichtigsten sind:

| Zugriffsmethode | Beschreibung                      |
|------------------|-----------------------------------|
| `df[]`           | Zugriff auf Spalten               |
| `df.loc[]`       | Label-basiert (Index & Spaltenname) |
| `df.iloc[]`      | Positionsbasiert (Zeilen/Spaltennummer) |
| `df.at[]`        | Zugriff auf einen einzelnen Wert (Label) |
| `df.iat[]`       | Zugriff auf einen einzelnen Wert (Position) |
| Boolean Indexing | Zugriff über Bedingungen (Filter) |

---

## 📦 Beispiel-DataFrame

````python
daten = {
    "Name": ["Anna", "Ben", "Carla", "David"],
    "Alter": [25, 30, 22, 35],
    "Stadt": ["Berlin", "Hamburg", "München", "Köln"],
}

df = pd.DataFrame(daten, index=["a", "b", "c", "d"])
print(df)
````

#### 🔹 `df["Spaltenname"]` → Zugriff auf Spalten

````python
df["Name"]  # Gibt eine Series zurück
````

````python
df[["Name", "Stadt"]]  # Gibt ein DataFrame zurück
````

#### 🔸 `loc[]` → **Label-basiert** (Indexnamen & Spaltennamen)

````python
# Eine Zeile per Label
df.loc["a"]
````

````python
# Mehrere Zeilen
df.loc[["a", "c"]]
````

````python
# Zeilen & Spalten
df.loc["a", "Name"]  # Einzelwert
````

````python
df.loc[["a", "b"], ["Name", "Stadt"]]  # Teil-DataFrame
# Alle Zeilen, bestimmte Spalte
````

#### 🔸 `iloc[]` → **Positionsbasiert** (wie bei Arrays: 0, 1, 2, …)

````python
# Erste Zeile
df.iloc[0]
````

````python
# Erste 2 Zeilen
df.iloc[0:2]
````

````python
# Wert in 1. Zeile, 2. Spalte
df.iloc[0, 1]
````

````python
# Zeilen 0 und 2, Spalten 0 und 2
df.iloc[[0, 2], [0, 2]]
````

#### 🔹 `at[]` und `iat[]` → Schneller Zugriff auf **einzelne Zelle**

- `at[]`: über **Label**
- `iat[]`: über **Position**

````python
df.at["a", "Name"]  # "Anna"
````

````python
df.iat[0, 0]  # ebenfalls "Anna"
````

➡️ Vorteil: **schneller als `loc[]` / `iloc[]`**, wenn du nur **einen einzigen Wert** brauchst.


#### 🔍 Boolean Indexing → Zugriff über Bedingungen

````python
df[df["Alter"] > 25]  # Filtert Zeilen
````

````python
df[df["Name"].str.startswith("C")]  # Namen mit C
````

````python
df.loc[df["Stadt"] == "Berlin", "Name"]  # Nur Namen in Berlin
````

#### 🧠 Zusammenfassung der Zugriffsformen

| Methode      | Typ                 | Beispiel                                | Beschreibung                         |
|--------------|---------------------|-----------------------------------------|--------------------------------------|
| `df["Spalte"]` | Spalte             | `df["Name"]`                             | Zugriff auf eine Spalte (Series)     |
| `df[["S1", "S2"]]` | Spalten (mehrere) | `df[["Name", "Stadt"]]`                  | Teil-DataFrame mit bestimmten Spalten |
| `df.loc[row, col]` | Label-basiert    | `df.loc["a", "Name"]`                    | Zugriff über Zeilen- und Spaltenlabel |
| `df.iloc[row, col]`| Positionsbasiert | `df.iloc[0, 0]`                          | Zugriff über Zahlenpositionen        |
| `df.at[row, col]`  | Label-basiert    | `df.at["a", "Name"]`                     | Sehr schnell für einzelne Zelle      |
| `df.iat[row, col]` | Positionsbasiert | `df.iat[0, 0]`                           | Sehr schnell für einzelne Zelle      |
| `df[bedingung]`    | Boolean Indexing | `df[df["Alter"] > 30]`                   | Zugriff auf gefilterte Zeilen        |


**Erklärung:**
- Hier werden nur die Zeilen angezeigt, bei denen das Alter größer als 30 ist. Stell dir vor, du filterst eine Excel-Tabelle mit Bedingungen.
