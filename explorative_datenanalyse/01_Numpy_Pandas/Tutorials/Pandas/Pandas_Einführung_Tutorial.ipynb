### **Grundlagen von Pandas**

#### **1. Einführung**
Pandas ist eine Python-Bibliothek, die das Arbeiten mit tabellarischen Daten (ähnlich wie Excel) erleichtert. Die beiden Haupt-Datenstrukturen in Pandas sind:
- `Series`: Eine eindimensionale Liste mit Indizes.
- `DataFrame`: Eine tabellenähnliche Struktur mit Zeilen und Spalten.

#### **3. Importieren von Pandas**


```python
import pandas as pd
```

---

#### **4. DataFrame Erstellen**

##### **4.1. DataFrame aus einem Dictionary**


```python
# Beispiel: Erstellen eines DataFrames
data = {
    "Name": ["Anna", "Ben", "Charlie", "Diana"],
    "Alter": [23, 34, 29, 42],
    "Beruf": ["Ingenieur", "Lehrer", "Arzt", "Architekt"],
}

df = pd.DataFrame(data)
print(df)
```

---

#### **5. Datenzugriff und Bearbeitung**

##### **5.1. Zugriff auf Spalten**

```python
# Zugriff auf eine Spalte
print(df["Name"])

# Mehrere Spalten auswählen
print(df[["Name", "Beruf"]])
```

##### **5.2. Zugriff auf Zeilen**

```python
# Zugriff mit loc (basierend auf Labels)
print(df.loc[2, "Name"])  # Erste Zeile, Name,

# Zugriff mit iloc (basierend auf Indexpositionen)
print(df.iloc[1])  # Zweite Zeile
```

##### **5.3. Hinzufügen neuer Spalten**

```python
# Neue Spalte hinzufügen
df["Gehalt"] = [50000, 60000, 75000, 80000]
print(df)
```

##### **5.4. Werte ändern**

```python
# Einen Wert ändern
df.at[0, "Name"] = "Anne"
print(df)
```

---

#### **6. Filtern von Daten**

##### **6.1. Filtern nach Bedingungen**


```python
# Personen, die älter als 30 Jahre sind
print(df[df["Alter"] > 30])

# Personen mit einem bestimmten Beruf
print(df[df["Beruf"] == "Arzt"])
```

##### **6.2. Mehrere Bedingungen**

```python
# Älter als 30 und Gehalt über 60.000
filtered_df = df[(df["Alter"] > 30) & (df["Gehalt"] > 60000)]
print(filtered_df)
```

##### **6.3. Spalten filtern**

```python
# Nur die Namen und Gehälter der gefilterten Personen anzeigen
print(filtered_df[["Name", "Gehalt"]])
```
