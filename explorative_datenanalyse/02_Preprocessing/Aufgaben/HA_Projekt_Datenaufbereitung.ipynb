# Kaffeebohnen Projekt

Willkommen zu unserem Data-Science-Projekt! Heute nehmen wir dich mit auf eine unterhaltsame Reise durch die chaotische, aber liebenswerte Welt der **Kaffeebohnen GmbH**. Dieses familiengeführte Unternehmen ist bekannt für seinen hervorragenden Kaffee – aber weniger für seine exzellente Datenverwaltung.

### Die manuelle Datenaufnahme bei der Kaffeebohnen GmbH

Bei der Kaffeebohnen GmbH ist die Datenpflege Familiensache. **Oma Erna**, das Herz des Unternehmens, kümmert sich persönlich um die Aufzeichnung aller Verkaufsdaten. Bewaffnet mit ihrem treuen alten Laptop (Baujahr 1999) und einer Tasse ihres Lieblingskaffees tippt sie fleißig jede Bestellung in Excel ein.

Doch Oma Erna ist eine Frau der Kreativität:

- **Spaltennamen nach Lust und Laune**: Was letztes Jahr noch "Produkt" hieß, nennt sich dieses Jahr "Kaffeesorte" und nächstes Jahr vielleicht "Artikelbezeichnung".
- **Individuelle Formate**: Datumsangaben variieren zwischen "31.12.2020", "2020/12/31" und "31-12-20". Manchmal schreibt sie das Datum sogar als "Silvester 2020".
- **Tippfehler und künstlerische Freiheit**: Aus "Espresso" wird schnell mal "Expresso" oder "Espreso". Und wer könnte ihr verdenken, dass sie "Cappuccino" als "Kapuccino" schreibt?

Natürlich meint es Oma Erna nur gut, aber ihre einzigartige Art der Dateneingabe stellt uns vor einige Herausforderungen.


### Unsere Mission

In diesem Projekt werden wir:

- Die Daten der letzten Jahre aus verschiedenen Excel-Dateien laden.
- Die unterschiedlichen Spaltennamen vereinheitlichen.
- Tippfehler bereinigen und Namen standardisieren.
- Datenformate korrigieren und konsistent machen.
- Fehlende Werte und Duplikate behandeln.


Mach dich bereit für eine spannende Datenrettungsaktion! Mit Python als unserem Werkzeug und einem Sinn für Humor werden wir Ordnung in das Datenchaos der Kaffeebohnen GmbH bringen.

---

**Disclaimer:** Dieses Projekt ist rein fiktiv, und die verwendeten Daten sind vollständig erfunden. Dennoch sind die dargestellten Probleme und Herausforderungen typisch und treten in der Praxis häufig auf... und vergiss nicht, zwischendurch eine Tasse Kaffee zu genießen – das hat bei Oma Erna schließlich auch immer geholfen!

## Daten laden

#### 1. Lese *Kaffeebohnen_2019.xlsx* korrekt ein

Öffne die Excel-Dateien zuerst aus dem Ordner *KaffeeDaten* manuell in Excel, um Folgendes zu überprüfen:

- Datenstruktur: Gibt es einheitliche Muster in den Tabellen?
- Header: Gibt es mehrere Headerzeilen oder eine korrekte Kopfzeile?
- Bereich: Liegen die relevanten Daten in einem spezifischen Bereich?

Informiere dich über die Parameter der Funktion `read_excel()` und wähle gezielt passende Einstellungen aus, um den relevanten Datenbereich effizient zu laden.

- Lade mit `pandas` alle drei Dateien in den DataFrames `df19`, `df20` und `df21`.

- Stelle sicher, dass bei allen DataFrames die Header-Zeile stimmt (z.B. mit `skiprows` oder `header`) und die gleiche Anzahl von Spalten geladen wird (z.B. mit `usecols`).

#### 2. Tabellen zusammenführen

Alle Excel-Dateien sind leider nicht einheitlich aufgebaut: Die Tabelle enthalten genau 6 Spalten, die aber unterschiedlich benannt sind, aber den gleichen Inhalt haben. 

- Verwende einheitliche Spaltennamen für alle drei DataFrames.

- Führe alle Excel-Dateien in einem einzigen DataFrame mit `pd.concat` zusammen. 

- Validiere, dass die Anzahl von Zeilen nach dem `pd.concat` auch passt.

Der Vorteil der Zusammenführung ist, dass weitere Bereinigungsmaßnahmen anschließend nur einmalig durchgeführt werden müssten.

#### 3. Verschaffe dir eine Übersicht über die Daten

- Gebe eine kompakte Zusammenfassung der Struktur des DataFrames, einschließlich Spaltennamen, Datentypen und Anzahl der nicht-leeren Werte

- Liefere grundlegende statistische Kennzahlen und erste Einblicke in die Verteilung der Daten zu erhalten

## Datenbereinigung

#### 4. Entfernen von leeren Zeilen

Wir entfernen zuerst ausschließlich die Zeilen, die in **allen** Spalten fehlende Werte enthalten, um sicherzustellen, dass teilweise ausgefüllte Datensätze erhalten bleiben.

Überprüfe nun erneut die `pd.info()`, um sicher zu sein, dass jetzt keine leeren Zeilen mehr im DataFrame enthalten sind.

#### 5. Anzeigen von Zeilen ohne Wert

Der nächste Schritt ist, die Zeilen anzuzeigen, in denen eine Zelle keinen Wert enthält. Dies hilft uns, potenzielle fehlende Informationen zu identifizieren, bevor wir entscheiden, wie wir damit umgehen. Mache dir Gedanken darüber, ob solche Zeilen entfernt werden sollten. 

In diesem Fall sehen wir, dass von den 222 Zeilen in der Spalte `Kunden_ID` nur 211 davon *non-null* sind. Zeige diese 11 Zeilen auf der Konsole.

#### 6. Überprüfung der Spalten auf Tippfehler und Korrektur

Jetzt geht es ans Eingemachte. Um mögliche Tippfehler in den Daten aufzuspüren und zu beheben, solltest du die folgenden Schritte durchführen:

**Identifikation einzigartiger Werte:** 

Zeige zunächst alle eindeutigen Werte der Spalte `Produkt` an. Dadurch erhältst du einen Überblick über die vorhandenen Einträge und erkennst mögliche Tippfehler oder Unstimmigkeiten. Rechne gerne auch die Häufigkeit dieser Werte mit `value_counts` aus.

Prüfe die einzigartigen Einträge auf Unregelmäßigkeiten, wie z. B. falsche Schreibweisen, unvollständige Begriffe oder ungewöhnliche Sonderzeichen. 

Entwickele eine Strategie, um diese Fehler zu korrigieren, z.B. durch Ersetzen häufiger Tippfehler, Harmonisierung von Synonymen oder Anpassung an einen definierten Standard.

**Spalten Typ anpassen**

- Zeige erstmal die `value_counts` für die Spalten `Preis`, `Menge` und `Datum`.

- In der Spalte `Preis` musst du noch das `,` mit einem `.` ersetzen. Dafür kannst du die Syntax `df['Preis'].astype(str).str.replace(",", ".")` nutzen.

- Konvertiere die Spalten ins passende Datenformat mit `to_numeric` und `to_datetime`.

- Nutze den Parameter `errors='coerse'`, um fehlerhafte Werte mit `NaN` zu erstetzen.

- Zeige am Ende die Zeilen für das Datum, wo man das nicht automatisch in einem einheitlichen Format bringen kann.

  
**Duplikate**

Zeige erstmal alle Duplikate auf der Konsole, bevor du diese mit `drop_duplicates` entfernst.

**Übersicht über die Daten**:

Du hast die Datenquelle erfolgreich bereinigt. Gebe nun wieder eine kompakte Zusammenfassung der Struktur des DataFrames, einschließlich Spaltennamen, Datentypen und Anzahl der nicht-leeren Werte und liefere grundlegende statistische Kennzahlen.

**Identifikation von Ausreißern:** 

Überlege, ob einige ungewöhnliche Einträge tatsächlich valide sind oder ob es sich um Fehler handelt. Vergleiche z.B. mit der `describe` Tabelle, welche Werte deutlich über den 75%-Wert liegen.

**Bereinigte Tabelle extrahieren**:

Speichere die bereinigte Tabelle als Excel-Datei mit `to_excel` wieder ab.
