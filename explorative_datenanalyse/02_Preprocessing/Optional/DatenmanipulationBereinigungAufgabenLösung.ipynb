### **Aufgabe 1: Fehlende Werte behandeln**
Gegeben ist ein DataFrame `df`:

````python
import numpy as np
import pandas as pd

data = {
    "Name": ["Anna", "Ben", np.nan, "Diana", "Ella"],
    "Alter": [25, np.nan, 30, 29, 22],
    "Stadt": ["Berlin", "Hamburg", "Berlin", "München", np.nan],
}
df = pd.DataFrame(data)
````

**Aufgabe**:
1. Fülle die fehlenden Werte in der Spalte `Name` mit `"Unbekannt"`.
2. Ersetze die fehlenden Werte in der Spalte `Alter` durch den Mittelwert der Spalte.
3. Fülle die fehlenden Werte in der Spalte `Stadt` mit `"Keine Angabe"`.

---

**Lösung**:

```python

````python
# Fehlende Werte in der Spalte "Name" füllen
df["Name"] = df["Name"].fillna("Unbekannt")

# Fehlende Werte in der Spalte "Alter" mit dem Mittelwert ersetzen
df["Alter"] = df["Alter"].fillna(df["Alter"].mean())

# Fehlende Werte in der Spalte "Stadt" füllen
df["Stadt"] = df["Stadt"].fillna("Keine Angabe")

display(df)
````

### **Aufgabe 2: Strings bereinigen**
Gegeben ist ein DataFrame `df`:


````python
data = {
    "Produkt": [" Apple ", "BaNaNa  ", "  Grape", "OranGe ", "apple"],
    "Preis": [1.2, 0.8, 2.5, 1.5, 1.2],
}
df = pd.DataFrame(data)
````

**Aufgabe**:
1. Entferne führende und nachfolgende Leerzeichen aus der Spalte `Produkt`.
2. Setze alle Werte in der Spalte `Produkt` auf Kleinbuchstaben.
3. Entferne doppelte Einträge in der Spalte `Produkt`.

---

**Lösung**:


````python
# 1. Entfernen führender und nachfolgender Leerzeichen
df["Produkt"] = df["Produkt"].str.strip()

# 2. Umwandeln aller Produktnamen in Kleinbuchstaben
df["Produkt"] = df["Produkt"].str.lower()

# 3. Entfernen von Duplikaten basierend auf der Spalte "Produkt"
df = df.drop_duplicates(subset="Produkt")

# Ergebnis anzeigen
display(df)
````

### **Aufgabe 3: Regex-Anwendung**
Gegeben ist ein DataFrame `df`:

````python
data = {
    "Text": [
        "Bestellung: #12345, Kunde: Max",
        "Bestellung: #67890, Kunde: Anna",
        "Kunde: Ben, Bestellung: #54321",
    ]
}
df = pd.DataFrame(data)
````

**Aufgabe**:
1. Extrahiere die Bestellnummer (z. B. `#12345`) und speichere sie in einer neuen Spalte `Bestellnummer`.
2. Extrahiere den Namen des Kunden und speichere ihn in einer neuen Spalte `Kunde`.

---

**Lösung**:

````python
# 1. Extrahieren der Bestellnummer mit einem Regex
df["Bestellnummer"] = df["Text"].str.extract(r"(#\d+)")  # Regex: '#' gefolgt von Ziffern

# 2. Extrahieren des Kundennamens mit einem Regex
df["Kunde"] = df["Text"].str.extract(
    r"Kunde:\s*(\w+)"
)  # Regex: 'Kunde:' gefolgt von einem oder mehreren Buchstaben

# Ergebnis anzeigen
print(df)
````

### **Aufgabe 4: Werte aufteilen**
Gegeben ist ein DataFrame `df`:


````python
data = {
    "Person": ["Max Mustermann", "Anna Schmidt", "Lisa Müller"],
    "Kontakt": ["max@gmail.com", "anna@outlook.com", "lisa@yahoo.com"],
}
df = pd.DataFrame(data)
````


**Aufgabe**:
1. Teile die Spalte `Person` in zwei Spalten: `Vorname` und `Nachname`.
2. Extrahiere die Domain der E-Mail-Adresse (z. B. `gmail.com`) in eine neue Spalte `Domain`.

---

**Lösung**:


````python
# 1. Aufteilen der Spalte "Person" in "Vorname" und "Nachname", Leerzeichen als Trennzeichen
df[["Vorname", "Nachname"]] = df["Person"].str.split(" ", expand=True)

# 2. Extrahieren der Domain aus der E-Mail-Adresse
df["Domain"] = df["Kontakt"].str.extract(
    r"@([\w.]+)"
)  # Regex: '@' gefolgt von Wortzeichen oder Punkten

# Ergebnis anzeigen
display(df)
````

### **Aufgabe 5: Bedingte Manipulation**
Gegeben ist ein DataFrame `df`:

````python
data = {
    "Name": ["Anna", "Ben", "Chris", "Diana", "Ella"],
    "Alter": [25, 17, 35, 29, 22],
    "Punkte": [80, 45, 95, 70, 88],
}
df = pd.DataFrame(data)
````

**Aufgabe**:
1. Füge eine neue Spalte `Kategorie` hinzu, basierend auf dem Alter:
   - `Kind`, wenn Alter < 18
   - `Erwachsen`, wenn Alter >= 18
2. Aktualisiere die Spalte `Punkte`, indem du bei Erwachsenen 10 Punkte addierst.

---

**Lösung**:


````python
# 1. Kategorie basierend auf Alter hinzufügen
df["Kategorie"] = df["Alter"].apply(
    lambda x: "Kind" if x < 18 else "Erwachsen"
)  # Bedingte Zuordnung
# apply(): Wendet eine Funktion auf jede Zeile oder Spalte an.
# Lambda-Funktionen sind anonyme (kurze) Funktionen, die oft für eine einmalige Anwendung genutzt werden.

# Alternativ:
# def kategorisieren(x):
#    if x < 18:
#        return "Kind"
#    else:
#        return "Erwachsen"

# df["Kategorie"] = df["Alter"].apply(kategorisieren)

# 2. Punkte für Erwachsene um 10 erhöhen
df.loc[df["Kategorie"] == "Erwachsen", "Punkte"] += 10  # Nur für "Erwachsen" 10 Punkte hinzufügen
# df.loc[bedingung, spalte]:
# Filtert Zeilen basierend auf einer Bedingung und ändert nur die angegebenen Werte.

# Ergebnis anzeigen
display(df)
````
