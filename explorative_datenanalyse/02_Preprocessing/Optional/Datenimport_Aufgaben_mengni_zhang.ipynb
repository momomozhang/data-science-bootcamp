## **Übung: Dateien aus einem Verzeichnis laden**

### Aufgaben:
1. Lade alle CSV-Dateien aus dem Verzeichnis *temp_data*.
2. Kombiniere die Dateien in einem DataFrame.
3. Füge eine Spalte hinzu, die den Namen der Ursprungsdatei angibt.

```python
import glob
import os
from typing import Optional

import pandas as pd
```

```python
sub_folder = "temp_data/"
csv_files = glob.glob(f"{sub_folder}*.csv")
```

```python
dfs = []

for file in csv_files:
    df = pd.read_csv(file)
    df["source_file"] = os.path.basename(file)
    dfs.append(df)

combined_df = pd.concat(dfs, ignore_index=True)
```

```python
def load_multiple_csv_with_source(folder_name: str) -> Optional[pd.DataFrame]:
    csv_files = glob.glob(f"{folder_name}/*.csv")
    dfs = []

    for file in csv_files:
        df = pd.read_csv(file)
        df["source_file"] = os.path.basename(file)
        dfs.append(df)

    combined_df = pd.concat(dfs, ignore_index=True)
    return combined_df
```
