## Aufgabensammlung zum Thema Datenaufbereitung

### Aufgabe 1: Daten kombinieren (Verkäufe aus Januar und Februar)

#### Ziel:
Du hast zwei Datensätze, die Verkaufsdaten aus Januar und Februar enthalten (*sales_january.csv* und *sales_february.csv*). 
Kombiniere diese Daten zu einem einzigen Datensatz und füge eine neue Spalte hinzu, 
die den Gesamtumsatz aus beiden Monaten berechnet.
Da die Verkaufsdaten eine einfache Struktur haben (Produkt Spalte ist identisch), brauchst du sie lediglich spaltenweise zu verketten.

```python
import pandas as pd

# Datensätze laden
sales_january = pd.read_csv("sales/sales_january.csv")
sales_february = pd.read_csv("sales/sales_february.csv")

# Datensätze anzeigen
print("Verkaufsdaten Januar: ")
display(sales_january)

print("Verkaufsdaten Februar: ")
display(sales_february)

# Daten kombinieren
sales_combined = pd.concat([sales_january, sales_february], axis=1)

# Gesamtumsatz berechnen
sales_combined["Gesamtumsatz"] = sales_combined["Umsatz_Jan"] + sales_combined["Umsatz_Feb"]

# Ergebnis anzeigen
print("Verkaufsdaten kombiniert: ")

display(sales_combined)
```

Eine allgemeine (bessere) Lösung wäre eher das "Verknüpfen" der Datensätze über die Produkt-Spalte mittels `merge()`. Denn so haben wir die Spalte Produkt im neuen Datensatz doppelt...

### Aufgabe 2: Kunden und Bestellungen verknüpfen

#### Ziel:
Verknüpfe die beiden Datensätze *customers.csv* und *orders.csv*, um eine Übersicht zu erstellen, 
welche Kunden welche Produkte bestellt haben.


```python
# Datensätze laden
customers = pd.read_csv("sales/customers.csv")
orders = pd.read_csv("sales/orders.csv")

# Datensätze anzeigen
print("Kunden: ")
display(customers)

print("Bestellungen: ")
display(orders)

# Verknüpfung der Datensätze
customer_orders = pd.merge(customers, orders, on="CustomerID", how="left")

print("Kunden mit Bestellung: ")
# Ergebnis anzeigen
display(customer_orders)
```

### Aufgabe 3: Umsatzdaten vergleichen

#### Ziel:
Kombiniere die Umsatzdaten aus zwei Filialen (store_revenue CSVs), sodass die monatlichen Umsätze nebeneinander dargestellt werden. 
Berechne außerdem die Differenz zwischen den Umsätzen der beiden Filialen.
Tipp: Verwende `index_col=0` Parameter beim Laden der CSVs, um anschließend `join()`zu verwenden.


```python
# Datensätze laden
store1_revenue = pd.read_csv("sales/store1_revenue.csv", index_col=0)
store2_revenue = pd.read_csv("sales/store2_revenue.csv", index_col=0)

print("Filialen einzeln: ")
# Ergebnis anzeigen
display(store1_revenue)
display(store2_revenue)


# Daten kombinieren
combined_revenue = store1_revenue.join(store2_revenue, lsuffix="_Store1", rsuffix="_Store2")

# Differenz berechnen
combined_revenue["Store2_Difference_from_Store1"] = (
    combined_revenue["Revenue_Store1"] - combined_revenue["Revenue_Store2"]
)

# Ergebnis anzeigen
display(combined_revenue)
```
