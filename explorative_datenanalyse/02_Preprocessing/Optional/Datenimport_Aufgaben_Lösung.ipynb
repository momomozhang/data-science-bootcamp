## **Übung: Dateien aus einem Verzeichnis laden**

### Aufgaben:
1. Lade alle CSV-Dateien aus dem Verzeichnis *temp_data*.
2. Kombiniere die Dateien in einem DataFrame.
3. Füge eine Spalte hinzu, die den Namen der Ursprungsdatei angibt.

**Lösung:**


1. **Import der benötigten Bibliotheken**  


```python
import os

import pandas as pd
```

   - `os` wird genutzt, um mit Verzeichnissen und Dateien im Dateisystem zu arbeiten (z. B. Dateien auflisten, Pfade zusammenfügen, etc.).  

2. **Pfadangabe**  


```python
directory = "temp_data"
```

   - Hier wird das Verzeichnis angegeben, in dem sich alle gewünschten CSV-Dateien befinden. In diesem Beispiel heißt der Ordner `"temp_data"`.

3. **Auflisten aller CSV-Dateien im Ordner**  

```python
csv_files = [f for f in os.listdir(directory) if f.endswith(".csv")]
```

   - `os.listdir(directory)` listet alle Dateien und Ordner im Verzeichnis `directory` auf.  
   - In der anschließenden Listenkomprehension wird geprüft, ob der Dateiname auf `'.csv'` endet. Nur solche Dateien werden in die Liste `csv_files` aufgenommen.  
   - Ergebnis: `csv_files` enthält alle CSV-Dateien, die im Ordner `temp_data` zu finden sind.

4. **Initialisierung eines leeren DataFrames**  

```python
combined_data = pd.DataFrame()
```

   - Ein leerer DataFrame `combined_data` wird erstellt. In diesen werden später schrittweise die Daten aller CSV-Dateien eingefügt.

5. **Schleife über alle CSV-Dateien zum Einlesen und Zusammenführen**  

```python
for file in csv_files:
    file_path = os.path.join(directory, file)
    df = pd.read_csv(file_path)
    df["Source_File"] = file  # Ursprungsdateinamen hinzufügen
    combined_data = pd.concat([combined_data, df], ignore_index=True)
```

   - **Schritt für Schritt:**
     1. **`for file in csv_files`:**  
        Die Schleife durchläuft die Liste aller CSV-Dateien (`csv_files`), wobei jede Datei nacheinander in der Variablen `file` gehalten wird.
     2. **`file_path = os.path.join(directory, file)`:**  
        Hier wird der komplette Pfad zur CSV-Datei zusammengebaut. `os.path.join` setzt den Ordnernamen `directory` und den Dateinamen `file` zu einem Pfad zusammen, der für das verwendete Betriebssystem richtig formatiert ist.
     3. **`df = pd.read_csv(file_path)`:**  
        Mit `pandas.read_csv()` wird die entsprechende CSV-Datei eingelesen und als DataFrame `df` gespeichert.
     4. **`df['Source_File'] = file`:**  
        Eine neue Spalte namens `Source_File` wird erstellt und darin wird der Dateiname (`file`) gespeichert. So kann später nachvollzogen werden, aus welcher CSV-Datei jede Zeile stammt.
     5. **`combined_data = pd.concat([combined_data, df], ignore_index=True)`:**  
        Die DataFrames werden miteinander verbunden (konkateniert).  
        - Durch `[combined_data, df]` wird eine Liste mit dem schon vorhandenen kombinierten DataFrame und dem aktuellen DataFrame erstellt.  
        - `pd.concat(...)` führt alle Zeilen zusammen.  
        - `ignore_index=True` sorgt dafür, dass das Indexieren der Zeilen neu geschrieben wird. Anstatt die alten Indizes der einzelnen Dateien beizubehalten, bekommen alle Zeilen aufsteigende Indizes ab 0.

6. **Ausgabe des kombinierten DataFrames**  

```python
display(combined_data)
```

   - Zum Schluss wird der fertig kombinierte DataFrame auf der Konsole ausgegeben, sodass man sehen kann, wie die Daten aller CSV-Dateien zusammengefasst wurden und aus welcher Datei jede Zeile stammt.

---

**Zusammenfassung**  
- Das Skript durchsucht den Ordner `temp_data` nach allen CSV-Dateien.  
- Jede dieser Dateien wird eingelesen und in einen Pandas-DataFrame umgewandelt.  
- Um später zu erkennen, welche Zeilen aus welcher Datei stammen, wird eine Spalte `Source_File` angelegt, in der der jeweilige Dateiname abgelegt wird.  
- Anschließend werden alle einzelnen DataFrames mit `pd.concat()` zu einem einzigen großen DataFrame (`combined_data`) zusammengeführt.  
- Zuletzt wird dieser kombinierte DataFrame ausgegeben.
