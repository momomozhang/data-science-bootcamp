## **Einführung in Regex mit Pandas**

### **Was ist Regex?**
Reguläre Ausdrücke (Regex) sind Muster, die verwendet werden, um bestimmte Zeichenfolgen in Texten zu finden, zu extrahieren oder zu manipulieren. Sie sind extrem nützlich in der Datenanalyse, insbesondere beim Bereinigen und Transformieren von Textdaten.

---

### **1. Grundlegende Regex-Syntax**
Bevor wir Regex in Pandas verwenden, schauen wir uns die wichtigsten Bausteine an:

| Regex-Muster | Bedeutung                                                                 |
|--------------|---------------------------------------------------------------------------|
| `.`          | Beliebiges Zeichen außer Zeilenumbruch                                   |
| `\d`         | Eine Ziffer (0-9)                                                        |
| `\D`         | Kein Ziffernzeichen                                                      |
| `\w`         | Ein alphanumerisches Zeichen (a-z, A-Z, 0-9, _)                          |
| `\W`         | Kein alphanumerisches Zeichen                                            |
| `\s`         | Ein Leerzeichen (Leerzeichen, Tab, etc.)                                 |
| `\S`         | Kein Leerzeichen                                                        |
| `^`          | Start der Zeichenkette                                                  |
| `$`          | Ende der Zeichenkette                                                   |
| `[]`         | Zeichenklasse (z. B. `[aeiou]` sucht nach einem Vokal)                  |
| `()`         | Gruppierung für Muster oder Extraktion                                  |
| `*`          | 0 oder mehr Wiederholungen                                              |
| `+`          | 1 oder mehr Wiederholungen                                              |
| `?`          | 0 oder 1 Wiederholung                                                   |
| `{n}`        | Genau n Wiederholungen                                                  |
| `{n,m}`      | Zwischen n und m Wiederholungen                                          |
| `\`          | Escape-Zeichen (z. B. `\.` sucht nach einem Punkt)                      |

---

### **2. Daten vorbereiten**

Wir erstellen ein Beispiel-DataFrame, mit dem wir verschiedene Regex-Operationen demonstrieren können:

```python

````python
import pandas as pd

data = {
    "Name": ["Alice Wonderland", "Bob Marley", "Charlie Brown", "Dora Explorer"],
    "Email": [
        "alice@wonderland.com",
        "bob.marley@gmail.com",
        "charlie.brown@yahoo.com",
        "dora.explorer@adventure.com",
    ],
    "Phone": ["123-456-7890", "987.654.3210", "555-666-7777", "(123) 456-7890"],
    "Comment": ["Order #12345", "Query #54321", "Complaint #98765", "Feedback #67890"],
}

df = pd.DataFrame(data)
display(df)
````

---

### **3. Praktische Anwendung von Regex mit Pandas**

#### **3.1. Suche nach Text mit `str.contains()`**
`str.contains()` verwendet ein Regex-Muster, um Text in einer Spalte zu suchen. Beispiel: Finde alle E-Mails mit der Domain "gmail.com".

````python
filtered_df = df[df["Email"].str.contains(r"@gmail.com", na=False)]
display(filtered_df)
````

### **Regex-Erklärung**:

Das Präfix `r` in Python steht für **"raw string"** (roher String). Es wird verwendet, um Escape-Zeichen (`\`) in Strings korrekt zu interpretieren, ohne sie speziell behandeln zu müssen.

##### **Warum wird `r` benötigt?**
In regulären Ausdrücken (`Regex`) werden viele Zeichen mit einem Backslash (`\`) verwendet, wie z. B.:
- `\d` (für eine Ziffer)
- `\w` (für ein alphanumerisches Zeichen)
- `\.` (für einen Punkt)

Wenn du in Python einen normalen String ohne `r` verwendest, musst du den Backslash doppelt schreiben (`\\`), da der Backslash im normalen String als Escape-Zeichen interpretiert wird. Das führt zu mehr Schreibaufwand und macht den Code schwerer lesbar.


**Kurz gesagt:** Das `r` erleichtert die Arbeit mit Regex und vermeidet unnötiges doppeltes Escaping. 

---


#### **3.2. Extrahieren von Text mit `str.extract()`**
Extrahiere Muster wie die Bestellnummer (`#12345`) aus der Spalte `Comment`.

````python
df["Order_ID"] = df["Comment"].str.extract(r"#(\d+)")
display(df)
````

- **Regex-Erklärung**:
  - `#` sucht das `#`-Symbol.
  - `(\d+)`: 
    - `\d` steht für eine Ziffer (0-9).
    - `+` bedeutet "eine oder mehr Wiederholungen".
    - Die Klammern `()` gruppieren das Muster, um es zu extrahieren.

#### **3.3. Ersetzen von Text mit `str.replace()`**
Ersetze Punkte oder Klammern in Telefonnummern durch Bindestriche.

```python

````python
df["Phone_Cleaned"] = df["Phone"].str.replace(r"[()]", "", regex=True)
df["Phone_Cleaned"] = df["Phone_Cleaned"].str.replace(r"[.\s]", "-", regex=True)
display(df)
````

- **Regex-Erklärung**:
  - `[()]`: 
    - `()` sucht nach Klammern.
    - Die Zeichen sind in eckigen Klammern gruppiert, was bedeutet, dass eines davon ersetzt wird.
  - `[.\s()]`: 
    - `.` sucht nach einem Punkt.
    - `\s` sucht nach Leerzeichen (oder Tabs).
  - `-` ist der Ersatzwert.

### **4. Kombination von Regex und Pandas**

#### **4.1. Benutzerdefinierte Funktion mit Regex**
Kategorisiere E-Mails basierend auf ihrer Domain.

````python
def categorize_email(email):
    if pd.isna(email):
        return "Unknown"
    if "gmail.com" in email:
        return "Personal"
    elif "yahoo.com" in email:
        return "Personal"
    else:
        return "Work"


df["Email_Category"] = df["Email"].apply(categorize_email)
display(df)
````
