## 2. Mehrere Dateien aus einem Verzeichnis laden

Hast du ein Verzeichnis voller Excel- oder CSV-Dateien, die du alle gleichzeitig verarbeiten möchtest? Kein Problem! Mit Python und Pandas kannst du diese Dateien effizient in einem Rutsch einlesen. Lass uns das Schritt für Schritt durchgehen. 🚀

### **Warum mehrere Dateien laden?**
Manchmal liegen Daten in mehreren Dateien vor, beispielsweise monatliche Berichte oder verschiedene Datensätze einer Kampagne. Anstatt jede Datei einzeln zu laden, können wir alle Dateien im Verzeichnis automatisch einlesen und kombinieren.

### **1. Verzeichnis mit Dateien durchsuchen**
Wir verwenden das `os`-Modul, um alle Dateien im Verzeichnis zu finden.

```python
import os

# Pfad zum Verzeichnis mit den Excel-Dateien
verzeichnis_pfad = "sales_data"

# Liste aller Dateien im Verzeichnis
csv_daten = [
    f for f in os.listdir(verzeichnis_pfad) if f.endswith(".csv")
]  # nur Dateien mit der Endung .csv

print(csv_daten)
```

### **2. Dateien mit Pandas laden und kombinieren**
Jetzt können wir die Dateien nacheinander laden und in einem großen DataFrame kombinieren.

```python
# Leerer DataFrame für die gesammelten Daten
sales2019 = pd.DataFrame()

# Dateien einzeln laden und kombinieren
for datei in csv_daten:
    datei_pfad = os.path.join(verzeichnis_pfad, datei)
    df = pd.read_csv(datei_pfad)  # Excel-Datei laden
    sales2019 = pd.concat([sales2019, df], ignore_index=True)  # Daten hinzufügen
```
