## Datenbereinigung

```python
import pandas as pd
```

```python
# Daten laden
file_path = "sales_data/dirtyData.csv"
df = pd.read_csv(file_path)
print("Originaldaten:")
display(df)
```

### Spaltennamen vereinheitlichen
Aufgabe: Vereinheitliche die Spaltennamen (z.B. klein schreiben, Leerzeichen entfernen)

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_").str.replace("-", "_")

print("\nSpaltennamen vereinheitlicht:")

display(df)
```

### Spaltennamen gezielt ändern


```python
df = df.rename(columns={"e_mail_adresse": "email"})

# Überprüfen, ob der Spaltenname geändert wurde
print(df.columns)
```

### Datenübersicht

Wir schauen uns die Spaltenformate und Anzahl der Einträge an

```python
df.info()
```

```python
# Datenübersicht mittels display
display(df.describe(include="all"))
```

### Fehlende Werte 
Aufgabe: Identifiziere fehlende Werte

```python
print("\nFehlende Werte pro Spalte:")
print(df.isnull().sum())
```

```python
display(df)
```

```python
# Alterswerte behandeln
df["alter"] = df["alter"].fillna(df["alter"].median())  # Median verwenden
```

```python
display(df)
```

### Zahlen korrigieren

Kaufbeträge korrigieren: Punkte und Kommas vereinheitlichen


```python
df["kaufbetrag"] = df["kaufbetrag"].astype(str).str.replace(",", ".")
display(df)
```

```python
# Tausender Punkt entfernen mit Regex
df["kaufbetrag"] = df["kaufbetrag"].str.replace(r"(?<=\d)\.(?=\d{3})", "", regex=True)

# Spalte in numerische umwandeln
df["kaufbetrag"] = pd.to_numeric(df["kaufbetrag"], errors="coerce")
display(df)
```

### String-Funktionen anwenden

```python
# Aufgabe: Bereinige und vereinheitliche den Text
df["name"] = df["name"].str.strip().str.capitalize()
df["wohnort"] = df["wohnort"].str.strip().str.capitalize()
display(df)
```

### Duplikate entfernen
Aufgabe: Überprüfe und entferne doppelte Datensätze

```python
print(
    "\nAnzahl Duplikate vor Entfernung:",
    df.duplicated(subset=["name", "kaufbetrag", "registrierungsdatum"]).sum(),
)
df = df.drop_duplicates(subset=["name", "kaufbetrag", "registrierungsdatum"])
print(
    "Anzahl Duplikate nach Entfernung:",
    df.duplicated(subset=["name", "kaufbetrag", "registrierungsdatum"]).sum(),
)
```

### Datumskorrekturen
Aufgabe: Konvertiere das Datum ins richtige Format

```python
df["registrierungsdatum"] = (
    df["registrierungsdatum"].str.replace("-", ".").str.replace(",", ".").str.replace("/", ".")
)
```

```python
# Funktion zum Bereinigen der Datumsformate
def parse_date(date):
    formats = ["%Y.%m.%d", "%d.%m.%Y", "%m.%Y"]  # Liste möglicher Formate
    for fmt in formats:
        try:
            return pd.to_datetime(date, format=fmt)
        except ValueError:
            continue
    return pd.NaT  # Wenn kein Format passt, NaT zurückgeben


# Anwenden der Funktion auf die Spalte
df["registrierungsdatum"] = df["registrierungsdatum"].apply(parse_date)

display(df)
```

```python
df["email"] = df["email"].str.replace("(at)", "@")
display(df)
```

```python
### Spalte korrigieren mit einem Dictionary

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Abteilung": ["HR", "Financ", "IT", "HR", "Fin"],
}

df = pd.DataFrame(data)
display(df)
```

```python
# Korrektur-Dictionary
korrektur_dict = {
    "Financ": "Finance",
    "Fin": "Finance",
    "HR": "Human Resources",
    "IT": "Information Technology",
}

# Korrigieren der Spalte 'Abteilung' mit replace
df["Korrigierte_Abteilung"] = df["Abteilung"].replace(korrektur_dict)

# Ausgabe des korrigierten DataFrames
display(df)
```

Die Methode `.str.replace()` unterscheidet sich von `.replace()` in Pandas, da sie auf Zeichenkettenoperationen (Strings) spezialisiert ist. Sie wird verwendet, um **Teile von Zeichenketten** innerhalb einer Spalte (oder Serie) zu ersetzen, während `.replace()` zum **Ersetzen ganzer Werte** (egal welcher Datentyp) im DataFrame oder in einer Spalte dient.

Hier sind die Hauptunterschiede:

---

### **1. `.replace()`**
- Funktioniert auf dem gesamten DataFrame oder einer Serie.
- Kann ganze Werte (nicht nur Strings) ersetzen.
- Unterstützt die Verwendung von Dictionaries oder Listen.
- Eignet sich für das Ersetzen von **kompletten Übereinstimmungen**.

**Beispiel mit `.replace()`:**

```python
df = pd.DataFrame({"Abteilung": ["HR", "Finance", "Financ", "IT"]})

# Ersetzen kompletter Werte
df["Abteilung_korrigiert"] = df["Abteilung"].replace(
    {"Financ": "Finance", "IT": "Information Technology"}
)
print(df)
```

---

### **2. `.str.replace()`**
- Funktioniert nur auf textbasierten Daten (Strings).
- Ersetzt Teile von Zeichenketten basierend auf einem Suchmuster (Reguläre Ausdrücke sind möglich).
- Eignet sich für das **teilweise Ersetzen** von Zeichen in Strings.

**Beispiel mit `.str.replace()`:**

```python
df = pd.DataFrame({"Abteilung": ["HR", "Finance", "Financ", "IT"]})

# Ersetzen von Teilen der Zeichenkette
df["Abteilung_korrigiert"] = df["Abteilung"].str.replace("Financ", "Finance")
print(df)
```

Hier wird **nur "Financ" als Teil des Strings ersetzt**, ohne dass andere Werte geändert werden.

---

### **Vergleich anhand eines Beispiels**

**Datensatz:**

```python
data = {"Abteilung": ["HR", "Financ", "Finanzen", "IT"]}
df = pd.DataFrame(data)
```

1. **Mit `.replace()`:**


```python
df["Abteilung_korrigiert"] = df["Abteilung"].replace({"Financ": "Finance"})
```

- Ersetzt nur komplette Übereinstimmungen (`"Financ" → "Finance"`).
- `"Finanzen"` bleibt unverändert.

2. **Mit `.str.replace()`:**

```python
df["Abteilung_korrigiert"] = df["Abteilung"].str.replace("Financ", "Finance")
```

- Ersetzt **Teile von Zeichenketten**, also auch `"Finanzen"` wird angepasst.

---

### Fazit:
- Verwende `.replace()` für **komplette Werteersetzungen**.
- Verwende `.str.replace()` für **Teilersetzungen** innerhalb von Zeichenketten.
