# Date String Manipulations mit Python

In diesem Tutorial lernst du, wie du Datumsangaben in Python als Strings verarbeiten kannst. Wir verwenden dazu die `datetime`-Bibliothek und zeigen einfache Beispiele zur Umwandlung, Formatierung und Berechnung von Datumswerten.

## 1. Ein einfaches Datum als String formatieren



```python
from datetime import datetime

# Aktuelles Datum und Uhrzeit abrufen
jetzt = datetime.now()
jetzt
```

```python
# In einen lesbaren String umwandeln
datum_str = jetzt.strftime("%d.%m.%Y %H:%M:%S")
print("Formatiertes Datum:", datum_str)
```

**Erklärung:**
- `datetime.now()` holt das aktuelle Datum und die aktuelle Uhrzeit.
- `strftime()` wird genutzt, um das Datum in einen String zu verwandeln.
- Die Platzhalter `%d`, `%m`, `%Y` stehen für Tag, Monat und Jahr.

## 2. Einen Datum-String in ein `datetime`-Objekt umwandeln

```python
# Beispiel: Ein Datum als String
datum_str = "15.03.2025 14:30:00"

# Umwandlung in ein datetime-Objekt
datum_obj = datetime.strptime(datum_str, "%d.%m.%Y %H:%M:%S")
print("Datetime-Objekt:", datum_obj)
```

**Erklärung:**
- `strptime()` wird verwendet, um einen String in ein `datetime`-Objekt umzuwandeln.
- Die gleiche Formatierung wie zuvor (`%d.%m.%Y %H:%M:%S`) wird genutzt.

## 3. Datumsberechnungen durchführen

```python
from datetime import timedelta

# Ein Datum um eine Woche erhöhen
neues_datum = datum_obj + timedelta(days=7)
print("Neues Datum in einer Woche:", neues_datum.strftime("%d.%m.%Y %H:%M:%S"))
```

**Erklärung:**
- `timedelta(days=7)` erzeugt eine Zeitspanne von 7 Tagen.
- Durch Addition wird das ursprüngliche Datum um eine Woche verschoben.

## 4. Unterschied zwischen zwei Daten berechnen

```python
# Ein weiteres Datum
datum_alt = datetime(2024, 3, 1)

# Differenz berechnen
differenz = datum_obj - datum_alt
print("Tage zwischen den Daten:", differenz.days)
```

**Erklärung:**
- Die Differenz zweier `datetime`-Objekte liefert eine `timedelta`-Angabe.
- `.days` gibt die Anzahl der Tage zwischen den beiden Daten zurück.

## 5. Aktuelles Datum als String in verschiedene Formate umwandeln

```python
print("ISO-Format:", jetzt.isoformat())  # Standardformat
print("US-Format:", jetzt.strftime("%m/%d/%Y"))  # MM/TT/JJJJ
print("Deutsches Format:", jetzt.strftime("%d.%m.%y"))  # TT.MM.JJJJ
```

**Erklärung:**
- `isoformat()` gibt das Datum in der internationalen ISO-Formatierung aus.
- Mit `strftime()` können verschiedene Formate erstellt werden.

## 6. Datum in Quartal, Woche und Wochentag umwandeln

```python
import pandas as pd

# Quartal berechnen (einfachere Methode mit pandas)
quartal = pd.Timestamp(jetzt).quarter
print("Quartal:", quartal)

# Kalenderwoche berechnen
kalenderwoche = jetzt.strftime("%U")
print("Kalenderwoche:", kalenderwoche)

# Wochentag berechnen (0 = Montag, 6 = Sonntag)
wochentag = jetzt.weekday()
print("Wochentag (0=Montag, 6=Sonntag):", wochentag)

# Wochentag als Namen anzeigen
wochentag_name = jetzt.strftime("%A")
print("Wochentag (Name):", wochentag_name)
```

Hier ist eine Übersicht der gängigsten `strftime`-Funktionen in Python, die für die Formatierung von Datum und Uhrzeit verwendet werden:

| Platzhalter | Beschreibung                     | Beispiel (`datetime(2024, 3, 6, 14, 30, 45)`) |
|-------------|---------------------------------|---------------------------------------------|
| `%Y`        | Jahr (vierstellig)              | `2024`                                     |
| `%y`        | Jahr (zweistellig)              | `24`                                       |
| `%m`        | Monat (01–12)                   | `03`                                       |
| `%B`        | Monat (vollständig)             | `März`                                     |
| `%b`        | Monat (abgekürzt)               | `Mär`                                      |
| `%d`        | Tag (01–31)                     | `06`                                       |
| `%A`        | Wochentag (vollständig)         | `Mittwoch`                                 |
| `%a`        | Wochentag (abgekürzt)           | `Mi`                                       |
| `%w`        | Wochentag (0=Sonntag, 6=Samstag)| `3`                                        |
| `%j`        | Tag des Jahres (001–366)        | `066`                                      |
| `%U`        | Kalenderwoche (Sonntag als Start, 00–53) | `09`                              |
| `%W`        | Kalenderwoche (Montag als Start, 00–53)  | `10`                              |
| `%H`        | Stunde (00–23)                  | `14`                                       |
| `%I`        | Stunde (01–12)                  | `02`                                       |
| `%p`        | AM oder PM                      | `PM`                                       |
| `%M`        | Minute (00–59)                  | `30`                                       |
| `%S`        | Sekunde (00–59)                 | `45`                                       |
| `%f`        | Mikrosekunden (000000–999999)   | `000000`                                   |
| `%z`        | Zeitzonen-Offset                | `+0100` (z. B. MEZ)                        |
| `%Z`        | Zeitzonen-Name                  | `CET` (z. B. Mitteleuropäische Zeit)       |
| `%c`        | Datum und Zeit                  | `Mi 06 Mär 2024 14:30:45`                  |
| `%x`        | Datum                           | `06.03.2024` (formatabhängig)              |
| `%X`        | Uhrzeit                         | `14:30:45` (formatabhängig)                |


## 7. Weitere Timestamp-Manipulationen mit `pandas`

```python
# Aktuelles Datum als Pandas Timestamp
timestamp = pd.Timestamp.now()
print("Pandas Timestamp jetzt:", timestamp)

# Timestamp in einen String umwandeln
print("Timestamp als String:", timestamp.strftime("%Y-%m-%d %H:%M:%S"))

# Timestamp in Unix-Zeitstempel umwandeln
unix_time = timestamp.timestamp()
print("Unix-Timestamp:", unix_time)

# Timestamp aus Unix-Zeitstempel erstellen
neuer_timestamp = pd.to_datetime(unix_time, unit='s')
print("Wiederhergestellter Timestamp:", neuer_timestamp)

# Anfang des Tags berechnen
tag_start = timestamp.normalize()
print("0 Uhr:", tag_start)

# Weitere Zeitangaben aus einem Timestamp extrahieren
jahr = timestamp.year
monat = timestamp.month
woche = timestamp.isocalendar()[1]
tag_im_jahr = timestamp.day_of_year
stunde = timestamp.hour
minute = timestamp.minute
sekunde = timestamp.second

print("Jahr:", jahr)
print("Monat:", monat)
print("Woche im Jahr:", woche)
print("Tag im Jahr:", tag_im_jahr)
print("Stunde:", stunde)
print("Minute:", minute)
print("Sekunde:", sekunde)
```

**Erklärung:**
- `.year`, `.month`, `.day_of_year`, `.hour`, `.minute` und `.second` extrahieren verschiedene Teile des Datums.
- `timestamp.isocalendar()[1]` gibt die Kalenderwoche zurück.

## Fazit
Mit Python kannst du Datumswerte leicht als Strings verarbeiten. Du kannst sie formatieren, umwandeln und damit rechnen. Nutze `datetime`, `timedelta` und `pandas.Timestamp`, um mit Datum-Strings effizient zu arbeiten!
