## ❓ Forschungsfrage 4: **Gibt es Ausreißer – und was sagen sie uns?**

### ▶️ IQR-Methode anwenden

```python
q1 = df["body_mass_g"].quantile(0.25)
q3 = df["body_mass_g"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df_outliers = df[(df["body_mass_g"] < lower_bound) | (df["body_mass_g"] > upper_bound)]
df_outliers
```

Es gibt insgesamt betrachtet keine Ausreißer. Kann man Ausreißer innerhalb einer Gruppe identifizieren?

```python
# Liste zum Sammeln der Ausreißer-Zeilen
outlier_rows = []

# Gruppieren nach Art und Geschlecht (inkl. "Unknown")
grouped = df.dropna(subset=["body_mass_g"]).groupby(["species", "sex"])

for (species, sex), group in grouped:
    q1 = group["body_mass_g"].quantile(0.25)
    q3 = group["body_mass_g"].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = group[(group["body_mass_g"] < lower) | (group["body_mass_g"] > upper)]
    outlier_rows.append(outliers)

# Alle Ausreißer in einem DataFrame zusammenfassen
df_outliers_grouped = pd.concat(outlier_rows)
df_outliers_grouped
```

## ❓ Forschungsfrage 5: **Welche Merkmale hängen miteinander zusammen?**

### ▶️ Korrelation berechnen (numerisch)
Diese kann zwischen -1 und 1 liegen, wobei ein hoher bzw. niedriger Wert auf einen Zusammenhang hindeutet, und ein Wert nahe 0 auf keinen.
Dies sind aber keine Kausalitäten!

```python
df.corr(numeric_only=True)
```

🔍 **Antwort:**  
- **Flipperlänge und Körpergewicht** haben eine starke positive Korrelation (~0.87) → größere Flossen = schwerere Pinguine.
- Schnabellänge korreliert ebenfalls leicht mit Körpergewicht.
