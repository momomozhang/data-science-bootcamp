# Normalverteilung – Einführung

Tipp: Die Bibliotheken zur Visualisierung lernen wir erst nächste Woche kennen. Den Code brauchst du also noch nicht zu verstehen und soll erstmal nur demonstrieren, wie sich Werte in einer Normalverteilung verteilen.

```python
# Notwendige Bibliotheken importieren ---

import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
#  Plotten der Wahrscheinlichkeitsdichtefunktion (PDF) der Normalverteilung ---

# Erzeugen eines Wertebereichs für x
x = np.linspace(140, 210, 1000)

# Berechnen der PDF-Werte
pdf = norm.pdf(x, loc=175, scale=10)

# Plotten der PDF
plt.figure()
plt.plot(x, pdf, label="PDF der Normalverteilung")
plt.title("Wahrscheinlichkeitsdichtefunktion (PDF) der Normalverteilung")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.show()
```

```python
# Stichprobe aus der Standardnormalverteilung ziehen und Histogramm erstellen ---

# Wir ziehen 10.000 Zufallswerte aus der Normalverteilung.
samples = np.random.normal(
    loc=175, scale=10, size=1000
)  # Gerne für size andere Zahlen ausprobieren

# Erstellen eines Histogramms
plt.figure()
plt.hist(samples, bins=50, density=True, alpha=0.6, label="Stichprobe (Histogramm)")

# Überlagern der theoretischen PDF
x_fine = np.linspace(140, 210, 1000)
pdf_fine = norm.pdf(x_fine, loc=175, scale=10)
plt.plot(x_fine, pdf_fine, label="Theoretische PDF", linewidth=2)

plt.title("Stichprobe aus der Standardnormalverteilung")
plt.xlabel("x")
plt.ylabel("Dichte")
plt.legend()
plt.show()
```

```python
# Einige statistische Kennwerte unserer Stichprobe ---

mean_sample = np.mean(samples)
std_sample = np.std(samples)

print(f"Gemittelter Wert aus der Stichprobe: {mean_sample:.4f}")
print(f"Standardabweichung aus der Stichprobe: {std_sample:.4f}")
```
