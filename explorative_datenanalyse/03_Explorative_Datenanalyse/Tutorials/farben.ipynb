# Farben
Das waren jetzt viel Theorie mit vielen verschiedenen Plots zum kennenlernen. Deswegen jetzt ein kurzer Exkurs zum Thema Farben. Denn was macht an Plots am meisten Spaß, wenn nicht ihr Aussehen zu visualisieren!

Und das sieht meistens nicht nur schön aus; mit der richtigen Farbskala erleichterst du deinen Lesern auch das Verstehen der Daten erheblich!
```python
import matplotlib.colors as mlc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

Zunächst einmal kann bei einfachen Plots die Farbe über den Parameter `color` geändert werden. Hierfür stehen einige Farben aus der Matplotlib Bibliotheck zur Verfügung, es können aber auch Hexcodes oder RGB angegeben werden. 

```python
# Ausgeben der verfügbaren 148 Farben
print(mlc.CSS4_COLORS)
```

```python
np.random.seed(111)
x = np.random.rand(100)
y = np.random.rand(100)
df = pd.DataFrame({"x": x, "y": y})
```

```python
f = plt.figure(figsize=(17, 6))
gs = f.add_gridspec(1, 3)

# erstellt jeweils einen Subplot und weißt sie den jeweiligen Stellen zu
ax1 = f.add_subplot(gs[0, 0])
ax2 = f.add_subplot(gs[0, 1])
ax3 = f.add_subplot(gs[0, 2])
sns.scatterplot(df, x="x", y="y", color="aqua", ax=ax1)
sns.scatterplot(df, x="x", y="y", color="#65081F", ax=ax2)  # mit einem Hexcode
sns.scatterplot(
    df, x="x", y="y", color=(50 / 255, 205 / 255, 50 / 255), ax=ax3
)  # mit einem RGB-Code, wobei die Werte normiert werden müssen.

plt.tight_layout()
plt.show()
```

Da wir oft mehrere Variablen oder unterschiedliche Gruppen einfärben wollen, verfügt Seaborn auch über einige Farbpaletten.

## der Default
Seaborn hat per default eine festgelegte Palette, die zur Gruppe der qualitative Paletten gehört (dazu gleich mehr). 
Du kannst dir deine aktuell festsgelegte Palette via `.color_palette()`anzeigen lassen.
Seaborn verfügt über mehrere integrierte Farbpaletten, darunter `deep`, `muted`, `pastel`, `dark` und `colorblind`. Wenn du eine dieser Paletten als Argument in `.color_palette()` angibst, kannst du dir diese via `palplot` ausgeben lassen:

```python
sns.reset_defaults()  # setzt Seaborn Einstellungen auf den default zurück
print("Das ist deine aktuelle Palette: \n")
sns.palplot(sns.color_palette())
plt.show()
print("Das ist eine ausgewählte Palette: \n")
sns.palplot(sns.color_palette("pastel"))
plt.show()
sns.set_palette("dark")  # default festlegen
print("Das ist deine aktualisierte default Palette: \n")
sns.palplot(sns.color_palette())
plt.show()
```

## Qualitative, sequentielle und divergiierende Farbpaletten
Farbpalleten können in *3* Gruppen unterteilt werden.

**qualitative Paletten** dienen dazu, Klassen zu unterscheiden, ohne eine unterschiedliche Wertung zu implizieren. Diese Farbpaletten eignen sich gut, wenn man kategorische Gruppen unterscheiden möchte. Stell dir z.B. vor, dass du einen Datensatz von 100 Hunden mit unterschiedlichen Rassen hast. Die verschiedenen Rassen kannst du dann mit einer qualitatien Farbpalette einfärben, ohne eine Rasse hervorzuheben!

**sequentielle Paletten** dienen, wie der Name schon vermuten lässt, um Abstufungen einer Sequenz zu verdeutlichen. Diese sind für Daten sinnvoll, dessen Werte einer Auf- oder Abstufung folgen.

**divergierende Paletten** werden benutzt, um starke Abweichungen von einem neutral eingefärbtem Mittelwert zu unterstreichen. Die Farbskale verläuft hierbei in zwei Richtungen, um sowohl negative als auch positive Abweichungen darzustellen.

In Seaborn sind viele Paletten dieser Gruppen durch die *Brewer palettes* zur Verfügung gestellt. Diese variieren jedoch in der Anzahl an ihnen zugehörigen Farben.


```python
# color Brewer palettes eingeteilt nach ihren Gruppen

qualitative_palettes = ["Accent", "Dark2", "Paired", "Pastel1", "Pastel2", "Set1", "Set2", "Set3"]

sequential_palettes = [
    "Blues",
    "BuGn",
    "BuPu",
    "GnBu",
    "Greens",
    "Greys",
    "Oranges",
    "OrRd",
    "PuBu",
    "PuBuGn",
    "PuRd",
    "Purples",
    "RdPu",
    "Reds",
    "YlGn",
    "YlGnBu",
    "YlOrBr",
    "YlOrRd",
]


diverging_palettes = [
    "BrBG",
    "PiYG",
    "PRGn",
    "PuOr",
    "RdBu",
    "RdGy",
    "RdYlBu",
    "RdYlGn",
    "Spectral",
    "coolwarm",
]
```

```python
def show_colormaps(colormaps, **kwargs):
    for cmap in colormaps:
        palette = sns.color_palette(cmap, **kwargs)
        print(cmap)
        display(palette)
```

```python
show_colormaps(qualitative_palettes)
```

```python
show_colormaps(sequential_palettes)
```

```python
show_colormaps(diverging_palettes)
```

### qualitative Farbpaletten

```python
# blobs für qualitatve Daten
np.random.seed(77)
x = np.random.rand(100)
y = np.random.rand(100)
groups = np.random.choice(range(5), 100)
data = pd.DataFrame({"x": x, "y": y, "group": groups})
```

```python
sns.scatterplot(data=data, x=x, y=y, hue=groups, palette=np.random.choice(qualitative_palettes))
plt.show()
```

### sequentielle Paletten

```python
# Nutze den Datensatz Diamonds für Sequenzielle Daten

diamonds = sns.load_dataset("diamonds")
diamonds_sample = diamonds.sample(n=600, random_state=7)

sns.scatterplot(
    data=diamonds_sample,
    x="carat",
    y="price",
    hue="clarity",
    palette=np.random.choice(sequential_palettes) + "_r",
)  # an jede zur Verfügung gestellte Palette kann ein '_r' angehängt werden, um die Abstufungsreihenfolge umzukehren
plt.show()
```

Um dir eine Abstufung einer Farbe ausgeben zu lassen und dir damit selber eine sequentielle Farbpalette zu erstellen, kannst du `light_palette` und `dark_palette` benutzen. Als zweites Argument kannst du einen Integer eingeben, um zu bestimmen, wie viele Abstufungen du haben willst. Diese müssen mit der Anzahl an Kategorien deines Datensatzes übereinstimmen.

**Hinweis**: Beachte, dass deine erstellte Palette in einer Variable gespeichert wird. Beim Aufruf in den plot-Parametern musst du sie also ohne Anführungszeichen eingeben.

```python
sns.palplot(sns.light_palette("pink"))
d_pinks = sns.dark_palette("pink", 8)
sns.palplot(d_pinks)
plt.show()
```

```python
sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette=d_pinks)
plt.show()
```

Du kannst mit Seaborn aber auch Paletten erstellen, die einen Verlauf bilden. Im Plot wird den Gruppen dann automatisch eine passende Farbe zugewiesen, und du musst nicht darauf achten, dass du exakt so viele Farben wie Gruppen hast. Der Schlüssel hierfür ist der Parameter `as_cmap`, wobei cmap für colormap steht.

Du kannst hierbei als Farbargumente einzelnde Farben, aber auch ganze Paletten angeben. Tausche das Farbargument unten aus und schaue dir an, wie sequentielle, divergierende und einzelnde Farben als colormap aussehen. Seaborn stellt auch hier wieder eigene Paletten zur Verfügung, wie `rocket`, `mako`, `flare` und `crest`. Von Matplotlib wurden noch die cmaps `viridis` und `magma` übernommen, die du hier auch benutzen kannst.

Du wirst merken, dass du bei selbst generierten Farbverläufen und qualitativen Paletten *keinen* Verlauf erhälst. Bei den selbst generierten Paletten kannst du das Argument `as_cmap` der `dark_palette`/`light_palette` auf True setzten und erhälst auch so einen Verlauf.

```python
sns.dark_palette("pink", as_cmap=True)
```

```python
sns.color_palette("magma", as_cmap=True)
```

```python
sns.scatterplot(
    data=diamonds_sample, x="carat", y="price", hue="clarity", palette="magma"
)  # bei Angabe einer cmap werden automatisch Farben aus dieser für die Gruppen ausgewäht.
plt.show()
```

### divergierende Paletten
Auch von den divergierenden Paletten stellt Seaborn wieder welche zur Verfügung: `vlag` und `icefire`. Um eine eigene Palette zu erstellen gibt es die Funktion `diverging_palette`, in der ein Start-/Endton angegeben wird, sowie der Parameter `as_cmap=True`. Die ersten zwei ziffern geben hierbei Gradzahlen an, die dsich auf den HUSL Farbraum beziehen. Außerdem kann auch angegeben werden, ob das center 'light' oder 'dark' sein soll.
Divergierende Paletten eignen sich besonders gut für Heatmaps!

```python
sns.color_palette("icefire", as_cmap=True)
```

```python
sns.diverging_palette(300, -200, as_cmap=True, center="light")  # eigene Paletten erstellen
```

```python
# erzeuge eine Matrix an Zufalsdaten
np.random.seed(111)
data = np.random.rand(11, 11)
```

```python
sns.heatmap(
    data, cmap=np.random.choice(diverging_palettes), linewidths=0.5
)  # Farbe wird zufällig aus der Liste diverging_palettes gewählt
plt.show()
```

### Style 

Die Funktion `set_style()` erlaubt es, den Hintergrund eurer Plots als default festzulegen. Wenn die Funktion ohne Paramter aufgerufen wird, legt Seaborn den default auf `darkgrid` fest. Die Parameter `dark/whitegrid` sowie `white/dark` können übergeben werden, um den Hintergrund anzupassen. Der Parameter `ticks` kann außerdem angegeben werden, um die Achsen mit ticks zu versehen.

**Hinweis**: da Seaborn auf Matplotlib aufbaut, werden deine Einstellungen auch für die plt-Plots übernommen.

```python
# definiere eine Vorlage, auf der mehrere Plots abgebildet werden können
f = plt.figure(figsize=(10, 7))
gs = f.add_gridspec(2, 2)

# mit dem 'with'-statement kann man den Style temporär ändern. Dafür wird aber 'axes_style' statt 'set_style' verwendet
with sns.axes_style("darkgrid"):
    ax = f.add_subplot(gs[0, 0])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")
with sns.axes_style("whitegrid"):
    ax = f.add_subplot(gs[0, 1])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")
with sns.axes_style("dark"):
    ax = f.add_subplot(gs[1, 0])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")
with sns.axes_style("white"):
    ax = f.add_subplot(gs[1, 1])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")

# verhindert die Überlagerung von Titeln und beschriftungen
f.tight_layout()
plt.show()
```

### Skalierung

Es gibt außerdem die Möglichkeit, die Skalierung der Plotelemente zu ändern. Das Schlüsselwort hierfür ist `set_context()`: mit den Parametern `paper`, `notebook`, `talk` und `poster` können die plots angepasst werden, wobei notebook der default ist.

```python
# definiere eine Vorlage, auf der mehrere Plots abgebildet werden können
f = plt.figure(figsize=(12, 6))
gs = f.add_gridspec(1, 2)

# mit dem 'with'-statement kann man den Style temporär ändern. Dafür wird statte 'set' 'plotting' benutzt.
with sns.plotting_context("paper"):
    ax = f.add_subplot(gs[0, 0])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")
with sns.plotting_context("notebook"):
    ax = f.add_subplot(gs[0, 1])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")

f.tight_layout()
plt.show()
```

```python
f = plt.figure(figsize=(12, 6))
gs = f.add_gridspec(1, 2)
with sns.plotting_context("talk"):
    ax = f.add_subplot(gs[0, 0])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")
with sns.plotting_context("poster"):
    ax = f.add_subplot(gs[0, 1])
    sns.scatterplot(data=diamonds_sample, x="carat", y="price", hue="clarity", palette="Spectral")

# verhindert die Überlagerung von Titeln und beschriftungen
f.tight_layout()
plt.show()
```

Diese beiden Änderungen sind unter der Funktion `set_theme()` zusammengefasst. Darunter findet man auch weitere Parameter, wie z.B die Schriftart, Schriftgröße oder auch die Farbpalette.
