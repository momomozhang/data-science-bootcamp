## 🧱 Seaborn basiert auf Matplotlib – was bringt mir das?

Jeder Seaborn-Plot ist eigentlich ein **Matplotlib-Plot**, den du weiter bearbeiten kannst.  
Hier ein paar coole Möglichkeiten:

---

### 🎯 Möglichkeiten, Matplotlib mit Seaborn zu kombinieren:

| Was du ändern willst           | Matplotlib-Befehl        | Beispiel                           |
|-------------------------------|---------------------------|------------------------------------|
| **Titel hinzufügen**          | `plt.title()`             | `plt.title("Mein Plot")`          |
| **Achsenbeschriftung ändern** | `plt.xlabel()`, `plt.ylabel()` | `plt.xlabel("X-Wert")`         |
| **Achsenbereich setzen**      | `plt.xlim()`, `plt.ylim()` | `plt.ylim(0, 50)`               |
| **Legende anpassen**          | `plt.legend()`            | `plt.legend(loc="upper left")`    |
| **Mehrere Plots kombinieren** | `plt.subplots()`          | Für Subplots / mehrere Diagramme  |
| **Grafik speichern**          | `plt.savefig()`           | `plt.savefig("mein_plot.png")`    |

---

### 🧪 Beispiel: Seaborn + Matplotlib kombiniert

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("tips")

# Seaborn-Plot
sns.boxplot(data=df, x="day", y="total_bill", hue="sex")
# Matplotlib-Feintuning
plt.title("Rechnungshöhen nach Wochentag und Geschlecht")
plt.xlabel("Wochentag")
plt.ylabel("Rechnungsbetrag ($)")
plt.ylim(0, 60)
plt.legend(title="Geschlecht", loc="upper left")
plt.tight_layout()
```

---

### 🔍 Warum Seaborn, wenn es Matplotlib gibt?

- Seaborn baut auf Matplotlib auf.
- Macht vieles einfacher: weniger Code, bessere Standard-Stile.
- Statistische Infos (Fehlerbalken, Regression) gibt’s direkt eingebaut.

---

### 🔧 Setup – Installation & Import

```python
# Wenn du Seaborn noch nicht installiert hast, nutze:
# !pip install seaborn

import matplotlib.pyplot as plt
import seaborn as sns

# Schönes Standard-Design setzen
sns.set_theme()

# Beispiel-Datensatz laden
df = sns.load_dataset("tips")
df.head()
```

## 📊 3. Diagramm-Typen – Struktur & Überblick

## 🗂️ Überblick: Seaborn Diagramm-Typen

| Diagramm-Typ     | Seaborn-Funktion         | Für welche Fragen?                     | Typische `kind=` Optionen          |
|------------------|--------------------------|----------------------------------------|------------------------------------|
| **Verteilungsplot** | `displot()`, `histplot()`, `kdeplot()` | Wie ist eine Variable verteilt?       | histogram, kde                    |
| **Beziehungsplot** | `relplot()`, `scatterplot()`, `lineplot()` | Wie hängen 2 Variablen zusammen?     | scatter, line                     |
| **Kategorischer Plot** | `catplot()`, `barplot()`, `boxplot()`, `violinplot()` | Wie unterscheiden sich Gruppen?     | bar, box, violin, swarm, strip   |
| **Matrixplot**    | `heatmap()`             | Beziehungen zwischen vielen Variablen  | –                                  |
| **Multivariate**  | `pairplot()`, `jointplot()` | Zusammenspiel mehrerer Variablen     | scatter, kde, hist                |


---

### 📦 4. Verteilungsplots (`displot`, `histplot`, `kdeplot`)

```python
# Histogramm – zeigt Häufigkeit
sns.histplot(df["total_bill"])
plt.title("Histogramm der Rechnungsbeträge")
plt.show()
```

```python
# Dichteplot – glatter Verlauf
sns.kdeplot(df["total_bill"])
plt.title("Dichteverteilung der Rechnungsbeträge")
plt.show()
```

```python
# displot kombiniert beides
sns.displot(df["total_bill"], kde=True)
```

🔎 **Wofür?**
- Univariate Verteilung: Wie ist eine einzelne Variable verteilt?
- `displot()` ist flexibel und zeigt Histogramm + Dichte.

🎯 **Tipp:** `bins=` anpassen, um die Detailtiefe zu verändern.


---

### 🔗 5. Beziehungsplots (`relplot`, `scatterplot`, `lineplot`)

```python
# Scatterplot: Zwei quantitative Variablen
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.title("Rechnung vs. Trinkgeld")
plt.show()
```

```python
# Linienplot für Zeitreihen etc.
sns.lineplot(data=df, x="size", y="tip")
```

```python
# Mit zusätzlicher Gruppierung (Farbe)
sns.relplot(data=df, x="total_bill", y="tip", hue="sex", kind="scatter")
```

🔎 **Wofür?**
- Zeigt Beziehung/Zusammenhang zwischen zwei numerischen Variablen.
- `relplot()` kann je nach `kind` Scatter oder Linie sein.
- Farbe (`hue`), Form (`style`) und Größe (`size`) für Gruppierungen.

🎯 **Tipp:** Perfekt für Korrelationen oder Trends.


---

### 🧩 6. Kategorische Plots (`catplot`, `barplot`, `boxplot`, `violinplot`, `swarmplot`)

```python
# Balkendiagramm mit Mittelwerten & Fehlerbalken
sns.barplot(data=df, x="day", y="total_bill")
```

```python
# Boxplot: zeigt Median, Quartile & Ausreißer
sns.boxplot(data=df, x="day", y="total_bill")
```

```python
# Violinplot: Verteilung + Boxplot kombiniert
sns.violinplot(data=df, x="day", y="total_bill")
```

```python
# Swarmplot: Jedes Datenpunkt einzeln sichtbar
sns.swarmplot(data=df, x="day", y="total_bill")
```

🔎 **Wofür?**
- Vergleich zwischen Gruppen (z. B. Tage, Geschlechter).
- `catplot()` ist das „Framework“ – kann alle `kind=` Formen.
- Besonders nützlich für Mittelwert, Streuung & Ausreißer.

🎯 **Tipp:** Nutze `catplot()` mit verschiedenen `kind=`-Werten:

```python
# Alle Optionen schnell testen:
sns.catplot(data=df, x="day", y="total_bill", kind="box")  # oder: bar, box, swarm, strip
```

---

### 🧠 7. Fortgeschrittene Funktionen

```python
# Korrelationsmatrix
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
```

```python
# Pairplot: zeigt Scatter & Verteilungen aller Kombinationen
sns.pairplot(df, hue="sex")
```

```python
# Jointplot: Kombination aus Scatter + Histogrammen
sns.jointplot(data=df, x="total_bill", y="tip", kind="reg")  # auch: hex, kde, scatter
```

🔎 **Wofür?**
- Heatmaps → perfekte Übersicht über Zusammenhänge.
- Pairplot → Schnell viele Scatterplots auf einmal.
- Jointplot → Fokus auf zwei Variablen, aber mit Extras.


---

### 🎨 8. Stil & Design anpassen

```python
# Farbpalette ändern
sns.set_palette("pastel")

# Achsen & Titel mit Matplotlib ergänzen
sns.boxplot(data=df, x="sex", y="tip")
plt.title("Trinkgeld nach Geschlecht")
plt.xlabel("Geschlecht")
plt.ylabel("Trinkgeld in $")
```

🎯 **Tipp:** Seaborn kümmert sich um die meisten Styles – aber mit Matplotlib kannst du alles feinjustieren.


---

### 🧩 9. Seaborn + Matplotlib kombinieren (Subplots)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df, x="day", y="total_bill", ax=axes[0])
axes[0].set_title("Boxplot")

sns.violinplot(data=df, x="day", y="total_bill", ax=axes[1])
axes[1].set_title("Violinplot")

plt.tight_layout()
```
