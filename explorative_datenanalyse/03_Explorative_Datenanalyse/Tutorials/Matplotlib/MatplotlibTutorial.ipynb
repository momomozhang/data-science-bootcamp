# Matplotlib Basics – Dein Einstieg in die Welt der Datenvisualisierung

## Willkommen zu Matplotlib!
Hey! Heute wirst du lernen, wie du mit **Matplotlib** beeindruckende Diagramme erstellen kannst. Egal, ob du einfache Linienplots oder komplexe Diagramme mit mehreren Subplots basteln willst – am Ende dieses Tutorials hast du den Durchblick. 🎨📊

### Warum Matplotlib?
Matplotlib ist ideal, wenn du **volle Kontrolle** über die Darstellung brauchst. Es erlaubt dir, jedes Detail eines Plots anzupassen – von Achsenticks über Farben bis hin zu individuellen Markern. Perfekt für datengetriebene Visualisierungen, bei denen es auf Präzision ankommt!

## Import und Grundstruktur
Zuerst einmal müssen wir **Matplotlib** importieren. Der Standardweg ist:

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Matplotlib arbeitet mit **Figure** und **Axes**:
- Eine **Figure** ist die gesamte Zeichenfläche.
- Ein **Axes** ist ein einzelnes Diagramm innerhalb der Figure.

### Beispiel: Eine leere Figur mit einem Achsen-Objekt

```python
fig, ax = plt.subplots()  # Erstellt eine Figur mit einer Achse
ax.set_title("Mein erstes Diagramm")
plt.show()
```

## Erster Plot: Einfache Linie
Lass uns gleich loslegen mit einem **Linienplot**!

```python
# Unsere x- und y-Werte
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]

plt.plot(x, y)  # Erstellt ein Liniendiagramm
plt.show()  # Zeigt das Diagramm an
```

### Achsenbeschriftungen und Titel
Dein Plot sieht besser aus, wenn er beschriftet ist!

```python
plt.plot(x, y)
plt.xlabel("X-Werte")  # X-Achse benennen
plt.ylabel("Y-Werte")  # Y-Achse benennen
plt.title("Einfacher Linienplot")  # Titel setzen
plt.show()
```

## Weitere Diagramme

```python
# Zufallszahlengenerator mit einem festen Seed initialisieren,
# um reproduzierbare Ergebnisse zu gewährleisten
np.random.seed(42)

# Erstelle eine gleichmäßig verteilte Zahlenreihe von 0 bis 10 mit 100 Werten
x = np.linspace(0, 10, 100)

# Berechne die Sinuswerte für x und füge normalverteiltes Rauschen hinzu
# (Mittelwert 0, Standardabweichung 0.1) für realistischere Daten
y = np.sin(x) + np.random.normal(0, 0.1, 100)

# Ein DataFrame mit zufälligen Geschäftsdaten für die Jahre 2000-2019 erstellen
df = pd.DataFrame(
    {
        "Jahr": np.arange(2000, 2020),  # Jahreszahlen von 2000 bis 2019
        "Umsatz": np.random.randint(5000, 20000, 20),  # Zufälliger Umsatz zwischen 5000 und 20000
        "Kosten": np.random.randint(2000, 15000, 20),  # Zufällige Kosten zwischen 2000 und 15000
        "Kunden": np.random.randint(50, 500, 20),  # Zufällige Anzahl an Kunden zwischen 50 und 500
    }
)
```

```python
# Figur mit fester Größe erstellen
plt.figure(figsize=(10, 5))

# Linienplot für Umsatz mit Markern und gestrichelter Linie
plt.plot(df["Jahr"], df["Umsatz"], label="Umsatz", linestyle="--", marker="o")

# Achsenbeschriftungen setzen
plt.xlabel("Jahr")
plt.ylabel("Umsatz in €")

# X-Achsen-Ticks anpassen (Jahreszahlen leicht drehen)
plt.xticks(df["Jahr"], rotation=45)

# Titel und Legende hinzufügen
plt.title("Jährlicher Umsatz - Liniendiagramm")
plt.legend()

# Diagramm anzeigen
plt.show()
```

```python
# Figur mit fester Größe erstellen
plt.figure(figsize=(10, 5))

# Balkendiagramm für Umsatz mit grüner Farbe
plt.bar(df["Jahr"], df["Umsatz"], color="green", label="Umsatz")

# Achsenbeschriftungen setzen
plt.xlabel("Jahr")
plt.ylabel("Umsatz in €")

# X-Achse anpassen (Jahreszahlen leicht drehen)
plt.xticks(df["Jahr"], rotation=45)

# Titel und Legende hinzufügen
plt.title("Jährlicher Umsatz - Balkendiagramm")
plt.legend()

# Diagramm anzeigen
plt.show()
```

```python
# Vergleich von Balkendiagrammen - Umsatz vs. Kosten
plt.figure(figsize=(10, 5))
plt.bar(df["Jahr"], df["Umsatz"], color="blue", label="Umsatz")
plt.bar(df["Jahr"], df["Kosten"], color="red", alpha=0.7, label="Kosten")
plt.xlabel("Jahr")
plt.ylabel("Betrag in €")
plt.xticks(df["Jahr"], rotation=45)
plt.title("Vergleich Umsatz vs. Kosten")
plt.legend()
plt.show()
```

```python
# Figur mit fester Größe erstellen
plt.figure(figsize=(8, 5))

# Streudiagramm: Kundenanzahl vs. Umsatz
plt.scatter(df["Kunden"], df["Umsatz"], color="red", alpha=0.6)
# alpha = Transparenz (0 = durchsichtig, 1 = voll sichtbar)

# Achsenbeschriftungen setzen
plt.xlabel("Kundenanzahl")
plt.ylabel("Umsatz in €")

# Titel hinzufügen
plt.title("Zusammenhang zwischen Kunden und Umsatz")

# Diagramm anzeigen
plt.show()
```

```python
# Zufallszahlen aus einer Normalverteilung erzeugen
# (Mittelwert=50, Standardabweichung=15, 500 Werte)
random_data = np.random.normal(loc=50, scale=15, size=500)

# Figur mit fester Größe erstellen
plt.figure(figsize=(8, 5))

# Histogramm der Zufallszahlen mit 20 Bins und violetter Farbe
plt.hist(random_data, bins=30, color="purple", alpha=0.7)
# bins = Anzahl der Balken (je mehr Bins, desto feiner die Auflösung)
# alpha = Transparenz der Balken (0 = unsichtbar, 1 = volle Deckkraft)

# Achsenbeschriftungen setzen
plt.xlabel("Werte")
plt.ylabel("Häufigkeit")

# Titel hinzufügen
plt.title("Histogramm einer Normalverteilung")

# Diagramm anzeigen
plt.show()
```

```python
# Labels für die einzelnen Marktanteile
labels = ["Firma A", "Firma B", "Firma C", "Firma D"]

# Größen der Marktanteile in Prozent
sizes = [35, 25, 25, 15]

# Figur mit fester Größe erstellen
plt.figure(figsize=(6, 6))

# Kuchendiagramm erstellen
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=["blue", "green", "red", "purple"],
)
# autopct = Zeigt Prozentwerte im Diagramm an ('%1.1f%%' für eine Dezimalstelle)
# startangle = Startwinkel des ersten Segments (90° = beginnt oben)

# Titel hinzufügen
plt.title("Marktanteile")

# Diagramm anzeigen
plt.show()
```

```python
# Figur mit fester Größe erstellen
plt.figure(figsize=(8, 5))

# Linienplot mit Punkten, durchgezogener Linie und angepasster Farbe
plt.plot(x, y, linestyle="-", marker="o", color="darkblue", linewidth=2)
# linestyle = Linienstil ('-' = durchgezogen)
# marker = Markierung an den Datenpunkten ('o' = Kreise)
# color = Farbe der Linie ('darkblue')
# linewidth = Linienstärke

# Raster (Grid) aktivieren mit gestricheltem Stil und teilweiser Transparenz
plt.grid(True, linestyle="--", alpha=0.6)
# linestyle = Art des Gitters ('--' = gestrichelt)
# Figur mit fester Größe erstellen
plt.figure(figsize=(8, 5))

# Linienplot mit Punkten, durchgezogener Linie und angepasster Farbe
plt.plot(x, y, linestyle="-", marker="o", color="darkblue", linewidth=2)
# linestyle = Linienstil ('-' = durchgezogen)
# marker = Markierung an den Datenpunkten ('o' = Kreise)
# color = Farbe der Linie ('darkblue')
# linewidth = Linienstärke

# Raster (Grid) aktivieren mit gestricheltem Stil und teilweiser Transparenz
plt.grid(True, linestyle="--", alpha=0.6)
# linestyle = Art des Gitters ('--' = gestrichelt)
# alpha = Transparenz des Gitters (0 = unsichtbar, 1 = volle Deckkraft)

# Achsenbeschriftungen setzen
plt.xlabel("X-Werte")
plt.ylabel("Y-Werte")

# Titel hinzufügen
plt.title("Stilisiertes Diagramm")

# Diagramm anzeigen
plt.show()
```

```python
# Figur mit zwei Subplots (1 Zeile, 2 Spalten) erstellen
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
# nrows = Anzahl der Zeilen (1)
# ncols = Anzahl der Spalten (2)
# figsize = Größe der gesamten Figur
# `axes` ist eine Liste von Achsen-Objekten [axes[0], axes[1]],
# mit denen wir die einzelnen Plots ansprechen können.

# Erster Subplot: Liniendiagramm
axes[0].plot(x, y)  # Plot im ersten Subplot (links)
axes[0].set_title("Liniendiagramm")  # Titel für den ersten Plot setzen

# Zweiter Subplot: Balkendiagramm
axes[1].bar(df["Jahr"], df["Umsatz"], color="blue")  # Balkendiagramm im zweiten Subplot (rechts)
axes[1].set_title("Balkendiagramm")  # Titel für den zweiten Plot setzen

# Gesamte Figur anzeigen
plt.show()
```

| **Subplots**             | `fig, ax = plt.subplots(nrows=1, ncols=2, figsize=…)`                                      | `fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(8,6))`<br>`axes[0,0].plot(x, y)`<br>`axes[1,1].bar(x, y)`                                                              | Erstellt mehrere Unterplots in einer Figur. Jeder `ax` ist eine eigene Achse, auf der geplottet werden kann.                                                         |
| **Titel setzen**         | `plt.title("Mein Titel")`<br>`ax.set_title("Mein Titel")`                                  | `plt.plot(x, y)`<br>`plt.title("Mein Plot")`                                                                                                                               | Setzt den Titel für das aktuelle Plot/Achsen-Objekt.                                                                                                                 |

## Plot mit vielen Anpassungen

```python
# x-Werte in Grad
x_deg = np.linspace(0, 180, 50)  # Erzeugt 50 gleichmäßig verteilte Werte von 0° bis 180°

# Daten basierend auf Tangens, angepasst für Gradwerte
y3 = (
    np.tan(np.radians(x_deg) / 2) - 0.5
)  # Berechnet den Tangens (nach Umwandlung in Radiant) und skaliert ihn

z = np.linspace(1, 50, 50)  # Für Farbskala

# Figur und Achse erstellen
fig, ax = plt.subplots(
    figsize=(16, 9), dpi=100
)  # Erstellt eine Figur mit einer Achse (16x9 Zoll, 100 dpi)

# Linienplot für Sinus
ax.plot(
    x_deg,
    np.sin(np.radians(x_deg)),
    color="tab:blue",
    linestyle="--",
    marker="o",
    markersize=6,
    markeredgecolor="black",
    linewidth=2,
    label="Sinus",
)
# Plottet die Sinus-Funktion (nach Umwandlung in Radiant)
# Farbe: Blau, gestrichelte Linie, Kreismarker, schwarze Markerumrandung

# Linienplot für Kosinus
ax.plot(
    x_deg,
    np.cos(np.radians(x_deg)),
    color="tab:red",
    linestyle=":",
    marker="s",
    markersize=6,
    markeredgecolor="black",
    linewidth=2,
    label="Kosinus",
)
# Plottet die Kosinus-Funktion (nach Umwandlung in Radiant)
# Farbe: Rot, gepunktete Linie, Quadratmarker, schwarze Markerumrandung

# Scatter-Plot für Tangens-Werte
scatter = ax.scatter(
    x_deg,
    y3,
    c=z,
    cmap="viridis",
    alpha=0.9,
    s=100,
    edgecolor="black",
    linewidth=1,
    label="Tangens-Datenpunkte",
)
# Scatter-Plot mit Tangens-Werten, farbcodiert mit der viridis-Farbskala
# alpha = Transparenz (0.9), s = Punktgröße (100), edgecolor = schwarze Umrandung

# Achsentitel und Labels setzen
ax.set_title("Beispiel-Plot mit X-Achse in Grad", fontsize=14)  # Titel mit Schriftgröße 14
ax.set_xlabel("X-Achse (Grad)", fontsize=12)  # X-Achsenbeschriftung mit Schriftgröße 12
ax.set_ylabel("Y-Achse (Wert)", fontsize=12)  # Y-Achsenbeschriftung mit Schriftgröße 12

# Achsenlimits setzen
ax.set_xlim(0, 180)  # X-Achse von 0° bis 180° begrenzen
ax.set_ylim(-1.5, 1.5)  # Y-Achse von -1.5 bis 1.5 begrenzen

# Ticks auf sinnvolle Grad-Werte setzen
ax.set_xticks([0, 30, 60, 90, 120, 150, 180])  # Setzt spezifische X-Ticks bei Gradwerten
ax.set_xticklabels(
    ["0°", "30°", "60°", "90°", "120°", "150°", "180°"], rotation=0
)  # Setzt die Tick-Beschriftung als Gradzeichen
ax.grid(
    True, linestyle="--", alpha=0.5
)  # Aktiviert das Gitter mit gestrichelter Linie und 50% Transparenz

# Hintergrundfarbe ändern
ax.set_facecolor("#f0f0f0")  # Setzt die Hintergrundfarbe des Plots auf hellgrau

# Annotation hinzufügen
ax.annotate(
    "Maximalwert Sinus",
    xy=(90, 1),
    xytext=(110, 1.2),
    arrowprops=dict(arrowstyle="->", color="black"),
    fontsize=10,
)
# Fügt eine Annotation bei x=90° (maximaler Sinuswert) mit einem Pfeil ein

# Legende anzeigen
ax.legend(loc="upper right")  # Legende in der oberen rechten Ecke platzieren

# Farbskala hinzufügen
cbar = plt.colorbar(scatter, ax=ax)  # Fügt eine Farbskala für die Scatter-Punkte hinzu
cbar.set_label("Farbskala (Werte von 1-50)")  # Beschriftung für die Farbskala

# Layout optimieren
plt.tight_layout()  # Optimiert die Platzierung der Elemente für bessere Darstellung

# Plot anzeigen
plt.show()  # Zeigt den fertigen Plot an
```

### 3D-Plots (kurzer Einblick)

```python
# 3D-Plot-Modul importieren
import numpy as np

# 3D-Figur erstellen
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")  # 3D-Achse hinzufügen

# X- und Y-Werte gleichmäßig im Bereich -5 bis 5 erzeugen
X = np.linspace(-5, 5, 100)
Y = np.linspace(-5, 5, 100)

# 2D-Gitter aus X- und Y-Werten erstellen
X, Y = np.meshgrid(X, Y)

# Z-Werte basierend auf einer Sinus-Funktion berechnen
Z = np.sin(np.sqrt(X**2 + Y**2))

# 3D-Oberflächenplot erstellen mit der Farbkarte 'viridis'
ax.plot_surface(X, Y, Z, cmap="viridis")

# Plot anzeigen
plt.show()
```
