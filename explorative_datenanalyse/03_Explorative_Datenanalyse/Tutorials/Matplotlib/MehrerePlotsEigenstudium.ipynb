In Python gibt es verschiedene Möglichkeiten, mehrere Plots auf einer **Figure** darzustellen. Hier sind die gängigsten Methoden:

### 1. **`plt.subplot()`**
Mit `plt.subplot(rows, cols, index)` kannst du eine Figure in ein Raster mit mehreren Subplots unterteilen.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.figure(figsize=(10, 5))

plt.subplot(2, 2, 1)  # 2x2 Grid, 1. Plot
plt.plot(x, np.sin(x))
plt.title("Sinus")

plt.subplot(2, 2, 2)  # 2. Plot
plt.plot(x, np.cos(x))
plt.title("Cosinus")

plt.subplot(2, 2, 3)  # 3. Plot
plt.plot(x, np.tan(x))
plt.title("Tangens")

plt.subplot(2, 2, 4)  # 4. Plot
plt.plot(x, np.exp(-x))
plt.title("Exponentiell")

plt.tight_layout()  # Verhindert Überlappungen
plt.show()
```

---

### 2. **`plt.subplots()` (Empfohlen für mehr Kontrolle)**
Diese Methode erstellt eine Figure mit mehreren Subplots und gibt eine **Figure** und **Achsen-Objekte** zurück.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 5))

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Sinus")

axes[0, 1].plot(x, np.cos(x))
axes[0, 1].set_title("Cosinus")

axes[1, 0].plot(x, np.tan(x))
axes[1, 0].set_title("Tangens")

axes[1, 1].plot(x, np.exp(-x))
axes[1, 1].set_title("Exponentiell")

plt.tight_layout()
plt.show()
```

---

### 3. **`plt.axes()` für individuelle Positionierung**
Falls du eine feinere Kontrolle über die Position von Plots brauchst.

```python
plt.figure(figsize=(8, 6))

ax1 = plt.axes([0.1, 0.6, 0.3, 0.3])  # [x, y, Breite, Höhe]
ax1.plot(x, np.sin(x))
ax1.set_title("Sinus")

ax2 = plt.axes([0.6, 0.6, 0.3, 0.3])
ax2.plot(x, np.cos(x))
ax2.set_title("Cosinus")

ax3 = plt.axes([0.1, 0.1, 0.3, 0.3])
ax3.plot(x, np.tan(x))
ax3.set_title("Tangens")

ax4 = plt.axes([0.6, 0.1, 0.3, 0.3])
ax4.plot(x, np.exp(-x))
ax4.set_title("Exponentiell")

plt.show()
```

---

### 4. **`GridSpec` für flexible Layouts**
Falls du eine unregelmäßige Anordnung der Subplots möchtest.

```python
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(8, 6))
gs = gridspec.GridSpec(3, 3, figure=fig)  # 3x3 Raster

ax1 = fig.add_subplot(gs[0, :])  # Obere Zeile über die ganze Breite
ax1.plot(x, np.sin(x))
ax1.set_title("Sinus")

ax2 = fig.add_subplot(gs[1, :-1])  # Mittlere Zeile, aber nicht die letzte Spalte
ax2.plot(x, np.cos(x))
ax2.set_title("Cosinus")

ax3 = fig.add_subplot(gs[1:, -1])  # Rechte Spalte, mittlere und untere Zeile
ax3.plot(x, np.tan(x))
ax3.set_title("Tangens")

ax4 = fig.add_subplot(gs[-1, 0])  # Letzte Zeile, erste Spalte
ax4.plot(x, np.exp(-x))
ax4.set_title("Exponentiell")

plt.tight_layout()
plt.show()
```

---

### 5. **Mehrere Plots auf einer Achse (Overlapping Plots)**
Falls du mehrere Kurven in einem einzigen Plot darstellen möchtest.

```python
plt.figure(figsize=(8, 5))

plt.plot(x, np.sin(x), label="Sinus")
plt.plot(x, np.cos(x), label="Cosinus")
plt.plot(x, np.tan(x), label="Tangens")
plt.plot(x, np.exp(-x), label="Exponentiell")

plt.legend()
plt.title("Mehrere Plots auf einer Achse")
plt.show()
```

---

💡 **Wann welche Methode nutzen?**
- `plt.subplot()` → Schnell und einfach für regelmäßige Raster
- `plt.subplots()` → Empfohlen für Kontrolle über einzelne Achsen
- `plt.axes()` → Individuelle Positionierung
- `GridSpec` → Für flexible Layouts mit unregelmäßigen Größen
- Mehrere Plots auf einer Achse → Wenn alle Daten gemeinsam dargestellt werden sollen



#### **Wann `plt.title()` und wann `set_title()`?**

| **Situation** | **Methode** | **Grund** |
|--------------|------------|-----------|
| **Ein einzelner Plot** | `plt.title()` | Bezieht sich auf den gesamten Plot |
| **Subplots mit `plt.subplot()`** | `plt.title()` | Weil `plt.subplot()` das aktive Subplot setzt |
| **Subplots mit `plt.subplots()`** | `ax.set_title()` | Weil explizite Achsen (`ax[i, j]`) genutzt werden |

✅ **Kurz gesagt:**  
- `plt.title()` funktioniert nur, wenn Matplotlib **weiß, welches Subplot aktiv ist** (wie bei `plt.subplot()`).
- Bei `plt.subplots()` gibt es **keine aktive Achse**, daher muss `set_title()` verwendet werden.

- **Hast du nur einen Plot?** ➝ `plt.title()`
- **Hast du mehrere Subplots?** ➝ `ax.set_title()`
- **Soll die ganze Figure einen Titel haben?** ➝ `fig.suptitle()`

**💡 Pro-Tipp:**  
Wenn du moderne und saubere Plots möchtest, solltest du `plt.subplots()` mit `ax.set_title()` bevorzugen. 😊
