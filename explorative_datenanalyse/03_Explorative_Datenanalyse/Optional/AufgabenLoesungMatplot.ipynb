### **📌 Aufgaben für den Titanic-Datensatz**
💡 **Lade den Datensatz zuerst in pandas:**

```python
import matplotlib.pyplot as plt

### Laden über `seaborn`
import seaborn as sns

titanic = sns.load_dataset("titanic")
display(titanic)
```

---

### **1️⃣ Balkendiagramm: Überlebensrate nach Geschlecht**
👉 Erstelle ein **Balkendiagramm**, das zeigt, wie viele Männer und Frauen überlebt haben.

**Hinweis:**  
- Verwende `plt.bar()`, um die Anzahl der Überlebenden (`Survived == 1`) nach Geschlecht (`Sex`) darzustellen.

💡 **Erwarteter Plot:**
- **X-Achse:** "Männlich" vs. "Weiblich"
- **Y-Achse:** Anzahl der Überlebenden
- **Farben:** Unterschiedliche Farben für die Balken

```python
# 1️⃣ Balkendiagramm: Überlebensrate nach Geschlecht

# Anzahl der Überlebenden nach Geschlecht berechnen
survival_by_gender = titanic.groupby("sex")["survived"].sum()

# Figur erstellen
plt.figure(figsize=(6, 4))

# Balkendiagramm zeichnen
plt.bar(survival_by_gender.index, survival_by_gender.values, color=["pink", "blue"])

# Achsenbeschriftungen setzen
plt.xlabel("Geschlecht")
plt.ylabel("Anzahl der Überlebenden")
plt.title("Überlebensrate nach Geschlecht")

# Diagramm anzeigen
plt.show()
```

---

### **2️⃣ Kuchendiagramm: Anteil der Passagierklassen**
👉 Erstelle ein **Kuchendiagramm**, das die **Verteilung der Passagierklassen (`Pclass`)** zeigt.

**Hinweis:**  
- Nutze `plt.pie()`, um die Anzahl der Passagiere in **1., 2. und 3. Klasse** anzuzeigen.
- Setze **Prozentwerte** mit `autopct='%1.1f%%'`.

💡 **Erwarteter Plot:**
- **Drei Segmente**: 1. Klasse, 2. Klasse, 3. Klasse
- **Prozentangaben** auf den Segmenten

```python
# Anzahl der Passagiere pro Klasse berechnen
class_counts = titanic["pclass"].value_counts()
display(class_counts)
```

```python
# 2️⃣ Kuchendiagramm: Anteil der Passagierklassen

# Figur erstellen
plt.figure(figsize=(6, 6))

# Kuchendiagramm zeichnen
plt.pie(
    class_counts,
    labels=["3. Klasse", "1. Klasse", "2. Klasse"],
    autopct="%1.1f%%",
    colors=["lightblue", "gold", "lightgreen"],
    startangle=90,
)

# Titel setzen
plt.title("Verteilung der Passagierklassen")

# Diagramm anzeigen
plt.show()
```

### **3️⃣ Histogramm: Altersverteilung der Passagiere**
👉 Erstelle ein **Histogramm**, das zeigt, wie alt die Passagiere waren.

**Hinweis:**  
- Nutze `plt.hist()`, um die Verteilung der Spalte **`Age`** zu visualisieren.
- Wähle z. B. **20 Bins**, um eine sinnvolle Gruppierung zu erhalten.

💡 **Erwarteter Plot:**
- **X-Achse:** Alter der Passagiere
- **Y-Achse:** Anzahl der Passagiere in jedem Altersbereich
- **Tipp:** Setze eine **Transparenz (`alpha=0.7`)** für eine schönere Darstellung.

```python
# 3️⃣ Histogramm: Altersverteilung der Passagiere

# Figur erstellen
plt.figure(figsize=(8, 5))

# Histogramm für die Altersverteilung
plt.hist(titanic["age"].dropna(), bins=20, color="purple", alpha=0.7, edgecolor="black")

# Achsenbeschriftungen setzen
plt.xlabel("Alter")
plt.ylabel("Anzahl der Passagiere")
plt.title("Altersverteilung der Titanic-Passagiere")

# Diagramm anzeigen
plt.show()
```

---

### **4️⃣ Streudiagramm: Alter vs. Fahrpreis**
👉 Erstelle ein **Streudiagramm**, das zeigt, ob es eine Verbindung zwischen **Alter (`Age`)** und **Fahrpreis (`Fare`)** gibt.

**Hinweis:**  
- Verwende `plt.scatter()`, um **Alter (X)** gegen **Ticketpreis (Y)** zu plotten.
- Nutze `alpha=0.5`, um überlappende Punkte besser sichtbar zu machen.

💡 **Erwarteter Plot:**
- **X-Achse:** Alter
- **Y-Achse:** Ticketpreis
- **Tipp:** Verwende eine **andere Farbe für Überlebende (`Survived == 1`)**, um Muster zu erkennen.

```python
# 4️⃣ Streudiagramm: Alter vs. Ticketpreis mit Farbe für Überlebende

# Figur erstellen
plt.figure(figsize=(8, 5))

# Scatter-Plot mit Farbkennzeichnung für Überlebende (grün) und Nicht-Überlebende (rot)
plt.scatter(
    titanic[titanic["survived"] == 1]["age"],
    titanic[titanic["survived"] == 1]["fare"],
    alpha=0.5,
    color="green",
    label="Überlebt",
)
plt.scatter(
    titanic[titanic["survived"] == 0]["age"],
    titanic[titanic["survived"] == 0]["fare"],
    alpha=0.5,
    color="red",
    label="Nicht überlebt",
)

# Achsenbeschriftungen setzen
plt.xlabel("Alter")
plt.ylabel("Ticketpreis (€)")
plt.title("Alter vs. Ticketpreis auf der Titanic")

# Legende hinzufügen
plt.legend()

# Diagramm anzeigen
plt.show()

# Alternativ:
# plt.scatter(titanic["age"], titanic["fare"], c=titanic["survived"], cmap="coolwarm", alpha=0.5)
```

### **5️⃣ Gruppiertes Balkendiagramm: Überlebensrate pro Klasse**
👉 Erstelle ein **gruppiertes Balkendiagramm**, das zeigt, wie viele Passagiere in jeder **Klasse (`Pclass`)** überlebt haben oder nicht.

**Hinweis:**  
- Verwende `plt.bar()` für zwei Balken pro Klasse: **Überlebende vs. Nicht-Überlebende**.
- Nutze `legend()`, um die Farben zu kennzeichnen.

💡 **Erwarteter Plot:**
- **X-Achse:** 1. Klasse, 2. Klasse, 3. Klasse
- **Y-Achse:** Anzahl der Passagiere
- **Farben:** **Blau = Überlebt, Rot = Nicht-Überlebt**

```python
# 5️⃣ Gruppiertes Balkendiagramm: Überlebensrate pro Klasse

# Anzahl der Überlebenden und Nicht-Überlebenden pro Klasse berechnen
survival_by_class = titanic.groupby(["pclass", "survived"]).size().unstack()
# 👉 `groupby(["pclass", "survived"])` → Gruppiert die Daten nach Passagierklasse (`pclass`) und Überlebensstatus (`survived`).
# 👉 `.size()` → Zählt die Anzahl der Passagiere für jede Kombination von `pclass` und `survived`.
# 👉 `.unstack()` → Wandelt das Gruppen-Ergebnis in eine Tabelle um, sodass `survived=0` und `survived=1` als Spalten erscheinen.

# Figur erstellen
fig, ax = plt.subplots(figsize=(8, 5))
# 👉 `plt.subplots(figsize=(8, 5))` → Erstellt eine neue Figur mit 8x5 Zoll Größe.
# 👉 `ax` ist das Achsenobjekt, auf dem das Diagramm gezeichnet wird.

# Balkendiagramm für Überlebende (survived=1) und Nicht-Überlebende (survived=0)
survival_by_class.plot(kind="bar", stacked=False, color=["red", "green"], ax=ax)
# 👉 `.plot(kind="bar", stacked=False, color=["red", "green"], ax=ax)`
#     - `kind="bar"` → Erstellt ein Balkendiagramm.
#     - `stacked=False` → Die Balken für `survived=0` (Nicht überlebt) und `survived=1` (Überlebt) werden **nebeneinander** dargestellt.
#     - `color=["red", "green"]` → Weist Farben zu: **Rot = Nicht überlebt**, **Grün = Überlebt**.
#     - `ax=ax` → Zeichnet das Diagramm auf die vorher erstellte Achse `ax`.

ax.tick_params(axis="x", rotation=0)  # X-Achse wird sonst um 90 Grad gedreht

# Achsenbeschriftungen setzen
plt.xlabel("Passagierklasse")  # X-Achse beschriften
plt.ylabel("Anzahl der Passagiere")  # Y-Achse beschriften
plt.title("Überlebensrate pro Klasse")  # Diagrammtitel setzen

# Legende hinzufügen
plt.legend(["Nicht überlebt", "Überlebt"])

# Diagramm anzeigen
plt.show()
```
