## **Stichprobe ziehen – Reduzierung der Datenmenge**

Der `diamonds`-Datensatz ist sehr groß, was ihn ideal für statistische Analysen wie Regression und Machine Learning macht. Mehr Datenpunkte ermöglichen oft genauere Modelle.  

Allerdings kann eine große Datenmenge bei Visualisierungen problematisch sein. Manche Plots werden durch die hohe Anzahl an Punkten unübersichtlich, sodass keine aussagekräftigen Muster erkennbar sind. Ein Beispiel ist der folgende [`catplot`](https://seaborn.pydata.org/generated/seaborn.catplot.html), der die Preisverteilung für verschiedene Diamant-Schliffe (`cut`) darstellt:



```python
sns.catplot(data=diamonds, x="cut", y="price")
```


Wie du sehen kannst, sind durch die riesige Datenmenge keine sinnvollen Zusammenhänge erkennbar. Um dieses Problem zu umgehen, arbeiten wir ab jetzt mit einer zufälligen Stichprobe der Daten.

---

### **Aufgaben:**
1. Ziehe mit der Methode [`sample`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sample.html) eine zufällige Stichprobe von **500 Datenpunkten** aus dem ursprünglichen `diamonds`-DataFrame. Speichere diese in einer neuen Variablen.
2. Gib die **Dimensionen** des neuen DataFrames auf der Konsole aus.
3. Zeige die **ersten 10 Zeilen** der Stichprobe an.
4. Lass dir die **Datentypen der Spalten** ausgeben.
5. Erstelle den **`catplot` für die Stichprobe** (wie oben), um zu prüfen, ob die Daten nun besser lesbar sind.


```python
import matplotlib.pyplot as plt  # Matplotlib zum Anzeigen von Plots

# 2. Ziehe eine zufällige Stichprobe von 500 Datenpunkten
#    - sample(n=500) wählt zufällig 500 Zeilen aus dem DataFrame aus
#    - random_state=42 sorgt dafür, dass die Stichprobe bei jeder Ausführung gleich bleibt (Reproduzierbarkeit)
sample_diamonds = diamonds.sample(n=500, random_state=42)

# 3. Gib die Dimensionen des neuen DataFrames aus (Anzahl der Zeilen und Spalten)
print(f"Größe der Stichprobe (Zeilen, Spalten): {sample_diamonds.shape}")

# 4. Zeige die ersten 10 Zeilen der Stichprobe an
print("\nErste 10 Zeilen der Stichprobe:")
print(sample_diamonds.head(10))

# 5. Lass dir die Datentypen der Spalten ausgeben, um die Struktur des DataFrames zu analysieren
print("\nDatentypen der Spalten:")
print(sample_diamonds.dtypes)

# 6. Erstelle den catplot für die Stichprobe, um die Preisverteilung nach Diamant-Schliff zu visualisieren
#    - 'kind="strip"' zeigt die einzelnen Datenpunkte
#    - 'jitter=True' sorgt für eine leichte Streuung der Punkte, um Überlappungen zu reduzieren
sns.catplot(data=sample_diamonds, x="cut", y="price", kind="strip", jitter=True)

# 7. Zeige den Plot an
plt.show()
```

### **Aufgabe: Countplot des Features `cut`**
Erstelle ein **[Countplot](https://seaborn.pydata.org/generated/seaborn.countplot.html)** für das Feature **`cut`**.

2. **Welche Gruppe kommt am häufigsten vor?**

```python
# 1. Erstelle ein Countplot für das kategorische Feature 'cut'
#    - kind="count" erstellt ein Balkendiagramm der Kategorien
#    - x="cut" gibt die x-Achse vor (die unterschiedlichen Kategorien)
#    - Der Plot zeigt, wie häufig jede Kategorie in der Spalte 'cut' vorkommt
sns.catplot(data=diamonds, x="cut", kind="count")

# 2. Zeige den Plot an
plt.show()
```

```python
# 3. Bestimme die Anzahl der Gruppen im Feature 'cut'
unique_cuts = diamonds["cut"].nunique()  # Zählt die eindeutigen Kategorien
print(f"Anzahl der Gruppen im Feature 'cut': {unique_cuts}")
```


## **Aufgabe 3 – Verteilungsdiagramme**  

### **Ziel:**  
Untersuche die **Verteilung der Karatzahl (`carat`)** mithilfe eines **KDE-Plots**.  
Ein **KDE-Plot (Kernel Density Estimation)** zeigt die Wahrscheinlichkeitsdichte der Werte und gibt eine glatte Schätzung der Verteilung an.  

> 🔹 **INFO:** In dieser Aufgabe nutzen wir [`seaborn.kdeplot`](https://seaborn.pydata.org/generated/seaborn.kdeplot.html), um die Verteilung von `carat` zu visualisieren.  
> Der Parameter `hue="cut"` sorgt dafür, dass die Werte für die verschiedenen `cut`-Kategorien farblich unterschieden werden.  
>  
> Zusätzlich testen wir den Parameter `multiple` mit den Werten:  
> - `"layer"` (überlagerte Kurven)  
> - `"stack"` (gestapelte Verteilungen)  
> - `"fill"` (normierte Flächen)  

---

### **Aufgabe:**  
1. Erstelle ein **KDE-Plot** für `carat`, gruppiert nach `cut` (`hue="cut"`).  
2. Teste verschiedene Darstellungen für den Parameter `multiple` (`"layer"`, `"stack"`, `"fill"`).  
3. Beantworte die Frage:  
   - **Bei wie viel Karat haben die meisten Diamanten ihre Werte?**  


```python
# 1. Erstelle KDE-Plots für die Verteilung von 'carat', getrennt nach 'cut'

# Variante 1: multiple="layer" (überlagerte Kurven)
plt.figure(figsize=(8, 5))  # Optional: Größe des Plots anpassen
sns.kdeplot(data=diamonds, x="carat", hue="cut", multiple="layer", fill=False)
plt.title("KDE-Plot von Carat (multiple='layer')")
plt.show()

# Variante 2: multiple="stack" (gestapelte Verteilungen)
plt.figure(figsize=(8, 5))
sns.kdeplot(data=diamonds, x="carat", hue="cut", multiple="stack", fill=True)
plt.title("KDE-Plot von Carat (multiple='stack')")
plt.show()

# Variante 3: multiple="fill" (normierte Flächen)
plt.figure(figsize=(8, 5))
sns.kdeplot(data=diamonds, x="carat", hue="cut", multiple="fill", fill=True)
plt.title("KDE-Plot von Carat (multiple='fill')")
plt.show()

# 2. Bestimme den häufigsten Karatwert (Peak der Verteilung)
#    - Verwende value_counts(), um die häufigsten Werte in 'carat' zu finden
most_common_carat = diamonds["carat"].value_counts().idxmax()
print(f"Der häufigste Karatwert ist: {most_common_carat}")
```


## **Scatterplot von `price` und `carat` mit `relplot` (vertikale Anordnung)**  

### **Ziel:**  
Untersuche die **Beziehung zwischen `price` (Preis) und `carat` (Karat)** für verschiedene **Diamantschliffe (`cut`)** mithilfe eines **Scatterplots (`relplot`)**, wobei jede `cut`-Kategorie **untereinander** angeordnet wird.  

> 🔹 **INFO:**  
> - [seaborn.relplot](https://seaborn.pydata.org/generated/seaborn.relplot.html) erlaubt es, **mehrere Scatterplots** für verschiedene Gruppen (`cut`) zu erstellen.  
> - Mit `col="cut"` würden die Scatterplots nebeneinander erscheinen, aber mit `row="cut"` werden sie **untereinander** angezeigt.  
> - Durch die vertikale Anordnung lassen sich Unterschiede zwischen den `cut`-Kategorien leichter erkennen.  

---

### **Aufgabe:**  
1. Erstelle ein **Scatterplot (`relplot`)** mit:  
   - `x="price"` (x-Achse)  
   - `y="carat"` (y-Achse)  
   - `row="cut"` (separater Scatterplot für jede `cut`-Kategorie, **untereinander**)  
2. Nutze den Parameter `alpha=0.5`, um Überlappungen der Punkte zu reduzieren.  
3. Stelle die Diagrammgröße mit `height=3.5, aspect=2` so ein, dass sie gut lesbar sind.  
4. Beantworte die Frage:  
   - **Welche Unterschiede lassen sich zwischen den `cut`-Kategorien in Bezug auf Preis und Karat feststellen?**  

---




```python
# 1. Erstelle einen Scatterplot mit relplot
#    - x="price": Preis auf der x-Achse
#    - y="carat": Karatzahl auf der y-Achse
#    - row="cut": Separate Scatterplots für jede 'cut'-Kategorie (untereinander)
#    - alpha=0.5: Setzt die Transparenz der Punkte, um Überlappungen zu reduzieren
#    - height=3.5, aspect=2: Anpassung der Plotgröße für bessere Lesbarkeit
sns.relplot(
    data=diamonds, x="price", y="carat", row="cut", kind="scatter", alpha=0.5, height=3.5, aspect=2
)

# 2. Titel für das Gesamtbild hinzufügen
plt.suptitle("Scatterplot von Price und Carat nach Diamant-Schliff", y=1.02)

# 3. Zeige den Plot an
plt.show()
```

#### **Unterschiede zwischen den `cut`-Kategorien in Bezug auf `price` (Preis) und `carat` (Karat):**  

Nach der Analyse des **Scatterplots (`relplot` mit `row="cut"`)** lassen sich folgende Unterschiede zwischen den **Diamantschliffen (`cut`)** erkennen: 
##### **Zusammenfassung der Trends:**  
| `cut`       | Typische Karatzahl | Preisspanne | Preis pro Karat |
|-------------|------------------|------------|----------------|

| **Premium** | 0.5 - 2 Karat     | $500 - $15k+ | Hoch |
| **Ideal**   | 0.3 - 1.5 Karat   | $500 - $15k+ | Sehr hoch |

---

##### **Fazit:**  

- **Je besser der Schliff (`cut`), desto teurer sind die Diamanten für dieselbe Karatzahl.**  
- **Große Karatwerte (>2 Karat) sind hauptsächlich bei "Fair" und "Good" zu finden.**  
- **"Ideal" und "Premium" haben oft kleinere Karatzahlen, aber sehr hohe Preise.**  
- **"Fair" ist die beste Wahl für große Steine mit moderatem Preis, während "Ideal" die beste Brillanz bietet.**  


## **Aufgabe 5 – Regression: Zusammenhang zwischen `carat` und `price`**  

### **Ziel:**  
Untersuche die **Zusammenhang zwischen `carat` (Karat) und `price` (Preis)** mithilfe einer **Regressionsanalyse**.  

> 🔹 **INFO:**  
> - Die Funktion [`seaborn.regplot`](https://seaborn.pydata.org/generated/seaborn.regplot.html) erstellt eine **lineare Regressionskurve** und zeigt die zugrunde liegende Korrelation zwischen zwei numerischen Variablen.  
> - Über den Parameter `scatter` kann gesteuert werden, ob die Datenpunkte zusätzlich angezeigt werden sollen.  

---

### **Aufgabe:**  
1. Erstelle einen **Regressionsplot (`regplot`)** mit:  
   - `x="carat"` (x-Achse)  
   - `y="price"` (y-Achse)  
   - `scatter=True`, um die Regressionsgerade mit den Originaldatenpunkten anzuzeigen.  
2. Analysiere anhand des Plots:  
   - **Welcher Zusammenhang besteht zwischen `carat` und `price`?**  
   - **Sind die Features stark korreliert?**  

---

```python
# 1. Erstelle einen Regressionsplot für 'carat' (x) und 'price' (y)
#    - scatter=True: Zeigt sowohl die Datenpunkte als auch die Regressionslinie
#    - line_kws={"color": "red"}: Setzt die Farbe der Regressionsgeraden auf rot
plt.figure(figsize=(8, 5))
sns.regplot(data=diamonds, x="carat", y="price", scatter=True, line_kws={"color": "red"})

# 2. Titel und Achsenbeschriftungen hinzufügen
plt.title("Regressionsanalyse: Zusammenhang zwischen Carat und Price")
plt.xlabel("Carat")
plt.ylabel("Preis (USD)")
plt.show()

# 3. Korrelation zwischen 'carat' und 'price' berechnen
correlation = diamonds["carat"].corr(diamonds["price"])
print(f"Die Korrelation zwischen Carat und Price beträgt: {correlation:.2f}")

# 4. Interpretation des Plots und der Korrelation
if correlation > 0.7:
    print(
        "Es gibt eine starke positive Korrelation: Größere Karatwerte führen in der Regel zu höheren Preisen."
    )
elif correlation > 0.4:
    print("Es gibt eine moderate positive Korrelation zwischen Carat und Price.")
else:
    print("Die Korrelation zwischen Carat und Price ist schwach oder nicht vorhanden.")
```
