# Logische Operatoren 

Logische Operatoren haben in Python die Aufgabe Vergleiche bzw. Wahrheitswerte zu verknüpfen. 

Es gibt 3 logische Operatoren in Python:

\begin{array}{c|c|}
   Operator & Anwendung & Funktion \\
   \hline
   and & \text{a and b} & \text{True, wenn a und b wahr sind} \\
   \hline 
   or & \text{a or b} & \text{True, wenn mindestens a oder b wahr ist} \\
   \hline 
   not & \text{not a} & \text{Vertauscht den Wahrheitswert von a} \\ 
\end{array} 

\
\
 Wir werden uns die Operatoren in praktischen Beispielen anschauen.
*** 
 ## Operator `and`
 

 ### Beispiele

```python
print("Wahr und Wahr   --->", True and True)
print("Wahr und Falsch --->", True and False)
```

Wir können einzelne Aussagen miteinander verknüpfen. Im Folgenden ist sowohl die linke als auch die rechte Seite wahr, daher ist der gesamte Ausdruck wahr:

```python
3 == 3 and 1 > 0
```

Wir können die Aussagen auch komplizierter machen:

```python
2 < 0 and 1 + 1 == 2
```

### Wahrheitstabelle für `and`

Wahrheitstabellen zeigen für alle Kombinationen Wahrheitswerten und `and` das Resultat. 

Wir programmieren alle möglichen Kombinationen:


```python
f = False
t = True
print(t, "and", t, "--->", t and t)
print(t, "and", f, "--->", t and f)
print(f, "and", t, "--->", f and t)
print(f, "and", f, "--->", f and f)
```

Oder, wenn wir es tabellarisch anzeigen wollen:
\begin{array}{c|c||c}
   A & B & \text{A and B} \\
   \hline
   \hline
   t & t & t \\ 
   \hline
   t & f & f \\
   \hline 
   f & t & f \\
   \hline 
   f & f & f \\ 
   \end{array} 
   
Hier entspricht `t` dem Wert `True` und `f` dem Wert `False`. 

Wir sehen, dass eine Verknüpfung zweier Wahrheitswerte $A$ und $B$ mit `and` nur dann `True` ergibt, wenn sowohl $A$ als auch $B$ `True` sind.

***
## Operator `or`

### Beispiele
 
 Im folgenden einige Beispiel für den `or`-Operator. Mache dir klar, wieso die jeweilige Ausgabe erfolgt.

```python
print(True or False)
print(2 > 1 or 3 == 5)
print(5 % 2 == 0 or 3 < 2)
```

### Wahrheitstabelle für `or`

Auch für `or` können wir uns sämtliche Möglichkeiten in einer Wahrheitstabelle ausgeben lassen.

```python
f = False
t = True
print(t, "or", t, "--->", t or t)
print(t, "or", f, "--->", t or f)
print(f, "or", t, "--->", f or t)
print(f, "or", f, "--->", f or f)
```

bzw. tabellarisch:
\begin{array}{c|c||c}
   A & B & \text{A or B} \\
   \hline
   \hline
   t & t & t \\ 
   \hline
   t & f & t \\
   \hline 
   f & t & t \\
   \hline 
   f & f & f \\ 
   \end{array} 
   
\
Wir sehen, die `oder`-Verknüpfung von $A$ und $B$ ist genau dann `False`, wenn sowohl $A$ als auch $B$ `False` sind. 

***
## Operator `not`

Der `not`-Operator ist im Gegensatz zu `and` und `or` nur auf einen Wahrheitswert anwendbar. 

Der Wahrheitswert wird von `not` ins "Gegenteil" umgewandelt.

### Beispiele

```python
print(not True)
print(not False)
print(not 3==3)
```

### Wahrheitstabelle für `not`

Wir können uns wieder die Wahrheitstabelle anzeigen lassen:

```python
f = False
t = True
print("not", t, "--->", not t)
print("not", f, "--->", not f)
```

\begin{array}{c||c}
   A &  \text{not A} \\
   \hline
   \hline
   t & f \\ 
   \hline
   f & t \\
   \hline 
   \end{array} 
   
Der **not**-Operator verändert jeden Wahrheitswert in sein Gegenteil. 

Aus `True` wird `False` und umgekehrt.
